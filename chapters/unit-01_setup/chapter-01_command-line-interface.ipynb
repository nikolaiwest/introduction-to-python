## **Chapter 1.1:** Understanding the Command Line Interface

### 🔍 What is the CLI?

The **Command Line Interface (CLI)**, also known as the terminal, console, or command prompt, is a text-based interface for interacting with a computer's operating system. Unlike *graphical user interfaces* (GUIs) that use visual elements like buttons and menus, the CLI relies on typed commands to perform actions.

The CLI presents a prompt where you can type commands, which are then executed by the system. This direct interaction offers precise control over your computer's operations, making it an essential tool for developers, system administrators, and power users.

<img src="https://www.freecodecamp.org/news/content/images/2022/08/ss5-1.png" height="400">

### 🧩 The Importance of CLI

While graphical interfaces dominate everyday computer use, the command line remains vital for programming and development work for several compelling reasons:

* **Efficiency:** Once you're familiar with CLI commands, many tasks can be performed much faster than with a GUI.
* **Automation:** CLI commands can be easily scripted, enabling the automation of repetitive tasks.
* **Remote Access:** CLIs are essential for working with remote servers or cloud-based systems where graphical interfaces aren't available.
* **Resource Management:** CLIs typically use fewer system resources than GUIs, which can be crucial when working with large datasets or complex operations.
* **Precision:** CLI commands offer exact control over operations, allowing for specific options and parameters that might not be available in GUI equivalents.

As you progress in your Python journey, you'll find the CLI becomes an indispensable part of your development toolkit, serving as the foundation for many advanced tools and workflows.

### 📘 Basic CLI Commands

The following commands are presented using **Windows Command Prompt** syntax, as it's a common operating system for personal computers. Users of other systems (macOS or Linux) may need to adjust certain commands, though many are similar across platforms.

#### Print Working Directory: `cd`

This command displays your current directory path, helping you understand where you are in the file system. In Windows, typing `cd` without arguments serves this purpose, unlike Unix-based systems that use `pwd`.

```cmd
C:\Users\user> cd
```

Output:
```cmd
C:\Users\user
```

#### List Directory Contents: `dir`

The `dir` command shows files and folders in the current directory. This essential command helps you view directory contents, including file names, sizes, and modification dates.

```cmd
C:\Users\user\projects\introduction-to-python> dir
```

Output:
```cmd
Volume in drive C has no label.
Volume Serial Number is 1234-5678

Directory of C:\Users\user\projects\introduction-to-python

05/01/2024  09:30 AM    <DIR>          .
05/01/2024  09:30 AM    <DIR>          ..
05/01/2024  09:30 AM               256 main.py
05/01/2024  09:30 AM               128 script.py
05/01/2024  09:30 AM    <DIR>          unit-01_setup
05/01/2024  09:30 AM    <DIR>          unit-02_syntax

        2 File(s)            384 bytes
        2 Dir(s)  100,000,000,000 bytes free
```

#### Change Directory: `cd <any/path>`

The `cd` command is used to navigate between directories in the file system. This fundamental command allows you to move through the directory structure to access different locations.

```cmd
C:\Users\user\projects\introduction-to-python> cd unit-01_setup
```

Output:
```cmd
C:\Users\user\projects\introduction-to-python\unit-01_setup>
```

The `cd` command can be used in several ways to navigate efficiently:

* `cd ..`: Moves up one level to the parent directory
* `cd .`: Refers to the current directory (doesn't change location)
* `cd \`: Moves to the root directory of the current drive
* `cd /d D:`: Changes both the drive and directory at once (to drive D: in this example)

#### Make Directory: `mkdir`

The `mkdir` command creates a new directory, helping you organize your files into logical structures.

```cmd
C:\Users\user\projects\introduction-to-python> mkdir unit-00_test
```

Output after running `dir`:
```cmd
C:\Users\user\projects\introduction-to-python> dir
main.py  unit-00_test  unit-01_setup  unit-02_syntax ...
```

#### Create Empty File: `echo.>`

This command creates a new, empty file with the specified name. It's useful for quickly creating placeholder files or starting new scripts.

```cmd
C:\Users\user\projects\introduction-to-python> echo.> new_script.py
```

Output after running `dir`:
```cmd
C:\Users\user\projects\introduction-to-python> dir
main.py  new_script.py  unit-00_test  unit-01_setup  unit-02_syntax ...
```

#### ⚠️ Common Mistakes with Basic Commands

* **Using incorrect path separators:** Windows uses backslashes (`\`), while Unix-based systems use forward slashes (`/`).
* **Forgetting to quote paths with spaces:** If a path contains spaces, it should be enclosed in quotes.
* **Trying to `cd` into a file:** The `cd` command only works with directories, not files.

### 📘 More CLI Commands

As you become more comfortable with the basic commands, these additional commands will expand your CLI capabilities:

#### Copy File: `copy`

This command copies a file from one location to another, useful for creating backups or duplicating files.

```cmd
C:\Users\user\projects\introduction-to-python> copy new_script.py backup_script.py
```

Output:
```cmd
    1 file(s) copied.
```

#### Move File: `move`

The `move` command relocates a file from one location to another and can also be used to rename files.

```cmd
C:\Users\user\projects\introduction-to-python> move new_script.py backups\
```

Output:
```cmd
    1 file(s) moved.
```

#### Rename File: `ren`

This command allows you to change the name of a file without moving it to a different location. The syntax is simple and straightforward.

```cmd
C:\Users\user\projects\introduction-to-python> ren old_filename.py new_filename.py
```

You won't see any output message when the rename is successful. To verify the file was renamed, you can use the `dir` command to list the directory contents.
This command can also be used to rename multiple files at once using wildcards:

```cmd
C:\Users\user\projects\introduction-to-python> ren *.txt *.bak
```

This would rename all files with `.txt` extension to have `.bak` extension instead.

#### Delete File: `del`

This command removes a file from the file system. Use it carefully as it permanently deletes files without sending them to the recycle bin.

```cmd
C:\Users\user\projects\introduction-to-python> del backup_script.py
```

#### Remove Directory: `rmdir`

The `rmdir` command deletes a directory. The basic version can only remove empty directories.

```cmd
C:\Users\user\projects\introduction-to-python> rmdir empty_folder
```

For removing directories with contents, use additional options:

* `/s`: Removes all subdirectories and files in the directory tree
* `/q`: Runs in quiet mode, not asking for confirmation

```cmd
C:\Users\user\projects\introduction-to-python> rmdir /s /q not_empty_folder
```

#### Clear Screen: `cls`

This command clears the command prompt screen, giving you a clean workspace.

```cmd
C:\Users\user\projects\introduction-to-python> cls
```

#### ⚠️ Common Mistakes with Advanced Commands

* **Using `rmdir /s /q` without caution:** This command deletes all contents without confirmation and cannot be undone.
* **Confusing `copy` and `move`:** Remember that `copy` leaves the original file intact, while `move` removes it from the source location.

## **Chapter 1.2:** Using CLI Interactive Mode for Python

### 🔍 Install Python

Before you can use Python's interactive mode, you need to have Python installed on your system. Follow these steps to install Python:

1. Go to the official Python website (https://www.python.org/downloads/) and download the Python installer for your operating system.

2. Open the downloaded installer.

3. Important: Select the 'Add Python to PATH' option at the bottom of the installation window. This ensures Python is automatically added to your system's environment variables, making it accessible from the command line.

4. Click on 'Install Now' to proceed with the default installation.

5. After installation, verify that Python is correctly installed by opening a command prompt and running:
   ```cmd
   python --version
   ```
   This should display the installed Python version.

<img src="https://www.pythontutorial.net/wp-content/uploads/2020/09/Install-Python-Windows-Step-1.png" height="300">

### 📘 Start the Interactive Mode for Python

Python's interactive mode provides an immediate, interactive environment for testing code and executing Python commands. To access it:

1. Open your command line interface (Command Prompt, Terminal, etc.).

2. Type `python` (or `python3` on some systems) and press Enter. This launches the Python interpreter in interactive mode.

```bash
python
```

You'll see Python's version information and the interactive prompt (`>>>`) appears, indicating you're now in Python's interactive shell:

```bash
Python 3.11.4 (default, Jul 28 2023, 12:59:40) 
[GCC 9.3.0] on linux
Type "help", "copyright", "credits" or "license" for more information.
>>>
```

You can now type Python code directly into this interface. For example, try your first Python command:

```python
>>> print("Hello World!")
```

This will immediately execute and display:
```
Hello World!
```

The interactive mode is perfect for quick experimentation, testing small code snippets, or learning Python concepts without creating full script files.

### 📘 Using Python as a Calculator in CLI Interactive Mode

One of the simplest ways to start using Python's interactive mode is as a calculator. Python supports all standard arithmetic operations and can handle complex calculations.

#### Basic Arithmetic Operations

Start by trying these simple calculations:

```python
>>> 5+3
8
>>> 10-8
2
>>> 5*3
15
>>> 6/2
3.0
```

Notice that division (`/`) returns a floating-point number, even when the result is a whole number.

#### Advanced Arithmetic Operations

Python supports more complex operations like:

```python
>>> (3+5)*2  # Parentheses for order of operations
16
>>> 2**3     # Exponentiation (2 raised to power 3)
8
>>> 10//3    # Integer division (returns whole number)
3
>>> 10%3     # Modulo (returns remainder)
1
```

#### Using Variables for Calculations

You can assign values to variables and use them in calculations:

```python
>>> Force = 10
>>> Area = 5.2
>>> pressure = Force/Area
>>> pressure  # Display the value
1.9230769230769231
```

This capability becomes powerful when working with more complex formulas and calculations, as you can store intermediate results and reuse them.

#### ⚠️ Common Mistakes with Calculations

* **Forgetting operator precedence:** Python follows the standard order of operations (PEMDAS).
* **Confusing `/` and `//`:** Regular division (`/`) always returns a float, while floor division (`//`) returns an integer.
* **Using commas in large numbers:** Python doesn't recognize commas as number separators (use `1000000`, not `1,000,000`).

### 📘 Writing Programs with Multiple Lines in Interactive Mode

Python's interactive mode also supports multi-line programming constructs like loops, functions, and conditional statements. This allows you to test more complex code directly in the shell.

When you start a code block (such as a loop, function, or conditional statement), the prompt changes from `>>>` to `...` to indicate that you're inside a continuing block of code.

For example, to print "Hello World" three times using a for loop:

```python
>>> for i in range(3):
...     print("Hello World!")
...
Hello World!
Hello World!
Hello World!
```

Key points about multi-line programming in interactive mode:

* Use the Tab key or spaces to indent code blocks properly
* The block continues as long as you maintain the indentation
* Press Enter twice (on an empty line at the same indentation level) to end the block and execute it
* You can define functions, create classes, and implement any Python construct

This capability makes the interactive mode useful not just for simple calculations, but also for testing more complex logic and algorithms.

### 📘 Exiting the Interactive Mode

When you're finished using Python's interactive mode, there are several ways to exit and return to your regular command prompt:

**1. Using the exit() function:**
```python
>>> exit()
```

**2. Using the Ctrl + D shortcut (Unix/Linux/Mac):**
Press `Ctrl + D` on your keyboard. This sends an EOF (End of File) signal to Python. On some systems, you might need to press it twice.

**3. Using the Ctrl + Z followed by Enter (Windows):**
Press `Ctrl + Z` and then press Enter. This sends an EOF signal in Windows environments.

All these methods safely terminate the Python interpreter and return you to your system's command line interface.

## **Chapter 1.3:** Creating and Executing Python Scripts

While Python's interactive mode is excellent for experimentation and quick tests, most Python development involves creating and running script files. This allows you to save your code, build more complex programs, and execute them whenever needed.

### 🔍 Create a Python File with `echo`, `touch` and `nano`

There are several ways to create Python script files using the command line. The method you choose may depend on your operating system and personal preference.

#### Using `echo`

On Windows, you can use the `echo` command to create and populate Python files:

**Creating an empty file:**
```bash
echo.> filename.py
```
This creates an empty Python file named `filename.py` in the current directory. You can then open it with a text editor to add code.

**Creating a file with content:**
```bash
echo print("First row") > filename.py
echo print("Second row") >> filename.py
echo print("Third row") >> filename.py
```

Note the difference between `>` and `>>`:
- `>` creates a new file or overwrites an existing file
- `>>` appends content to an existing file

#### Using `touch` (Unix-based systems)

On macOS, Linux, or Windows with Git Bash installed, the `touch` command creates an empty file:

```bash
cd repos/python_for_engineers/programs
touch testfile.py
```

The key difference between `touch` and `echo.>` is that `touch` won't overwrite an existing file's content; it only updates the timestamp if the file already exists.

#### Using `nano` (Unix-based systems)

For Unix-based systems or Windows with appropriate tools installed, `nano` is a simple text editor that allows you to create and edit files directly in the terminal:

```bash
nano testfile.py
```

This opens the nano text editor where you can write your code. To save and exit:
1. Press `Ctrl + X`
2. Press `Y` to confirm saving
3. Press `Enter` to confirm the filename

Nano is particularly useful when working on remote servers where graphical editors may not be available.

#### ⚠️ Common Mistakes When Creating Files

* **Using the wrong file extension:** Python files should end with `.py`
* **Forgetting to properly indent code:** Python relies on indentation for code blocks
* **Using incorrect line endings:** Different operating systems use different line endings, which can sometimes cause issues

### 📘 Executing Python Scripts

Once you've created a Python script file, you can execute it using the Python interpreter. This process involves telling Python to run the code contained in your file.

#### Running a script

To execute a Python script, navigate to the directory containing the script using the `cd` command, then use the `python` command followed by the filename:

```bash
cd C:/repos/python_for_engineers/programs
python testfile.py
```

The Python interpreter will read the file, compile it to bytecode, and execute it, displaying any output in the terminal.

#### Alternative execution methods

Depending on your system configuration, you might also be able to:

* Run a script by double-clicking it (if file associations are set up)
* Use `./script.py` on Unix-based systems if the script has execution permissions and a proper shebang line
* Use `py script.py` on Windows systems with the Python launcher installed

#### ⚠️ Common Script Execution Issues

* **"Python is not recognized" error:** This indicates Python is not in your system PATH
* **Permission denied:** The script may not have execution permissions (Unix-based systems)
* **Module not found errors:** Python can't locate imported modules (often related to your working directory or PYTHONPATH)
