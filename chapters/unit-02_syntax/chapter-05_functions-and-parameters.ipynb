## **Chapter 5.1:** Function Fundamentals

### 🔍 Understanding Functions

Functions are one of the most **fundamental concepts** in programming. Think of a function as a *reusable recipe*. Just like a recipe takes ingredients (inputs), follows a set of instructions (processing), and produces a dish (output), a function takes **parameters**, processes them, and **returns** a result. Functions help us organize our code into manageable, reusable pieces.

### Why Do We Need Functions?

Imagine you're writing a program that needs to calculate the area of different circles. **Without functions**, you'd need to write the same calculation code every time:

````python
# Without functions - repetitive and hard to maintain
radius1 = 5
area1 = 3.14159 * radius1 * radius1
print(f"Area of circle 1: {area1}")

radius2 = 7
area2 = 3.14159 * radius2 * radius2
print(f"Area of circle 2: {area2}")

# What if we found a bug? We'd need to fix it in multiple places!
````

With functions, we can write the calculation once and reuse it:


````python
def calculate_circle_area(radius):
    """Calculate the area of a circle given its radius."""
    return 3.14159 * radius * radius

# Now we can easily calculate areas for any number of circles
area1 = calculate_circle_area(5)
area2 = calculate_circle_area(7)
````

### 🧩 Important Components

* **Function Definition:** The `def` keyword followed by the function name and parameters.
* **Function Body:** The *indented code block* that contains the operations of the function.
* **Parameters:** Variables listed in the function definition that receive values when the function is called.
* **Return Statement:** Used to send a result back from the function (optional).
* **Docstring:** Documentation string that explains what the function does.
* **Function Call:** How you *execute* a function, with arguments corresponding to parameters.

### 📘 Main Content

#### Defining and Calling Functions

In Python, you define a **function** using the `def` keyword, followed by the function name and a pair of parentheses `()`. The code block within every function starts with a colon `:` and is indented.

Here's a simple function that prints a greeting:

````python
def say_hello():
    print("Hello, World!")

# Calling the function
say_hello()  # Output: Hello, World!
````

Even in this simple `say_hello` example, we can see two key parts:

1. The function definition (starting with `def`)
2. The function body (the indented code)

Now, let's build up to a more complete function by examining each component in detail:

````python
def greet_user(name, greeting="Hello"):
    """
    Display a personalized greeting to the user.
    
    Args:
        name (str): The user's name
        greeting (str, optional): The greeting to use. Defaults to "Hello"
    
    Returns:
        str: The complete greeting message
    """
    message = f"{greeting}, {name}!"
    return message
````

Let's break down each part of this function:

1. `def` keyword
   * This tells Python you're **creating a new function**
   * *Always required* at the start of a function definition

2. Function name (`greet_user`)
   * Choose **descriptive names** that explain what the function does
   * Follow Python naming conventions (*lowercase with underscores*)
   * Should be a *verb or verb phrase* since functions perform actions

3. Parameters in parentheses
   * The **inputs** your function accepts
   * Multiple parameters are separated by commas
   * Can have **default values** (like `greeting="Hello"`)

4. Colon (`:`)
   * Marks the end of the function header
   * Tells Python that the function body is about to start

5. Docstring in triple quotes `"""`
   * **Explains** what the function does
   * **Documents** parameters and return values
   * Helps other developers **understand** your code

6. Function body
   * The *indented code block* that runs when the function is called
   * Must be indented (usually 4 spaces)
   * Can be multiple lines

7. `return` statement
   * Specifies what **value** the function sends back
   * Function ends when `return` is reached

Let's see this function in action:

````python
# Using the function with all arguments
message1 = greet_user("Alice", "Good morning")
print(message1)  # Output: Good morning, Alice!

# Using the function with default greeting
message2 = greet_user("Bob")
print(message2)  # Output: Hello, Bob!
````

#### Understanding the Flow

When you call a function, Python:

1. Jumps to the function definition
2. Creates space in memory for any parameters
3. Executes the function body
4. Returns to where the function was called
5. Uses the returned value (if any)

````python
# Here's what happens step by step:
def multiply_by_two(number):
    result = number * 2  # Step 3: Multiply the number
    return result       # Step 4: Send back the result

x = 5
doubled = multiply_by_two(x)  # Steps 1-2: Jump to function with x=5
print(doubled)  # Step 5: Print the returned value (10)
````

A few simple things to remember:

* Functions should do one specific task
* Choose clear, descriptive names
* Include helpful documentation
* Return values when needed

### 🔍 Function Parameters and Arguments

Functions can take *parameters*, which are variables that hold values passed to the function when it's called. These values are called **arguments**.

````python
def greet(name):
    print(f"Hello, {name}!")

# Calling the function with an argument
greet("Alice")  # Output: Hello, Alice!
greet("Bob")    # Output: Hello, Bob!
````

### 🔍 Return Statements

Functions can **return** values using the `return` statement. This allows a function to produce a result that can be used elsewhere in your code.

````python
def add(a, b):
    return a + b 

result = add(3, 5)
print(f"The sum is: {result}")  # Output: The sum is: 8
````

### Variable Scope and Lifetime

Variables defined inside a function have a local scope, meaning they can only be accessed within that function. Variables defined outside of any function have a global scope and can be accessed throughout the program.

````python
x = 10  # Global variable

def print_x():
    x = 20  # Local variable
    print(f"Local x: {x}")

print_x()        # Output: Local x: 20
print(f"Global x: {x}")  # Output: Global x: 10
````

In this example, the `x` inside the function is a different variable from the global `x`.

#### ⚠️ Common Mistakes with Function Definitions

* **Forgetting the colon**: The function definition line must end with a colon
* **Incorrect indentation**: All code in the function body must be indented consistently
* **Missing return statement**: If you don't include a return statement, the function returns `None` by default
* **Confusing parameters and arguments**: Parameters are in the function definition, arguments are the values passed when calling

## **Chapter 5.2:** Function Arguments

### 🔍 Understanding Function Arguments

When you create a function, you often want to pass information to it. This information comes in the form of **arguments**. Think of arguments as *inputs* that your function needs to do its job. Just like a coffee machine needs water and coffee beans to make coffee, functions often need data to perform their tasks.

Arguments are the values you pass into the function when you call it. Python offers flexibility in how arguments can be passed to functions, catering to various scenarios.

* **Positional Arguments:** These are arguments that need to be in order based on the function's definition. The first argument fills the first parameter, the second fills the second, and so on.

* **Keyword Arguments:** Allow you to specify arguments by the parameter name, making your function calls more readable and allowing you to skip certain default parameters.

* **Default Parameters:** You can assign default values to parameters, making them optional during a function call.

### 🧩 Important Components

* **Positional Arguments:** Arguments matched to parameters by position (order)
* **Keyword Arguments:** Arguments matched to parameters by name
* **Default Parameters:** Parameters with pre-assigned values if no argument is provided
* **Variable-Length Arguments:** Special parameters that can accept any number of arguments
  * `*args`: For variable positional arguments
  * `**kwargs`: For variable keyword arguments
* **Argument Order:** The correct sequence for different types of arguments in function definitions and calls

### 📘 Main Content

#### Positional Arguments

Positional arguments are the most basic way to pass information to functions. They're called **"positional"** because the order matters - the first value you pass matches the first parameter, the second value matches the second parameter, and so on.

````python
def make_coffee(coffee_type, size):
    """Make a coffee of specified type and size."""
    return f"Making a {size} {coffee_type}"

# The order matters with positional arguments
print(make_coffee("latte", "large"))     # Output: Making a large latte
print(make_coffee("large", "latte"))     # Output: Making a latte large (oops!)
````

In the example above, swapping the order of arguments changes the meaning of our function call. This can lead to errors that might not be immediately obvious.

#### Keyword Arguments

To avoid confusion with argument order, you can use keyword arguments. These explicitly match arguments to parameters by name:

````python
def create_user(username, email, age, is_active=True):
    """
    Create a new user profile.
    
    Args:
        username (str): User's display name
        email (str): User's email address
        age (int): User's age
        is_active (bool, optional): Account status. Defaults to True.
    
    Returns:
        dict: User profile information
    """
    return {
        "username": username,
        "email": email,
        "age": age,
        "is_active": is_active
    }

# Using keyword arguments makes the code more readable
user1 = create_user(
    username="john_doe",
    email="john@example.com",
    age=25
)

# You can mix positional and keyword arguments
# Positional arguments must come first!
user2 = create_user("jane_doe", email="jane@example.com", age=28)

# This would raise an error - positional argument after keyword:
# user3 = create_user(email="bob@example.com", "bob_doe", age=22)  # Error!
````

#### Default Arguments

Sometimes you want a parameter to have a default value that's used when no argument is provided:


````python
def greet(name, greeting="Hello", punctuation="!"):
    """
    Create a greeting message.
    
    Args:
        name (str): Person's name
        greeting (str, optional): Greeting to use. Defaults to "Hello".
        punctuation (str, optional): End punctuation. Defaults to "!".
    
    Returns:
        str: Formatted greeting message
    """
    return f"{greeting}, {name}{punctuation}"

# Using default values
print(greet("Alice"))                 # Output: Hello, Alice!
print(greet("Bob", "Hi"))             # Output: Hi, Bob!
print(greet("Carol", "Hey", "..."))   # Output: Hey, Carol...
````

````python


# Using keyword arguments with defaults
print(greet("David", punctuation="?")) # Output: Hello, David?
````

**Important notes about default parameters:**

* Default values are evaluated **when the function is defined**, not when it's called
* **Never use mutable objects** (like lists or dictionaries) as default values
* Default parameters **must come after** non-default parameters

Here's an example of what **NOT** to do:

````python
# BAD: Using a mutable default value
def add_to_list(item, my_list=[]):
    my_list.append(item)
    return my_list

# This might not work as expected
print(add_to_list("apple"))  # Output: ['apple']
print(add_to_list("banana")) # Output: ['apple', 'banana'] (Surprise!)
````

````python
# GOOD: Use None as default and create list inside function
def add_to_list_correct(item, my_list=None):
    if my_list is None:
        my_list = []
    my_list.append(item)
    return my_list
````

#### Variable-Length Arguments (`*args` and `**kwargs`)

Sometimes you don't know in advance how many arguments a function will receive. Python provides two special syntaxes for these cases:

**Arbitrary Positional Arguments: `*args`**

The `*args` syntax allows a function to accept any number of positional arguments.

````python
def calculate_average(*numbers):
    """
    Calculate the average of any number of values.
    
    Args:
        *numbers: Variable number of numeric values
        
    Returns:
        float: Average of all numbers, or 0 if no numbers provided
    """
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)

# Using *args
print(calculate_average(2, 4, 6))          # Output: 4.0
print(calculate_average(1, 2, 3, 4, 5))    # Output: 3.0
print(calculate_average())                  # Output: 0
````

````python
# You can also combine *args with regular parameters
def print_student_info(name, *scores):
    """Print a student's name and all their scores."""
    print(f"Student: {name}")
    print(f"Scores: {scores}")
    if scores:
        print(f"Average: {sum(scores) / len(scores):.2f}")

print_student_info("Alice", 85, 92, 78)
# Output:
# Student: Alice
# Scores: (85, 92, 78)
# Average: 85.00
````

**Arbitrary Keyword Arguments: `**kwargs`**

The `**kwargs` syntax allows a function to accept any number of keyword arguments:

````python
def create_profile(**info):
    """
    Create a profile from arbitrary keyword arguments.
    
    Args:
        **info: Variable number of keyword arguments
        
    Returns:
        dict: Profile information
    """
    # Add default values for missing fields
    defaults = {
        "occupation": "Unknown",
        "location": "Unknown"
    }
    
    # Update defaults with provided info
    return {**defaults, **info}

# Using **kwargs
profile1 = create_profile(
    name="Alice",
    age=30,
    occupation="Engineer"
)
print(profile1)
# Output: {'occupation': 'Engineer', 'location': 'Unknown', 'name': 'Alice', 'age': 30}
````

````python
# Combining all types of arguments
def process_order(order_id, *items, **details):
    """Process an order with items and additional details."""
    print(f"Order ID: {order_id}")
    print("Items:")
    for item in items:
        print(f"- {item}")
    print("Details:")
    for key, value in details.items():
        print(f"- {key}: {value}")

process_order(
    "12345",
    "Coffee", "Muffin", "Sandwich",
    payment="Credit Card",
    delivery="Express"
)
# Output:
# Order ID: 12345
# Items:
# - Coffee
# - Muffin
# - Sandwich
# Details:
# - payment: Credit Card
# - delivery: Express
````

#### Combining Different Types of Arguments

You can use all these types of arguments in a single function definition. 

The order matters: 
1. **positional arguments** first, 
2. then `*args`, 
3. then **default arguments**, 
4. and finally `**kwargs`.

````python
def example_function(pos1, pos2, *args, default1="default", **kwargs):
    print(f"Positional: {pos1}, {pos2}")
    print(f"Args: {args}")
    print(f"Default: {default1}")
    print(f"Kwargs: {kwargs}")

example_function(1, 2, 3, 4, 5, default1="custom", key1="value1", key2="value2")
````

#### Common Mistakes and Best Practices

When working with function arguments, there are several important pitfalls to avoid and best practices to follow.

Let's explore each one in detail:

**1. Argument Order Matters**

The order of different types of arguments in your function definition is crucial. Python has strict rules about this order:

```python
# Always put arguments in this order:
def function_name(
    regular_args,       # Regular positional arguments
    *args,              # Variable positional arguments
    default_args=value, # Default arguments
    **kwargs            # Variable keyword arguments
):
    pass
```

Why This Matters:

* Python needs to know which arguments match with which parameters
* Putting arguments in the wrong order will cause syntax errors
* The order helps Python correctly handle both positional and keyword arguments

*Examples of what not to do:*

````python
# This will cause a SyntaxError
def bad_function(default_arg=10, positional_arg):  # Error!
    return positional_arg + default_arg
````

````python
# This will also cause an error
def another_bad_function(**kwargs, *args):  # Error!
    pass
````

**2. Mutable Default Arguments**

This is one of the most common and trickiest mistakes in Python. When you use a mutable object (like a list or dictionary) as a default argument, it's created only once when the function is defined, not each time the function is called.

Why This Is Dangerous:

* The same mutable object is reused across all function calls
* Changes to the default object persist between function calls
* This can lead to very confusing bugs that are hard to track down

*Example of the Problem:*

````python
def add_item(item, shopping_list=[]):  # BAD!
    shopping_list.append(item)
    return shopping_list

# First call seems fine
print(add_item("apple"))  # Output: ['apple']

# But watch what happens on the second call
print(add_item("banana")) # Output: ['apple', 'banana']  # Unexpected!

# A third call makes it even worse
print(add_item("orange")) # Output: ['apple', 'banana', 'orange']  # Yikes!
````

*The correct way to approach this:*


````python
def add_item(item, shopping_list=None):  # GOOD!
    # Create a new list if None is provided
    if shopping_list is None:
        shopping_list = []
    shopping_list.append(item)
    return shopping_list

# Now each call starts fresh
print(add_item("apple"))   # Output: ['apple']
print(add_item("banana"))  # Output: ['banana']
print(add_item("orange"))  # Output: ['orange']

# Unless we want to reuse a list
my_list = ["milk"]
add_item("eggs", my_list)
print(my_list)  # Output: ['milk', 'eggs']
````

**3. Clear Parameter Names**

Parameter names should be descriptive and meaningful. This is part of making your code self-documenting and easier to maintain.

Why This Matters:

* Clear names make code more readable
* Good names serve as documentation
* It reduces the chance of using arguments incorrectly
* Makes code maintenance easier

*Examples of Poor Naming:*

````python
# Bad: Unclear parameter names
def calc(x, y, z, f=False):
    if f:
        return x * y * z
    return x + y + z

# Usage is confusing
result = calc(10, 20, 30, f=True)  # What does 'f' mean?
````

*Better approach:*

````python
# Good: Clear, descriptive parameter names
def calculate_volume(length, width, height, use_multiplication=False):
    if use_multiplication:
        return length * width * height
    return length + width + height

# Usage is clear and self-documenting
room_volume = calculate_volume(
    length=10,
    width=20,
    height=30,
    use_multiplication=True
)
````

#### ⚠️ Common Mistakes with Function Arguments

* **Confusing positional and keyword arguments**: Remember that positional arguments must come before keyword arguments
* **Using mutable default arguments**: Never use mutable objects (lists, dictionaries) as default arguments
* **Incorrect order of parameter types**: Always put required parameters first, then *args, then parameters with defaults, and finally **kwargs
* **Forgetting that default values are evaluated only once**: Default values are created when the function is defined, not each time it's called

## **Chapter 5.3:** Advanced Function Concepts

### 🔍 Lambda Functions *(Anonymous Functions)*

Lambda functions are small, **one-line functions** that can be created on the fly. Think of them as quick, disposable tools for simple operations. They're called "*anonymous*" because they don't need a formal name.

#### Why Use Lambda Functions

* When you need a **simple function** for a short period
* In functions that expect other functions as **arguments**
* For quick **data transformations**

Lambda functions use the following syntax:

### 🔍 Recursive Functions

A recursive function is one that **calls itself**. It's like a set of nested Russian dolls - each function call opens up to reveal another call inside.

#### Understanding Recursion

Every recursive function needs:

* A **base case** (when to stop)
* A **recursive case** (how to break down the problem)

Here's a simple example of recursion with a countdown function:

````python
def countdown(n):
    """
    Demonstrate recursion with a countdown.
    
    Args:
        n (int): Number to start counting from
        
    Returns:
        None: Prints countdown sequence
    """
    # Base case
    if n <= 0:
        print("Liftoff!")
        return
    
    # Recursive case
    print(n)
    countdown(n - 1)  # Function calls itself

# Example usage
countdown(5)
# Output:
# 5
# 4
# 3
# 2
# 1
# Liftoff!
````

Let's look at a more practical example - calculating factorial:


````python
def factorial(n):
    """Calculate the factorial of n (n!)."""
    # Base case
    if n <= 1:
        return 1
    
    # Recursive case
    return n * factorial(n - 1)

# Example usage
print(factorial(5))  # Output: 120 (5 × 4 × 3 × 2 × 1)
````

To understand how this works, let's trace through the calls for `factorial(5)`:
1. `factorial(5)` returns `5 * factorial(4)`
2. `factorial(4)` returns `4 * factorial(3)`
3. `factorial(3)` returns `3 * factorial(2)`
4. `factorial(2)` returns `2 * factorial(1)`
5. `factorial(1)` returns `1` (base case)
6. Now we can calculate: `2 * 1 = 2, 3 * 2 = 6, 4 * 6 = 24, 5 * 24 = 120`

#### When to Use Recursion

* For problems that have a **natural recursive structure**
  * Tree traversal
  * Fractals
  * Some mathematical calculations (like factorial)
* When it makes the solution **clearer**

#### When NOT to Use Recursion

* For **deep recursion** (risk of stack overflow)
* When **iteration would be simpler**
* When **performance is critical** (recursion has overhead)

### 🔍 Higher-Order Functions

**Higher-order functions** are functions that work with other functions. They can either:
* Take one or more functions as **arguments**

This powerful concept allows us to write more flexible and reusable code.

#### Functions as Arguments

The most common use of higher-order functions is **passing functions as arguments**:

````python
def apply_to_list(numbers: list, operation: callable) -> list:
    """
    Apply an operation to each number in a list.
    
    Args:
        numbers: List of numbers to process
        operation: Function to apply to each number
        
    Returns:
        List with operation applied to each number
    """
    return [operation(num) for num in numbers]

# Simple operations
def double(x): return x * 2
def square(x): return x ** 2
def negative(x): return -x

numbers = [1, 2, 3, 4, 5]

# Using different operations
doubled = apply_to_list(numbers, double)
squared = apply_to_list(numbers, square)
negated = apply_to_list(numbers, negative)


print(f"Original: {numbers}")   # [1, 2, 3, 4, 5]
print(f"Doubled: {doubled}")    # [2, 4, 6, 8, 10]
print(f"Squared: {squared}")    # [1, 4, 9, 16, 25]
print(f"Negated: {negated}")    # [-1, -2, -3, -4, -5]
````

### 🔍 Closures

**Closures** are functions that *remember values* from their creation environment, even after that environment is gone. They can be thought of as functions with built-in memory.

#### Basic Closure Example

````python
def create_multiplier(factor):
    """
    Create a function that multiplies numbers by a fixed factor.
    
    This demonstrates a basic closure where the returned function
    remembers the 'factor' value.
    """
    def multiplier(x):
        return x * factor  # 'factor' is remembered from outer scope
    return multiplier

# Create specialized multiplier functions
double = create_multiplier(2)
triple = create_multiplier(3)
half = create_multiplier(0.5)

print(double(10))  # 20
print(triple(10))  # 30
print(half(10))    # 5
````

#### Practical Example: Configurable Logger


````python
def create_logger(prefix, log_level=1):
    """
    Create a logging function with predefined prefix and level.
    
    Args:
        prefix: String to prepend to all log messages
        log_level: Minimum level for messages to be logged
        
    Returns:
        Logging function with embedded prefix and level
    """
    def log(message, level=1):
        if level >= log_level:
            print(f"[{prefix}] {message}")
    
    return log

# Create specialized loggers
debug_logger = create_logger("DEBUG", log_level=1)
error_logger = create_logger("ERROR", log_level=2)
critical_logger = create_logger("CRITICAL", log_level=3)

# Use the loggers
debug_logger("Starting application...", level=1)     # Prints
error_logger("Invalid input", level=1)              # Doesn't print
error_logger("Database connection failed", level=2)  # Prints
critical_logger("System crash", level=3)            # Prints
````

#### Advanced Example: Memoization with Closure


````python
def create_memoized_function(func):
    """
    Create a memoized version of a function that caches results.
    
    Args:
        func: Function to memoize
        
    Returns:
        Memoized version of the function
    """
    cache = {}  # This cache persists in the closure
    
    def memoized(*args):
        if args not in cache:
            cache[args] = func(*args)
        return cache[args]
    
    return memoized

# Example: Memoized factorial function
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n-1)

# Create memoized version
memoized_factorial = create_memoized_function(factorial)

# First call: calculates factorial
print(memoized_factorial(5))  # Calculates: 120

# Second call: returns cached result
print(memoized_factorial(5))  # Instant return: 120
````

## **Chapter 5.4:** Function Best Practices

### 🔍 Writing Clean and Efficient Functions

In this section, we'll explore best practices for creating functions that are not just functional but also well-designed, maintainable, and efficient. Following these guidelines will help you write better code and make your functions more valuable to yourself and others.

#### The Single Responsibility Principle

One of the most important principles in function design is the **Single Responsibility Principle**: each function should do **one thing** and do it *well*.

````python
# Bad practice: Function does too many things
def process_data(file_path):
    """Load data, clean it, analyze it, and save results."""
    # Load data
    with open(file_path, 'r') as f:
        data = f.read()
    
    # Clean data
    data = data.strip().lower()
    
    # Analyze data
    word_count = len(data.split())
    
    # Save results
    with open('results.txt', 'w') as f:
        f.write(f"Word count: {word_count}")
    
    return word_count
````

````python
# Good practice: Each function has a single responsibility
def load_data(file_path):
    """Load data from a file."""
    with open(file_path, 'r') as f:
        return f.read()

def clean_data(data):
    """Clean the input data."""
    return data.strip().lower()

def count_words(text):
    """Count words in text."""
    return len(text.split())

def save_results(results, file_path):
    """Save results to a file."""
    with open(file_path, 'w') as f:
        f.write(f"Word count: {results}")
````

By breaking the larger function into smaller, focused functions, you make your code:
- More **maintainable**: easier to find and fix bugs
- More **reusable**: each function can be used in different contexts
- More **testable**: each function can be tested independently
- More **readable**: easier to understand what each part does

#### Descriptive Function Names

Function names should **clearly indicate** what the function does:


````python
# Bad practice: Unclear function name
def f(x, y):
    return x + y

# Good practice: Descriptive function name
def add_numbers(x, y):
    """Add two numbers together."""
    return x + y
````

Good function names:
- Use **verbs** or **verb phrases** (e.g., `calculate_area`, `get_user_data`)
- Are **specific** enough to understand what they do
- Follow a **consistent style** (Python standard is `snake_case`)
- Are neither too short nor too long

#### Limiting Arguments

- **Harder to understand**

- **More likely to contain bugs**

### 🔍 Handling Errors and Exceptions in Functions

Robust functions handle errors gracefully rather than crashing. This is achieved through proper **exception handling**.

#### Input Validation

Always validate the inputs to your functions to prevent errors:

````python
def calculate_average(numbers):
    """Calculate the average of a list of numbers."""
    # Validate input
    if not numbers:
        raise ValueError("Cannot calculate average of empty list")
    
    if not all(isinstance(n, (int, float)) for n in numbers):
        raise TypeError("All items in the list must be numbers")
    
    return sum(numbers) / len(numbers)
````

#### Using Try-Except Blocks

Use **try-except blocks** to handle potential errors gracefully:

````python
def divide_numbers(a, b):
    """Safely divide two numbers."""
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Division by zero!")
        return None
    except TypeError:
        print("Error: Inputs must be numbers!")
        return None
    else:
        return result
    finally:
        print("Division operation attempted.")
````

This function handles two potential exceptions:
- `ZeroDivisionError`: when trying to divide by zero
- `TypeError`: when the inputs aren't numbers

The `else` clause runs if no exceptions occur, and the `finally` clause runs regardless of whether an exception was raised.

#### Re-raising Exceptions

Sometimes you want to catch an exception, do something about it, and then re-raise it or raise a different exception:

````python
def get_user_data(user_id):
    """Get data for a user, with additional error information."""
    try:
        return database.query(user_id)
    except DatabaseError as e:
        logger.error(f"Database error for user {user_id}: {e}")
        raise UserDataError(f"Could not retrieve data for user {user_id}") from e
````

The `raise ... from e` syntax preserves the original exception information while providing a more context-specific error message.

### 🔍 Performance Considerations

While writing clear and maintainable code is usually the priority, there are times when performance matters:

#### Avoid Unnecessary Computations

````python
# Less efficient: Calculates the length of items multiple times
def process_items(items):
    for i in range(len(items)):
        print(f"Processing item {i+1} of {len(items)}")
        # Process item...
````

````python
# More efficient: Calculate length once
def process_items_efficient(items):
    items_count = len(items)
    for i in range(items_count):
        print(f"Processing item {i+1} of {items_count}")
        # Process item...
````

#### Use Generators for Large Datasets

When working with large sequences, generators can save memory:

````python
# Memory intensive: Creates a full list in memory
def get_even_numbers(n):
    return [x for x in range(n) if x % 2 == 0]

# Memory efficient: Uses a generator
def get_even_numbers_generator(n):
    for x in range(n):
        if x % 2 == 0:
            yield x
````

````python
# Check memory size of objects and generators

import sys 

# List comprehension
numbers_list = [x for x in range(1000)]
print(sys.getsizeof(numbers_list))  # Output: 8856 Bytes

# Generator expression
numbers_gen = (x for x in range(1000))
print(sys.getsizeof(numbers_gen))   # Output: 200 Bytes
````

#### Memoization for Expensive Calculations

For functions with expensive calculations that might be called multiple times with the same arguments:

````python
def memoize(func):
    """Memoization decorator for functions taking one or more arguments."""
    cache = {}
    
    def wrapper(*args):
        if args not in cache:
            cache[args] = func(*args)
        return cache[args]
    
    return wrapper

@memoize
def fibonacci(n):
    """Calculate the nth Fibonacci number."""
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
````

This dramatically improves performance for recursive functions like `fibonacci` by storing previously calculated results.

## 📚 Key Terms

- **Function:** A named block of reusable code that performs a specific task
- **Parameter:** A variable in a function definition that accepts values from function callers
- **Argument:** A value passed to a function when it is called
- **Return Statement:** The mechanism for functions to send results back to the caller
- **Default Parameter:** A parameter that assumes a default value if no argument is provided
- **Keyword Argument:** An argument passed to a function by explicitly naming the parameter
- **Positional Argument:** An argument passed to a function based on its position in the parameter list
- **`*args`:** Syntax for accepting a variable number of positional arguments as a tuple
- **`**kwargs`:** Syntax for accepting a variable number of keyword arguments as a dictionary
- **Lambda Function:** A small, anonymous function defined with the `lambda` keyword
- **Recursive Function:** A function that calls itself to solve problems by breaking them down
- **Higher-Order Function:** A function that takes other functions as arguments or returns functions
- **Closure:** A function that retains access to variables from its defining scope, even after that scope has completed
- **Docstring:** A string at the beginning of a function that documents its purpose and usage
