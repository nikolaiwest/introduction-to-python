## **Chapter 6.1:** Classes and Objects Fundamentals

### 🔍 Understanding Classes

In Python, **classes** provide a way to bundle data (attributes) and functionality (methods) together. They are the blueprint or template that defines what an object will look like and how it will behave. Think of a class as a *factory* that produces objects with certain characteristics and capabilities.

Classes help us model real-world entities or abstract concepts in our code. For example, instead of managing separate lists for student names, ages, and grades, we can create a `Student` class that keeps all this information together with relevant methods like calculating grade averages.

### Why Do We Need Classes?

Consider how we might model a bank account using simple variables and functions:

````python
# Without classes - procedural approach
def create_account(account_number, balance):
    return {
        "account_number": account_number,
        "balance": balance
    }

def deposit(account, amount):
    account["balance"] += amount
    
def withdraw(account, amount):
    if amount <= account["balance"]:
        account["balance"] -= amount
        return True
    return False

def get_balance(account):
    return account["balance"]
````

````python
# Using these functions
account1 = create_account("12345", 1000)
account2 = create_account("67890", 500)

deposit(account1, 100)
deposit(account2, 250)

print(f"Account {account1['account_number']}: Balance = ${get_balance(account1)}")
print(f"Account {account2['account_number']}: Balance = ${get_balance(account2)}")
````

Now let's see how we could implement the same functionality using a class:


````python
# With classes - object-oriented approach
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance
        
    def deposit(self, amount):
        self.balance += amount
        
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return True
        return False
    
    def get_balance(self):
        return self.balance

# Using the class
account1 = BankAccount("12345", 1000)
account2 = BankAccount("67890", 500)

account1.deposit(100)
account2.deposit(250)

print(f"Account {account1.account_number}: Balance = ${account1.get_balance()}")
print(f"Account {account2.account_number}: Balance = ${account2.get_balance()}")
````

While both approaches accomplish the same task, the class-based approach offers several advantages:

1. **Encapsulation**: Data (account_number, balance) and behavior (deposit, withdraw) are bundled together
2. **Code organization**: The association between data and related functions is explicit
3. **Clarity**: Methods like `account.deposit()` read more naturally than `deposit(account)`
4. **Extensibility**: Classes can be extended through inheritance to add new features

### 🧩 Important Components

* **Class Definition**: The blueprint that defines attributes and methods using the `class` keyword
* **Attributes**: Variables that store data within a class or object
* **Methods**: Functions defined within a class that operate on class data
* **Constructor**: The special `__init__` method that initializes a new object
* **Self Parameter**: Reference to the specific instance being manipulated
* **Instances/Objects**: The actual objects created from the class blueprint

### 📘 Main Content

#### Defining a Class

In Python, you define a class using the `class` keyword followed by the class name and a colon. Class names in Python typically use **PascalCase** (also called CapWords) naming convention:

````python
class Person:
    """A class representing a person with a name and age."""
    
    def __init__(self, name, age):
        """Initialize a new Person object.
        
        Args:
            name (str): The person's name
            age (int): The person's age
        """
        self.name = name
        self.age = age
        
    def greet(self):
        """Return a greeting message from this person."""
        return f"Hello, my name is {self.name} and I am {self.age} years old."
````

Let's break down the components of this class:

1. **Class Definition**: `class Person:` declares a new class named Person
2. **Docstring**: Triple-quoted string documenting what the class represents
3. **Constructor**: `__init__` method that initializes new Person objects
4. **Instance Attributes**: `self.name` and `self.age` are data specific to each object
5. **Method**: `greet()` is a function that operates on the object's data

#### Creating Objects

Once a class is defined, you can create **instances** (also called objects) of that class:

````python
# Creating instances of the Person class
person1 = Person("Alice", 30)
person2 = Person("Bob", 25)

# Calling methods on the instances
print(person1.greet())
print(person2.greet())
````

When you create an instance, Python automatically calls the `__init__` method with the arguments you provide. The first parameter (`self`) is automatically filled with a reference to the new object being created.

#### Methods and the Self Parameter

All instance methods in a class have `self` as their first parameter. This parameter represents the specific instance that the method is being called on.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def have_birthday(self):
        """Increment the person's age by 1."""
        self.age += 1
        return f"{self.name} is now {self.age} years old."
    
# Create an instance
alice = Person("Alice", 30)

# Call the method on the instance
print(alice.have_birthday())
````

When you call `alice.have_birthday()`, Python automatically passes the `alice` object as the `self` parameter. This is why methods can access and modify the instance's attributes using `self.attribute_name`.

This mechanism is what allows different instances to maintain their own separate data while sharing the same method implementations.

#### Attributes

There are two main types of attributes in Python classes:

1. **Instance attributes**: Belong to a specific instance (defined with `self.attribute_name`)
2. **Class attributes**: Shared by all instances of the class (defined directly in the class)

````python
class Person:
    # Class attribute - shared by all instances
    species = "Human"
    
    def __init__(self, name):
        # Instance attribute - unique to each instance
        self.name = name

# Create instances
person1 = Person("Marco")
person2 = Person("Michelle")

# Access class attribute (shared)
print(Person.species)  # Accessing from the class
print(person1.species) # Accessing from instance 1
print(person2.species) # Accessing from instance 2

# Access instance attributes (unique)
print(person1.name)    # Marco
print(person2.name)    # Michelle
````

Class attributes are defined directly within the class but outside of any methods. They're useful for values that should be the same across all instances or for default values.

Instance attributes are usually defined in the `__init__` method using `self`. They can hold values that vary from one instance to another.

You can also add or modify attributes of an instance after it's created:

````python
# Adding an attribute after creation
person1.job = "Software Engineer"
print(person1.job)
````

#### Encapsulation and Information Hiding

Encapsulation is the bundling of data with the methods that operate on that data. It's one of the fundamental principles of OOP and helps to hide the internal state of an object and only expose what is necessary.

While Python doesn't have strong access control mechanisms like some other languages, it follows a convention where names starting with an underscore are considered "private" (though not enforced by the language):

````python
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self._balance = balance  # "Private" attribute by convention
        
    def get_balance(self):
        return self._balance
    
    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            return self._balance
        raise ValueError("Deposit amount must be positive")
    
    def withdraw(self, amount):
        if 0 < amount <= self._balance:
            self._balance -= amount
            return True
        return False
````

````python
# Using the encapsulated class
account = BankAccount("12345", 1000)
print(f"Balance: ${account.get_balance()}")
````

````python
# Use methods instead of directly accessing _balance
print(f"New balance after deposit: ${account.deposit(100)}")

withdrawal = account.withdraw(500)
print(f"Withdrawal successful: {withdrawal}")
print(f"New balance after withdrawal: ${account.get_balance()}")
````

````python
# Try to withdraw too much
withdrawal = account.withdraw(1000)
print(f"Withdrawal successful: {withdrawal}")
print(f"Balance remains: ${account.get_balance()}")
````

In this example, the `_balance` attribute is marked as "private" by convention with the leading underscore. Instead of accessing it directly, we provide methods like `get_balance()`, `deposit()`, and `withdraw()` that allow controlled access to this data.

This approach has several benefits:
- We can validate inputs (e.g., checking that deposit amounts are positive)
- We can control how data is modified (e.g., preventing withdrawals exceeding the balance)
- We can change the internal implementation without affecting code that uses the class

#### ⚠️ Common Mistakes with Classes and Objects

* **Forgetting self**: Always include `self` as the first parameter in instance methods
* **Confusing instance and class attributes**: Remember that class attributes are shared by all instances
* **Direct attribute access**: Avoid accessing attributes directly when encapsulation is important
* **Forgetting to initialize attributes**: Always set all necessary attributes in `__init__`

## **Chapter 6.2:** Special Methods and Operators

### 🔍 Magic Methods (Dunder Methods)

In Python, **special methods** (also called **magic methods** or **dunder methods** because of their double underscores) allow you to define how objects of your class behave in various situations. These methods are called automatically by Python in response to certain operations or syntax.

You've already seen one magic method: `__init__`, which is called when a new object is created. But there are many others that let you customize how your objects behave.

Special methods enable your classes to integrate smoothly with Python's language features, allowing your objects to work with built-in functions and operators just like Python's built-in types.

### 🧩 Important Components

* **`__init__`**: Initialize a new object (constructor)
* **`__str__`** and **`__repr__`**: Define string representations
* **Comparison methods**: `__eq__`, `__lt__`, `__gt__`, etc.
* **Arithmetic operators**: `__add__`, `__sub__`, `__mul__`, etc.
* **Container behavior**: `__len__`, `__getitem__`, `__contains__`, etc.
* **Function-like behavior**: `__call__`

### 📘 Main Content

#### String Representation with `__str__` and `__repr__`

When you print an object or convert it to a string with `str()`, Python calls the object's `__str__` method. This method should return a user-friendly string representation of the object:

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        """Return a user-friendly string representation."""
        return f"Point({self.x}, {self.y})"

point = Point(3, 4)
print(point)  # Output: Point(3, 4)
````

The `__repr__` method is similar but is meant to provide an unambiguous representation that should, if possible, be valid Python code that could recreate the object:


````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"Point({self.x}, {self.y})"
    
    def __repr__(self):
        """Return a representation that could recreate this object."""
        return f"Point({self.x}, {self.y})"

point = Point(3, 4)
print(str(point))   # Calls __str__: Point(3, 4)
print(repr(point))  # Calls __repr__: Point(3, 4)
print([point])      # Lists use __repr__: [Point(3, 4)]
````

If you define `__repr__` but not `__str__`, Python will use `__repr__` for both. It's good practice to define both methods: `__str__` for user-friendly output and `__repr__` for debugging and development.

#### Operator Overloading

Python allows you to define how operators like `+`, `-`, `*`, etc. behave with your custom objects by implementing the corresponding special methods:

````python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"Vector({self.x}, {self.y})"
    
    def __add__(self, other):
        """Define vector addition (called for v1 + v2)."""
        return Vector(self.x + other.x, self.y + other.y)
    
    def __sub__(self, other):
        """Define vector subtraction (called for v1 - v2)."""
        return Vector(self.x - other.x, self.y - other.y)
    
    def __mul__(self, scalar):
        """Define scalar multiplication (called for v * s)."""
        return Vector(self.x * scalar, self.y * scalar)
    
    def __rmul__(self, scalar):
        """Reverse scalar multiplication (called for s * v)."""
        return self.__mul__(scalar)
    
    def __eq__(self, other):
        """Define equality comparison (called for v1 == v2)."""
        if not isinstance(other, Vector):
            return False
        return self.x == other.x and self.y == other.y

# Using the overloaded operators
v1 = Vector(3, 4)
v2 = Vector(1, 2)

print(v1 + v2)    # Vector(4, 6)
print(v1 - v2)    # Vector(2, 2)
print(v1 * 2)     # Vector(6, 8)
print(2 * v1)     # Vector(6, 8) (using __rmul__)
print(v1 == v2)   # False
````

Here's a table of common operators and their corresponding special methods:

| Operator | Method | Description |
|----------|--------|-------------|
| `+` | `__add__(self, other)` | Addition |
| `-` | `__sub__(self, other)` | Subtraction |
| `*` | `__mul__(self, other)` | Multiplication |
| `/` | `__truediv__(self, other)` | Division |
| `//` | `__floordiv__(self, other)` | Floor division |
| `%` | `__mod__(self, other)` | Modulo |
| `**` | `__pow__(self, other)` | Exponentiation |
| `==` | `__eq__(self, other)` | Equality |
| `!=` | `__ne__(self, other)` | Inequality |
| `<` | `__lt__(self, other)` | Less than |
| `<=` | `__le__(self, other)` | Less than or equal |
| `>` | `__gt__(self, other)` | Greater than |
| `>=` | `__ge__(self, other)` | Greater than or equal |

#### Container Behavior

You can also make your objects behave like containers (lists, dictionaries, etc.) by implementing the appropriate special methods:

````python
class Deck:
    def __init__(self):
        self.cards = ['A', 'K', 'Q', 'J', '10', '9', '8', '7', '6', '5', '4', '3', '2']
    
    def __len__(self):
        """Return the number of cards in the deck (called by len(deck))."""
        return len(self.cards)
    
    def __getitem__(self, position):
        """Return the card at the given position (called by deck[position])."""
        return self.cards[position]
    
    def __contains__(self, card):
        """Check if the card is in the deck (called by card in deck)."""
        return card in self.cards

# Using container behaviors
deck = Deck()
print(len(deck))     # 13
print(deck[0])       # A
print('K' in deck)   # True
print('7' in deck)   # True
print('1' in deck)   # False

# We can even iterate through the deck
for card in deck:
    print(card, end=' ')  # A K Q J 10 9 8 7 6 5 4 3 2
````

By implementing these methods, our `Deck` class now behaves like a sequence that we can iterate through, index into, check membership, and get the length of.

#### Callable Objects with `__call__`

The `__call__` method lets you make instances of your class callable, just like functions:

````python
class Adder:
    def __init__(self, n):
        self.n = n
    
    def __call__(self, x):
        """Make this object callable like a function."""
        return self.n + x

# Create callable objects
add5 = Adder(5)
add10 = Adder(10)

# Call them like functions
print(add5(3))   # 8 (5 + 3)
print(add10(3))  # 13 (10 + 3)
````

In this example, `add5` and `add10` are objects, but they can be called like functions. This is useful for creating function-like objects that maintain state between calls.

#### Other Useful Special Methods

There are many other special methods that you might find useful:

* **`__bool__`**: Determines the truth value of an object (called by `bool(obj)`)
* **`__iter__` and `__next__`**: Make an object iterable (used in `for` loops)
* **`__enter__` and `__exit__`**: Support context management with the `with` statement
* **`__del__`**: Called when an object is garbage-collected (similar to a destructor)

#### ⚠️ Common Mistakes with Special Methods

* **Forgetting parameter requirements**: Check the number and types of parameters that each special method requires
* **Returning the wrong type**: Many special methods expect specific return types
* **Modifying self instead of returning a new object**: Methods like `__add__` should return a new object, not modify the existing one
* **Not implementing reverse operations**: For operators, consider implementing the reverse methods (like `__radd__`) for mixed-type operations

## **Chapter 6.3:** Properties and Access Control

### 🔍 Properties

In Python, **properties** provide a way to control access to class attributes. Using properties, you can create attributes that behave like regular attributes but actually execute special methods when accessed, modified, or deleted.

Properties allow you to:
- Add validation when setting values
- Compute values on demand
- Implement read-only or write-only attributes
- Change the internal implementation without affecting the public interface

### 🧩 Important Components

* **`@property` decorator**: Defines a getter method for an attribute
* **`@name.setter`**: Defines a setter method for the same attribute
* **`@name.deleter`**: Defines a deleter method for the attribute
* **Name mangling**: Using a double underscore (`__`) prefix for "private" attributes
* **Class methods**: Methods that operate on the class rather than instances
* **Static methods**: Methods that don't require access to the class or instances

### 📘 Main Content

#### Creating Properties

Properties are created using the `@property` decorator and optional setter and deleter decorators:

````python
class Circle:
    def __init__(self, radius):
        self._radius = radius  # Note the single underscore
    
    @property
    def radius(self):
        """Getter method for radius."""
        return self._radius
    
    @radius.setter
    def radius(self, value):
        """Setter method for radius with validation."""
        if value <= 0:
            raise ValueError("Radius must be positive")
        self._radius = value
    
    @property
    def diameter(self):
        """Calculated property for diameter."""
        return self._radius * 2
    
    @property
    def area(self):
        """Calculated property for area."""
        return 3.14159 * self._radius * self._radius
````

````python
# Using properties
circle = Circle(5)
print(circle.radius)    # 5 (calls the radius getter)

circle.radius = 10      # Calls the radius setter
print(circle.radius)    # 10

print(circle.diameter)  # 20 (calculated on demand)
print(circle.area)      # 314.159 (calculated on demand)

# This would raise an error due to validation
# circle.radius = -5    # ValueError: Radius must be positive

# This would raise an error since diameter has no setter
# circle.diameter = 30  # AttributeError: can't set attribute
````

In this example:
- `radius` is a property with both a getter and a setter
- `diameter` and `area` are read-only properties (they have getters but no setters)
- The setter for `radius` includes validation to ensure the value is positive

Properties are accessed and modified just like regular attributes, but they execute the getter and setter methods behind the scenes.

#### Read-Only Properties

If you only define a getter method with `@property` but no setter method, you create a read-only property:

````python
class BankAccount:
    def __init__(self, account_number, balance):
        self._account_number = account_number
        self._balance = balance
    
    @property
    def account_number(self):
        """Read-only property for account number."""
        return self._account_number
    
    @property
    def balance(self):
        """Read-only property for balance."""
        return self._balance
    
    def deposit(self, amount):
        """Method to deposit money."""
        if amount <= 0:
            raise ValueError("Deposit amount must be positive")
        self._balance += amount
    
    def withdraw(self, amount):
        """Method to withdraw money."""
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount

# Using read-only properties
account = BankAccount("12345", 1000)
print(account.account_number)  # 12345
print(account.balance)        # 1000

# These would raise errors
# account.account_number = "54321"  # AttributeError: can't set attribute
# account.balance = 2000           # AttributeError: can't set attribute

# Use methods instead
account.deposit(500)
print(account.balance)  # 1500
````

Read-only properties are useful for attributes that should never be directly modified, only read or modified through specific methods.

#### Property Deleter

You can also define a deleter method that's called when an attribute is deleted with `del`:

````python
class Person:
    def __init__(self, name):
        self._name = name
    
    @property
    def name(self):
        """Getter for name."""
        return self._name
    
    @name.setter
    def name(self, value):
        """Setter for name."""
        if not value:
            raise ValueError("Name cannot be empty")
        self._name = value
    
    @name.deleter
    def name(self):
        """Deleter for name."""
        print(f"Deleting name: {self._name}")
        self._name = None

# Using a deleter
person = Person("Alice")
print(person.name)  # Alice

del person.name     # Prints: Deleting name: Alice
print(person.name)  # None
````

Deleters are less commonly used than getters and setters but can be useful for cleanup operations or to reset attributes to default values.

#### Access Control

Unlike some other languages, Python doesn't have built-in access modifiers like "private" or "protected". Instead, it follows a convention of using name prefixes to indicate the intended use of attributes and methods:

- **No prefix**: Public attributes and methods, intended for external use
- **Single underscore** (`_`): "Protected" attributes and methods, intended for internal use but still accessible
- **Double underscore** (`__`): "Private" attributes and methods, name-mangled to make external access more difficult

Let's look at an example using these conventions:

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner            # Public attribute
        self._balance = balance       # Protected attribute
        self.__transaction_log = []   # Private attribute
    
    def deposit(self, amount):
        """Public method."""
        self._balance += amount
        self.__log_transaction("deposit", amount)
    
    def withdraw(self, amount):
        """Public method."""
        if amount <= self._balance:
            self._balance -= amount
            self.__log_transaction("withdraw", amount)
            return True
        return False
    
    def _calculate_interest(self):
        """Protected method."""
        return self._balance * 0.05
    
    def __log_transaction(self, transaction_type, amount):
        """Private method."""
        self.__transaction_log.append((transaction_type, amount))

# Using different levels of access
account = BankAccount("Alice", 1000)

# Public access - intended for external use
print(account.owner)       # Alice
account.deposit(500)
print(account.withdraw(200))  # True

# Protected access - possible but discouraged
print(account._balance)    # 1300
interest = account._calculate_interest()  # Possible but not recommended

# Private access - not directly possible
# This would raise an AttributeError:
# print(account.__transaction_log)  
# account.__log_transaction("test", 100)  

# However, name mangling just changes the name:
print(account._BankAccount__transaction_log)  # [('deposit', 500), ('withdraw', 200)]
account._BankAccount__log_transaction("test", 100)
````

It's important to note that these conventions are not enforced by Python - they're just conventions. The double underscore prefix doesn't truly make attributes private; it just triggers name mangling (the attribute is renamed to `_ClassName__attribute`). 

The Python philosophy is "We're all consenting adults here" - developers are trusted to respect the conventions and not access attributes or methods they shouldn't.

#### Class and Static Methods

In addition to regular instance methods, Python supports two special types of methods:

1. **Class methods**: Methods that operate on the class itself rather than instances
2. **Static methods**: Methods that don't require access to either the class or its instances

Class methods are defined with the `@classmethod` decorator and take the class (`cls`) as their first parameter:

````python
class Employee:
    company_name = "TechCorp"
    all_employees = []
    
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.all_employees.append(self)
    
    @classmethod
    def set_company_name(cls, name):
        """Class method to change the company name."""
        cls.company_name = name
    
    @classmethod
    def from_string(cls, employee_string):
        """Alternative constructor that creates an employee from a string."""
        name, salary_str = employee_string.split(',')
        return cls(name, float(salary_str))
    
    @classmethod
    def average_salary(cls):
        """Calculate the average salary of all employees."""
        total = sum(employee.salary for employee in cls.all_employees)
        return total / len(cls.all_employees) if cls.all_employees else 0

# Using class methods
emp1 = Employee("Alice", 60000)
emp2 = Employee("Bob", 70000)

# Change class attribute for all instances
Employee.set_company_name("GlobalTech")
print(Employee.company_name)  # GlobalTech
print(emp1.company_name)      # GlobalTech

# Use alternative constructor
emp3 = Employee.from_string("Charlie,80000")
print(emp3.name, emp3.salary)  # Charlie 80000.0

# Calculate class-level statistics
print(Employee.average_salary())  # 70000.0
````

Class methods are useful for:
- Altering class attributes that affect all instances
- Creating factory methods or alternative constructors
- Performing operations that involve all instances of a class

Static methods, defined with the `@staticmethod` decorator, don't receive an implicit first parameter:

````python
class MathUtil:
    @staticmethod
    def is_prime(n):
        """Check if a number is prime."""
        if n < 2:
            return False
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                return False
        return True
    
    @staticmethod
    def factorial(n):
        """Calculate the factorial of n."""
        result = 1
        for i in range(2, n + 1):
            result *= i
        return result

# Using static methods
print(MathUtil.is_prime(17))    # True
print(MathUtil.is_prime(15))    # False
print(MathUtil.factorial(5))    # 120
````

Static methods are essentially regular functions that are logically grouped within a class. They don't access or modify class or instance data.

#### ⚠️ Common Mistakes with Properties and Access Control

* **Forgetting the `self` parameter**: Even in property methods, `self` is required as the first parameter
* **Inconsistent naming**: Make sure your property name doesn't conflict with the underlying attribute name
* **Direct access to protected or private attributes**: Respect the intended access level, even though Python doesn't enforce it
* **Overusing properties**: For simple attributes that don't need validation or computation, regular attributes are simpler

## **Chapter 6.4:** Inheritance and Polymorphism

### 🔍 Inheritance

**Inheritance** is a fundamental concept in object-oriented programming that allows a class to inherit attributes and methods from another class. The class that inherits is called a **subclass** (or **derived class**), and the class being inherited from is called a **superclass** (or **base class**).

Inheritance enables code reuse and the creation of hierarchical relationships between classes. It models "is-a" relationships, where a subclass is a specialized version of its superclass.

### 🧩 Important Components

* **Base class/Superclass**: The original class being inherited from
* **Derived class/Subclass**: The class that inherits from the base class
* **`super()` function**: Used to call methods from the parent class
* **Method overriding**: Redefining methods in subclasses
* **Multiple inheritance**: Inheriting from more than one class
* **Method Resolution Order (MRO)**: The order Python uses to resolve method calls in inheritance hierarchies
* **Abstract base classes**: Classes that define an interface but don't provide complete implementation

### 📘 Main Content

#### Creating Subclasses

To create a subclass, you simply specify the parent class(es) in parentheses after the class name:

````python
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        """Default implementation for animals that don't specialize."""
        return "Some generic animal sound"
    
    def move(self):
        return "Moving..."

class Dog(Animal):  # Dog inherits from Animal
    def speak(self):
        """Override the speak method for dogs."""
        return "Woof!"
    
    def fetch(self):
        """Add a new method specific to dogs."""
        return f"{self.name} is fetching the ball!"

# Create instances
generic_animal = Animal("Generic")
dog = Dog("Buddy")

# Access inherited attributes
print(dog.name)  # Buddy (inherited from Animal)

# Call inherited methods
print(dog.move())  # Moving... (inherited from Animal)

# Call overridden methods
print(generic_animal.speak())  # Some generic animal sound
print(dog.speak())  # Woof! (overridden in Dog)

# Call methods specific to the subclass
print(dog.fetch())  # Buddy is fetching the ball!
````

In this example, `Dog` inherits all attributes and methods from `Animal` but overrides the `speak` method to provide a dog-specific implementation. It also adds a new method `fetch` that only dogs have.

#### Using `super()` to Access Parent Methods

The `super()` function allows you to call methods from the parent class. This is particularly useful in the `__init__` method to ensure the parent's initialization is performed:

````python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def info(self):
        return f"{self.name} is {self.age} years old"

class Dog(Animal):
    def __init__(self, name, age, breed):
        # Call the parent's __init__ method
        super().__init__(name, age)
        # Add dog-specific attribute
        self.breed = breed
    
    def info(self):
        # Call the parent's info method and extend it
        basic_info = super().info()
        return f"{basic_info} and is a {self.breed}"

# Create a dog
dog = Dog("Buddy", 5, "Golden Retriever")
print(dog.info())  # Buddy is 5 years old and is a Golden Retriever
````

Using `super()` allows you to extend the functionality of parent methods rather than completely replacing them. This promotes code reuse and makes your classes more maintainable.

#### Method Overriding

Method overriding occurs when a subclass provides a different implementation of a method that's already defined in its parent class. This allows subclasses to specialize behavior while maintaining the same interface:

````python
class Shape:
    def __init__(self, color="black"):
        self.color = color
    
    def area(self):
        """Base implementation."""
        return 0
    
    def describe(self):
        return f"A {self.color} shape with area {self.area()}"

class Circle(Shape):
    def __init__(self, radius, color="red"):
        super().__init__(color)
        self.radius = radius
    
    def area(self):
        """Override area method for circles."""
        return 3.14159 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self, width, height, color="blue"):
        super().__init__(color)
        self.width = width
        self.height = height
    
    def area(self):
        """Override area method for rectangles."""
        return self.width * self.height

# Create shapes
circle = Circle(5)
rectangle = Rectangle(4, 6)

# Use polymorphic behavior
print(circle.describe())  # A red shape with area 78.53975
print(rectangle.describe())  # A blue shape with area 24
````

In this example, both `Circle` and `Rectangle` override the `area` method to provide shape-specific calculations, but they inherit the `describe` method which calls `area()`. This shows polymorphism in action - the `describe` method behaves differently depending on which type of shape it's called on.

#### Polymorphism

**Polymorphism** allows objects of different classes to be treated as objects of a common superclass. The most common use of polymorphism is when a parent class reference is used to refer to a child class object.

In Python, polymorphism is achieved through method overriding and duck typing (if it walks like a duck and quacks like a duck, it's treated as a duck):

````python
def describe_shape(shape):
    """Function that works with any shape object."""
    print(f"Description: {shape.describe()}")
    print(f"Area: {shape.area()}")

# Create different shapes
shapes = [
    Circle(3, "purple"),
    Rectangle(5, 7, "green"),
    Circle(10, "yellow")
]

# Process all shapes polymorphically
for shape in shapes:
    describe_shape(shape)
````

This function `describe_shape` works with any object that has `describe` and `area` methods, regardless of its actual class. This demonstrates how polymorphism allows for more flexible and extensible code.

#### Multiple Inheritance

Python supports **multiple inheritance**, allowing a class to inherit from more than one parent class:

````python
class FlyingMixin:
    def fly(self):
        return f"{self.name} is flying high!"

class SwimmingMixin:
    def swim(self):
        return f"{self.name} is swimming"

class Duck(Animal, FlyingMixin, SwimmingMixin):
    def speak(self):
        return "Quack!"

# Create a duck
duck = Duck("Donald")

# Use methods from all parent classes
print(duck.speak())  # Quack! (from Duck)
print(duck.move())   # Moving... (from Animal)
print(duck.fly())    # Donald is flying high! (from FlyingMixin)
print(duck.swim())   # Donald is swimming (from SwimmingMixin)
````

In this example, `Duck` inherits from `Animal` and two mixin classes that provide additional behaviors. This allows for creating complex class hierarchies and promoting code reuse.

#### Method Resolution Order (MRO)

When a class inherits from multiple parent classes, Python uses the **Method Resolution Order (MRO)** to determine which method to call. The MRO is a linearization of the inheritance graph that follows the C3 linearization algorithm.

You can check a class's MRO using the `__mro__` attribute or the `mro()` method:

````python
print(Duck.__mro__)
# Output: (<class '__main__.Duck'>, <class '__main__.Animal'>, <class '__main__.FlyingMixin'>, <class '__main__.SwimmingMixin'>, <class 'object'>)
````

The MRO determines the order in which Python searches for methods. In this example, if a method exists in both `Duck` and `Animal`, the one in `Duck` will be used. If it doesn't exist in `Duck` but exists in both `Animal` and `FlyingMixin`, the one in `Animal` will be used, and so on.

#### Abstract Base Classes

An **abstract base class (ABC)** is a class that cannot be instantiated and is designed to be subclassed. It defines an interface that derived classes must implement. In Python, ABCs are created using the `abc` module:

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self, color):
        self.color = color
    
    @abstractmethod
    def area(self):
        """All shapes must implement this method."""
        pass
    
    @abstractmethod
    def perimeter(self):
        """All shapes must implement this method."""
        pass
    
    def describe(self):
        return f"A {self.color} shape with area {self.area()} and perimeter {self.perimeter()}"

class Circle(Shape):
    def __init__(self, radius, color="red"):
        super().__init__(color)
        self.radius = radius
    
    def area(self):
        return 3.14159 * self.radius * self.radius
    
    def perimeter(self):
        return 2 * 3.14159 * self.radius

class Rectangle(Shape):
    def __init__(self, width, height, color="blue"):
        super().__init__(color)
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)

# This would raise an error
# shape = Shape("black")  # TypeError: Can't instantiate abstract class Shape

# Create concrete shapes
circle = Circle(5)
rectangle = Rectangle(4, 6)

print(circle.describe())
print(rectangle.describe())
````

Abstract base classes help enforce a common interface across related classes. If a subclass doesn't implement all the abstract methods, Python will raise an error when you try to instantiate it.

#### ⚠️ Common Mistakes with Inheritance and Polymorphism

* **Forgetting to call super().__init__**: Failing to initialize the parent class can lead to missing attributes
* **Diamond problem**: In multiple inheritance, if two parent classes have a method with the same name, it can be unclear which one should be used
* **Deep inheritance hierarchies**: Overly complex inheritance structures can make code harder to understand and maintain
* **Using inheritance when composition would be better**: Sometimes "has-a" relationships are more appropriate than "is-a" relationships

## **Chapter 6.5:** Modern Class Patterns

### 🔍 Dataclasses

Python 3.7 introduced **dataclasses**, a decorator and functions for automatically adding generated special methods to classes. Dataclasses are designed to reduce boilerplate code for classes that primarily store data, similar to "data transfer objects" or "records" in other languages.

The `dataclass` decorator automatically generates special methods like `__init__`, `__repr__`, and `__eq__` based on the class's fields, saving you from writing these common methods by hand.

### 🧩 Important Components

* **Dataclasses**: Simplified class creation for data containers
* **Named tuples**: Lightweight, immutable alternatives to classes
* **Composition**: Building complex objects from simpler ones
* **Inheritance vs. Composition**: Choosing the right approach for object relationships

### 📘 Main Content

#### Using Dataclasses

Dataclasses simplify the creation of classes that are primarily used to store data:

````python
from dataclasses import dataclass

@dataclass
class Point:
    x: float
    y: float

@dataclass
class Rectangle:
    width: float
    height: float
    position: Point
    
    def area(self):
        return self.width * self.height

# Create instances
p = Point(10, 20)
rect = Rectangle(5, 10, p)

print(p)  # Point(x=10, y=20)
print(rect)  # Rectangle(width=5, height=10, position=Point(x=10, y=20))
print(rect.area())  # 50

# Automatic equality comparison
p1 = Point(10, 20)
p2 = Point(10, 20)
print(p1 == p2)  # True
````

The `@dataclass` decorator automatically generated:
- An `__init__` method that initializes the fields
- A `__repr__` method that creates a string representation
- An `__eq__` method that compares instances based on their field values

Dataclasses support optional parameters to customize behavior:

````python
from dataclasses import dataclass, field

@dataclass(frozen=True)  # Make instances immutable
class ImmutablePoint:
    x: float
    y: float
    
    def distance(self):
        return (self.x**2 + self.y**2)**0.5

@dataclass
class Rectangle:
    width: float
    height: float
    color: str = "red"  # Default value
    tags: list = field(default_factory=list)  # Safely use mutable defaults
    
    def __post_init__(self):
        """This runs after the automatic __init__."""
        if self.width <= 0 or self.height <= 0:
            raise ValueError("Dimensions must be positive")
````

Key features of dataclasses:
- **Type annotations**: Fields are declared with type hints
- **Default values**: Fields can have default values
- **Post-initialization**: The `__post_init__` method runs after initialization for additional setup
- **Immutability**: With `frozen=True`, instances become immutable
- **Factory defaults**: The `field()` function helps with mutable default values

Dataclasses are particularly useful for:
- Classes that primarily store data
- Classes with many attributes
- Data transfer objects
- Configuration classes

#### Named Tuples

Another modern approach for simple data containers is **named tuples**. Named tuples are immutable, lightweight collections where the elements can be accessed by name as well as by index.

There are two main ways to create named tuples: the classic `collections.namedtuple` function and the newer `typing.NamedTuple` class (which supports type annotations):

````python
# Classic approach
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y'])
p1 = Point(1, 2)
print(p1.x, p1.y)  # 1 2
print(p1[0], p1[1])  # 1 2

# Modern approach with type annotations
from typing import NamedTuple

class Point(NamedTuple):
    x: float
    y: float
    
    def distance(self):  # You can add methods
        return (self.x**2 + self.y**2)**0.5

p2 = Point(3, 4)
print(p2.x, p2.y)  # 3 4
print(p2.distance())  # 5.0
````

Named tuples are more lightweight than classes and even dataclasses. They're ideal for small, immutable data structures.

Key differences between named tuples and dataclasses:
- Named tuples are always immutable; dataclasses can be mutable or immutable
- Named tuples are sequences (they can be accessed by index); dataclasses are not
- Named tuples have a more compact memory footprint
- Dataclasses offer more flexibility and customization options

#### Composition vs. Inheritance

When designing classes, you often need to decide between two main approaches to building complex objects:

1. **Inheritance**: "is-a" relationships where a subclass inherits from a superclass
2. **Composition**: "has-a" relationships where a class contains instances of other classes

Let's compare these approaches:

**Inheritance Example**:

````python
class Engine:
    def start(self):
        return "Engine started"
    
    def stop(self):
        return "Engine stopped"

class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model
        self.engine = Engine()
    
    def start(self):
        return f"{self.make} {self.model}: {self.engine.start()}"
    
    def stop(self):
        return f"{self.make} {self.model}: {self.engine.stop()}"

class Car(Vehicle):
    def __init__(self, make, model, num_doors):
        super().__init__(make, model)
        self.num_doors = num_doors
    
    def honk(self):
        return "Honk honk!"

# Using inheritance
car = Car("Toyota", "Camry", 4)
print(car.start())  # Toyota Camry: Engine started
print(car.honk())   # Honk honk!
````

**Composition Example**:


````python
class Engine:
    def start(self):
        return "Engine started"
    
    def stop(self):
        return "Engine stopped"

class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model
        self.engine = Engine()
    
    def start(self):
        return f"{self.make} {self.model}: {self.engine.start()}"
    
    def stop(self):
        return f"{self.make} {self.model}: {self.engine.stop()}"

class Horn:
    def honk(self):
        return "Honk honk!"

class Car:
    def __init__(self, make, model, num_doors):
        self.vehicle = Vehicle(make, model)
        self.horn = Horn()
        self.num_doors = num_doors
    
    def start(self):
        return self.vehicle.start()
    
    def stop(self):
        return self.vehicle.stop()
    
    def honk(self):
        return self.horn.honk()

# Using composition
car = Car("Toyota", "Camry", 4)
print(car.start())  # Toyota Camry: Engine started
print(car.honk())   # Honk honk!
````

Both approaches achieve similar results, but they have different tradeoffs:

**Inheritance Advantages**:
- Simpler when subclass truly "is a" specialized version of parent
- Automatically inherits implementation and interface
- Supports polymorphism through shared interfaces

**Inheritance Disadvantages**:
- Creates tight coupling between classes
- Can lead to complex hierarchies that are hard to understand
- Changes to the parent class can unexpectedly affect subclasses

**Composition Advantages**:
- More flexible and modular
- Classes are more focused on specific responsibilities
- Easier to change implementation details
- Avoids problems with deep inheritance hierarchies

**Composition Disadvantages**:
- May require more code to delegate methods
- Relationship between classes is less explicit

A common design principle is **"Favor composition over inheritance"**. This doesn't mean never use inheritance, but rather:
- Use inheritance for genuine "is-a" relationships
- Use composition for "has-a" relationships
- Prefer composition when in doubt, as it tends to be more flexible and maintainable

#### When to Use Each Pattern

Here's a quick guide for choosing the right approach for different scenarios:

**Use Dataclasses When**:
- You need a class primarily for storing data
- The class has many attributes
- You want automatic implementation of common special methods
- You need to define additional methods beyond just data storage

**Use Named Tuples When**:
- You need a lightweight, immutable data container
- You want to access elements by both name and position
- You don't need to define many methods
- Memory efficiency is important

**Use Inheritance When**:
- There's a clear "is-a" relationship (e.g., a Car is a Vehicle)
- The subclass is truly a specialized version of the parent
- You need polymorphic behavior through shared interfaces
- The inheritance hierarchy is shallow (not many levels deep)

**Use Composition When**:
- There's a "has-a" relationship (e.g., a Car has an Engine)
- You want to combine behaviors from multiple sources
- You need more flexibility to change implementations
- You want to avoid the complications of multiple inheritance

**General Guidelines**:
- Keep classes focused on a single responsibility
- Favor composition for most relationships
- Use the simplest approach that meets your needs
- Consider how your design will accommodate future changes

#### ⚠️ Common Mistakes with Modern Class Patterns

* **Overusing dataclasses**: Not every class needs to be a dataclass; use them specifically for data containers
* **Forgetting immutability**: Named tuples are always immutable; use a dataclass if you need mutability
* **Deep inheritance hierarchies**: Avoid creating complex inheritance structures; use composition instead
* **Using inheritance for code reuse**: Inheritance should model "is-a" relationships, not just for sharing code

### 🏆 Stress and Strain Calculator (Part 5/5)

In this coding challenge, we'll revisit our stress and strain calculator from previous chapters, but now we'll implement it using object-oriented programming principles. This will be the culmination of everything we've learned about classes and objects.

**Objective:**

Create a comprehensive material stress and strain analysis system using OOP principles. The system should:

1. Model different materials with their properties
2. Calculate stress, strain, and Young's modulus
3. Provide tools for comparing and analyzing materials
4. Implement proper validation and error handling
5. Use appropriate OOP patterns for a clean, maintainable design

**Requirements:**

1. **Create a Material Class Hierarchy:**
   - A base `Material` class with common properties and methods
   - Specific material types as subclasses (e.g., `Metal`, `Plastic`, `Composite`)
   - Appropriate validation for all inputs

2. **Implement Material Properties:**
   - Store essential properties like density, yield strength, and typical Young's modulus
   - Use properties (getters/setters) for controlled access to critical values
   - Include appropriate type hints for clarity

3. **Create a Stress-Strain Test Class:**
   - Model individual tests with force, area, original length, and change in length
   - Calculate stress, strain, and observed Young's modulus
   - Provide methods to evaluate if the material is likely to fail under the test conditions

4. **Implement Modern Class Patterns:**
   - Use dataclasses where appropriate for data-centric classes
   - Apply composition to create a flexible, modular design
   - Add useful string representations and comparisons using special methods

5. **Create a Test Analysis System:**
   - Ability to compare multiple materials and tests
   - Method to generate a summary report
   - Tools to visualize or describe the results in a meaningful way

**Getting Started:**

Here's a skeleton to help you begin:

````python
from dataclasses import dataclass
from typing import Optional, List

@dataclass
class MaterialProperties:
    """Properties of a material."""
    density: float  # kg/m³
    yield_strength: float  # MPa
    typical_youngs_modulus: float  # GPa
    
    def __post_init__(self):
        """Validate properties."""
        if self.density <= 0:
            raise ValueError("Density must be positive")
        if self.yield_strength <= 0:
            raise ValueError("Yield strength must be positive")
        if self.typical_youngs_modulus <= 0:
            raise ValueError("Young's modulus must be positive")

class Material:
    """Base class for all materials."""
    
    def __init__(self, name: str, properties: MaterialProperties):
        self.name = name
        self.properties = properties
    
    def __str__(self) -> str:
        return f"{self.name} (Density: {self.properties.density} kg/m³)"
    
    def can_withstand_stress(self, stress: float) -> bool:
        """Check if the material can withstand the given stress."""
        # Convert GPa to MPa for comparison
        return stress < self.properties.yield_strength

class StressStrainTest:
    """A single stress-strain test."""
    
    def __init__(self, material: Material, force: float, area: float, 
                 original_length: float, change_in_length: float):
        self.material = material
        self._force = force
        self._area = area
        self._original_length = original_length
        self._change_in_length = change_in_length
        
        # Validate inputs
        if force <= 0:
            raise ValueError("Force must be positive")
        if area <= 0:
            raise ValueError("Area must be positive")
        if original_length <= 0:
            raise ValueError("Original length must be positive")
        # Change in length can be negative (compression)
    
    @property
    def stress(self) -> float:
        """Calculate stress in MPa."""
        return self._force / self._area
    
    @property
    def strain(self) -> float:
        """Calculate strain (dimensionless)."""
        return self._change_in_length / self._original_length
    
    @property
    def youngs_modulus(self) -> float:
        """Calculate Young's modulus in GPa."""
        # Convert to GPa from MPa
        return (self.stress / self.strain) / 1000
    
    def will_fail(self) -> bool:
        """Determine if the material is likely to fail under this test."""
        return not self.material.can_withstand_stress(self.stress)
    
    def __str__(self) -> str:
        return (f"Test on {self.material.name}: "
                f"Stress={self.stress:.2f} MPa, "
                f"Strain={self.strain:.6f}, "
                f"Young's Modulus={self.youngs_modulus:.2f} GPa")

# Example material subclass
class Metal(Material):
    """A metal material."""
    
    def __init__(self, name: str, properties: MaterialProperties, 
                 is_ferrous: bool = False):
        super().__init__(name, properties)
        self.is_ferrous = is_ferrous
    
    def __str__(self) -> str:
        ferrous_text = "Ferrous" if self.is_ferrous else "Non-ferrous"
        return f"{self.name} ({ferrous_text} metal, Density: {self.properties.density} kg/m³)"

# Example usage
steel_properties = MaterialProperties(
    density=7850,
    yield_strength=250,  # MPa
    typical_youngs_modulus=200  # GPa
)

steel = Metal("Steel", steel_properties, is_ferrous=True)
test = StressStrainTest(steel, force=5000, area=25, original_length=100, change_in_length=0.5)

print(steel)
print(test)
print(f"Will the material fail? {'Yes' if test.will_fail() else 'No'}")
print(f"Calculated Young's modulus: {test.youngs_modulus:.2f} GPa")
print(f"Typical Young's modulus: {steel.properties.typical_youngs_modulus:.2f} GPa")
````
