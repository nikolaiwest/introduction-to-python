## **Chapter 1.1:** Introduction to Type Hinting

### 🔍 Understanding Type Systems

Programming languages are often categorized by their type systems, which govern how they handle data types. The main distinction is between **static typing** and **dynamic typing**:

**Static typing** means the type of every variable is known at compile time, before the program runs. Languages like Java, C++, and TypeScript require you to declare types upfront.

```java
// Java example - static typing
String name = "Alice";
int age = 30;
```

**Dynamic typing** means types are determined at runtime based on the values assigned to variables. Languages like Python, JavaScript, and Ruby don't require explicit type declarations.

````python
# Python example - dynamic typing
name = "Alice"  # Implicitly a string
age = 30        # Implicitly an integer
````


Python has always been a dynamically typed language, which contributes to its reputation for flexibility and ease of use. Variables in Python can change types throughout their lifetime:


````python
x = "hello"  # x is a string
x = 42       # Now x is an integer
x = [1, 2, 3]  # Now x is a list
````


This flexibility has advantages:
- Rapid development without type declaration overhead
- Adaptability to different data structures
- Concise code that focuses on logic rather than type details

But it also presents challenges:
- Type-related bugs might only appear at runtime
- Code can be harder to understand without type information
- Limited IDE assistance for autocompletion and refactoring
- Reduced confidence in code correctness, especially in larger codebases

**Gradual typing**, which Python's type hinting system implements, offers a middle ground. It allows you to add type information where it's most valuable while retaining the flexibility of dynamic typing.

### 🔍 Benefits of Type Hinting

Adding type hints to your Python code provides several important benefits:

**1. Improved code readability and self-documentation**
Type hints make your code's intent clearer, serving as built-in documentation about what kinds of data your functions expect and return.


````python
# Without type hints
def process_data(data, factor, use_cache):
    # What types should these parameters be?
    ...

# With type hints
def process_data(data: list[float], factor: float, use_cache: bool) -> list[float]:
    # Much clearer expectations!
    ...
````


**2. Better IDE support**
Modern IDEs and editors like PyCharm, Visual Studio Code, and others use type hints to:
- Provide more accurate autocompletion suggestions
- Offer better refactoring tools
- Show more informative documentation on hover

**3. Catching errors earlier**
Type checkers like mypy can catch potential type-related errors before you run your code:
- Passing incorrect argument types to functions
- Returning incorrect types from functions
- Performing invalid operations on incompatible types

**4. Easier maintenance**
Type hints make it easier to:
- Understand existing code when coming back to it months later
- Onboard new team members to a codebase
- Refactor with confidence, knowing the type checker will catch issues

**5. Gradual adoption**
You don't have to add type hints to your entire codebase at once. You can:
- Start with the most critical or complex parts
- Add type hints to new code while leaving legacy code as-is
- Increase type coverage over time

### 🔍 Type Hinting Evolution

Python's journey toward type hinting has been gradual and pragmatic:

**Python 3.0-3.4 (Function Annotations)**
Python 3.0 introduced the ability to add arbitrary annotations to function parameters and return values, but without a standard meaning:


````python
def greet(name: "Person's name", age: "Person's age") -> "Greeting message":
    return f"Hello {name}, you are {age} years old"
````


**Python 3.5 (PEP 484)**
PEP 484 introduced type hints, giving a standard meaning to annotations and introducing the `typing` module:


````python
from typing import List

def process_items(items: List[str]) -> None:
    for item in items:
        print(item)
````


**Python 3.6 (PEP 526)**
PEP 526 added syntax for variable annotations:


````python
name: str = "Alice"
age: int = 30
active: bool = True
````


**Python 3.7-3.9 (Typing Improvements)**
Subsequent Python versions added many improvements:
- Postponed evaluation of annotations (PEP 563)
- `typing.Protocol` for structural typing (PEP 544)
- `typing.TypedDict` for dictionaries with specific types
- Support for more standard collections

**Python 3.9+ (Simplified Syntax)**
Python 3.9+ allowed using built-in collection types directly in annotations:


````python
# Python 3.9+
def process_items(items: list[str]) -> None:
    ...

# Before Python 3.9
from typing import List
def process_items(items: List[str]) -> None:
    ...
````


**Future Direction**
Type hinting in Python continues to evolve, with ongoing improvements to the type system, better performance, and enhanced tooling.

#### ⚠️ Common Misconceptions about Type Hints

* **Type hints enforce types at runtime**: They don't. Python remains dynamically typed, and type hints are ignored during execution.
* **Type hints make Python code slower**: They don't affect runtime performance (though some overhead exists when reading annotations).
* **Type hints are mandatory**: They're completely optional. You can use as many or as few as you want.
* **Type hints eliminate all type-related bugs**: They help catch many issues, but runtime type checking is still your responsibility.

## **Chapter 1.2:** Basic Type Annotations

### 🔍 Variable Annotations

Variable annotations allow you to explicitly specify the expected type of a variable. This provides information to type checkers and IDEs without changing the runtime behavior of your code.

The basic syntax for variable annotations is:


````python
variable_name: type_annotation = value
````


Let's look at examples of annotating variables with Python's built-in types:


````python
# Basic type annotations
name: str = "Alice"
age: int = 30
height: float = 1.75
is_active: bool = True

# You can annotate a variable without assigning a value
coordinates: tuple[float, float]  # Will be assigned later
````


Variable annotations are particularly useful in class definitions, module-level variables, and to clarify complex or ambiguous types.


````python
class User:
    name: str
    email: str
    active: bool = True  # With default value
    login_count: int = 0
````


Note that annotating a variable doesn't initialize it. If you annotate without assignment, the variable doesn't exist yet:


````python
x: int  # Just an annotation, x doesn't exist yet
print(x)  # This raises a NameError
````


### 🔍 Function Annotations

Function annotations are the most common use of type hints in Python. They help document the expected types of function parameters and return values.

The basic syntax for function annotations is:


````python
def function_name(param1: type1, param2: type2) -> return_type:
    ...
````


Here are examples of function annotations with built-in types:


````python
# Simple function annotation
def greet(name: str) -> str:
    return f"Hello, {name}!"

# Multiple parameters with different types
def calculate_bmi(weight: float, height: float) -> float:
    return weight / (height ** 2)

# Function with no return value
def log_user_activity(user_id: int, action: str) -> None:
    print(f"User {user_id} performed {action}")
````


One significant benefit of function annotations is that they document your function's contract for callers. This helps prevent errors when using your functions:


````python
# Without type hints, it's unclear what types to pass
def process_data(data, factor):
    ...

# With type hints, it's clear what's expected
def process_data(data: list[float], factor: float) -> list[float]:
    ...
````


Default parameter values work alongside type hints:


````python
def create_user(
    name: str,
    email: str,
    active: bool = True,
    login_count: int = 0
) -> dict[str, object]:
    return {
        "name": name,
        "email": email,
        "active": active,
        "login_count": login_count
    }
````


### 🔍 Type Comments

Before PEP 484 introduced the annotation syntax, developers used type comments to add type information. While annotations are now preferred, you might encounter type comments in older code, especially code compatible with Python 2:


````python
# Variable type comment
age = 30  # type: int

# Function type comment
def greet(name):
    # type: (str) -> str
    return f"Hello, {name}!"

# Multiple parameters
def calculate_bmi(weight, height):
    # type: (float, float) -> float
    return weight / (height ** 2)
````


Type comments are still useful in some specific situations:
- For code that needs to remain compatible with Python 2
- When annotating variables in `for` loops or with complex assignments


````python
# Annotating a loop variable
for item in items:  # type: str
    print(item.upper())

# Annotating in complex assignments
a, b = 1, 2  # type: int, int
````


Type checkers like mypy recognize both annotation styles, but the annotation syntax is clearer and more maintainable for new code.

#### ⚠️ Common Mistakes with Basic Type Annotations

* **Annotating with instance instead of type**: `x: str = str()` (incorrect) vs `x: str = ""`
* **Forgetting the return type**: All functions should have a return type annotation, use `-> None` for functions that don't return a value
* **Using the wrong type for optional parameters**: Use `Optional[type]` or `type | None` for parameters that could be None
* **Incorrect annotation for variadic arguments**: `*args: int` only hints that each arg is an int, not that args is a tuple

## **Chapter 1.3:** Complex Type Hints

### 🔍 Collection Types

Python's `typing` module provides special types for collections that allow you to specify what types they contain. This gives much richer information than just `list` or `dict`.

Let's explore the various collection type annotations:

**Lists**

For lists, you can specify the type of elements they contain:


````python
# A list of strings
names: list[str] = ["Alice", "Bob", "Charlie"]

# A list of integers
numbers: list[int] = [1, 2, 3, 4, 5]

# An empty list intended to hold floats
measurements: list[float] = []
````


Before Python 3.9, you needed to import these types from the `typing` module:


````python
from typing import List

names: List[str] = ["Alice", "Bob", "Charlie"]
````


**Dictionaries**

For dictionaries, you specify both key and value types:


````python
# A dictionary mapping strings to integers
ages: dict[str, int] = {"Alice": 30, "Bob": 25}

# A dictionary mapping integers to boolean values
flags: dict[int, bool] = {1: True, 2: False, 3: True}

# More complex nested structures
user_stats: dict[str, dict[str, float]] = {
    "Alice": {"accuracy": 0.95, "speed": 42.0},
    "Bob": {"accuracy": 0.87, "speed": 38.5}
}
````


**Sets and Tuples**

Sets are similar to lists but contain unique elements:


````python
# A set of integers
unique_ids: set[int] = {1, 2, 3, 4, 5}
````


For tuples, you have two options:
1. When all elements have the same type: `tuple[type, ...]`
2. When elements have different types: `tuple[type1, type2, ...]`


````python
# A tuple of strings (any length)
names: tuple[str, ...] = ("Alice", "Bob", "Charlie")

# A tuple with specific types in a specific order
point: tuple[float, float] = (2.5, 3.7)  # x, y coordinates

# A tuple with three different types
user_info: tuple[str, int, bool] = ("Alice", 30, True)  # name, age, active
````


**Nested Collections**

You can nest these type annotations to represent complex data structures:


````python
# A list of lists of integers (a matrix)
matrix: list[list[int]] = [[1, 2], [3, 4]]

# A dictionary mapping strings to lists of strings
user_hobbies: dict[str, list[str]] = {
    "Alice": ["reading", "swimming"],
    "Bob": ["gaming", "painting"]
}
````


### 🔍 Optional and Union Types

Sometimes your variables can be of multiple possible types. Python's typing module provides tools for expressing these situations.

**Optional Types**

`Optional[T]` indicates a value that could be of type `T` or `None`:


````python
from typing import Optional

# A string that might be None
middle_name: Optional[str] = None

# A function that might return None
def find_user(user_id: int) -> Optional[dict[str, object]]:
    # Implementation...
    if user_exists:
        return user_data
    return None
````


Starting with Python 3.10, you can also use the more concise union syntax with `|`:


````python
# Python 3.10+ syntax
middle_name: str | None = None

def find_user(user_id: int) -> dict[str, object] | None:
    # Implementation...
````


**Union Types**

`Union[T1, T2, ...]` indicates a value that could be of any of the specified types:


````python
from typing import Union

# A variable that could be a string or an integer
identifier: Union[str, int] = "abc123"

# Later, it could be an integer
identifier = 42

# A function that accepts either strings or integers
def process_id(identifier: Union[str, int]) -> str:
    return str(identifier)
````


With Python 3.10+, you can use the `|` operator for unions:


````python
# Python 3.10+ syntax
identifier: str | int = "abc123"

def process_id(identifier: str | int) -> str:
    return str(identifier)
````


**Combining Unions and Collections**

You can combine these concepts for more complex type specifications:


````python
# A list that can contain both strings and integers
mixed_list: list[Union[str, int]] = ["apple", 42, "banana", 7]

# With Python 3.10+ syntax
mixed_list: list[str | int] = ["apple", 42, "banana", 7]

# A dictionary with string keys and values that could be strings, integers, or None
config: dict[str, Optional[Union[str, int]]] = {
    "name": "MyApp",
    "version": 2,
    "description": None
}

# With Python 3.10+ syntax
config: dict[str, str | int | None] = {
    "name": "MyApp",
    "version": 2,
    "description": None
}
````


### 🔍 Type Aliases

As type annotations get more complex, they can become verbose and repetitive. Type aliases let you define a name for a complex type, improving readability and maintainability.

The basic syntax is:


````python
from typing import TypeAlias

AliasName: TypeAlias = complex_type_definition
````


Here are examples of how type aliases can simplify your code:


````python
from typing import TypeAlias, Union, Dict, List, Optional

# Define type aliases
UserId: TypeAlias = Union[str, int]
UserData: TypeAlias = Dict[str, Union[str, int, bool]]
UsersList: TypeAlias = List[UserData]
OptionalUsersList: TypeAlias = Optional[UsersList]

# Using the aliases
def get_user(user_id: UserId) -> Optional[UserData]:
    # Implementation...
    pass

def get_department_users(dept_id: str) -> UsersList:
    # Implementation...
    pass
````


With Python 3.10+, you can use the simpler syntax:


````python
# Python 3.10+ syntax
UserId = str | int
UserData = dict[str, str | int | bool]
UsersList = list[UserData]
OptionalUsersList = UsersList | None
````


Type aliases are particularly useful for:
- Complex collection types
- Types used repeatedly throughout your codebase
- Domain-specific type definitions

They also improve maintenance: when you need to change a type definition, you only need to change it in one place.

#### ⚠️ Common Mistakes with Complex Type Annotations

* **Using `List`, `Dict`, etc. without type parameters**: `List` without `[T]` is equivalent to `list[Any]`, which loses most type checking benefits
* **Using Optional incorrectly**: `Optional[T]` means `T | None`, not "an optional parameter"
* **Confusing `Optional` with default values**: `def f(x: int = None)` should be `def f(x: Optional[int] = None)`
* **Complex nested types that are hard to read**: Consider using type aliases for readability

## **Chapter 1.4:** User-Defined and Generic Types

### 🔍 Class Type Annotations

So far, we've focused on built-in types, but real-world Python code often involves custom classes. Type annotations work seamlessly with your own classes.

**Basic Class Annotations**

Using a class as a type annotation is straightforward:


````python
class User:
    def __init__(self, name: str, email: str, active: bool = True):
        self.name = name
        self.email = email
        self.active = active

# Function that takes and returns a User object
def deactivate_user(user: User) -> User:
    user.active = False
    return user

# Create and use a User
alice: User = User("Alice", "alice@example.com")
inactive_alice = deactivate_user(alice)
````


**Forward References**

Sometimes you need to refer to a class before it's defined. Python allows string literals in annotations for this purpose:


````python
# Using a forward reference
def get_node_parent(node: "TreeNode") -> "TreeNode":
    return node.parent

class TreeNode:
    def __init__(self, value: str, parent: "TreeNode" = None):
        self.value = value
        self.parent = parent
        self.children: list["TreeNode"] = []
````


With Python 3.7+, you can also use `from __future__ import annotations` at the top of your file to postpone evaluation of all annotations:


````python
from __future__ import annotations

# No quotes needed with postponed evaluation
def get_node_parent(node: TreeNode) -> TreeNode:
    return node.parent

class TreeNode:
    def __init__(self, value: str, parent: TreeNode = None):
        self.value = value
        self.parent = parent
        self.children: list[TreeNode] = []
````


**Annotating Methods**

Methods in classes follow the same rules as functions, but remember to include `self` (without a type annotation):


````python
class Rectangle:
    def __init__(self, width: float, height: float):
        self.width = width
        self.height = height

    def area(self) -> float:
        return self.width * self.height

    def scale(self, factor: float) -> None:
        self.width *= factor
        self.height *= factor
````


**Type Checking With Classes**

When type checking with classes, subclass relationships are respected:


````python
class Animal:
    def make_sound(self) -> str:
        return "..."

class Dog(Animal):
    def make_sound(self) -> str:
        return "Woof!"

# Function expecting an Animal
def get_animal_sound(animal: Animal) -> str:
    return animal.make_sound()

# Works fine - Dog is a subclass of Animal
fido: Dog = Dog()
sound = get_animal_sound(fido)
````


### 🔍 Generic Types

Generic types allow you to create classes and functions that work with a variety of types while maintaining type safety. They're similar to templates in C++ or generics in Java.

**Creating Generic Classes**

To create a generic class, you use `TypeVar` from the `typing` module:


````python
from typing import TypeVar, Generic, List

# Define a type variable
T = TypeVar('T')

# Generic class that can hold any type
class Box(Generic[T]):
    def __init__(self, content: T):
        self.content = content

    def get_content(self) -> T:
        return self.content

    def replace_content(self, new_content: T) -> None:
        self.content = new_content

# Usage with different types
int_box: Box[int] = Box(42)
str_box: Box[str] = Box("Hello")

# Type checker ensures correct usage
x: int = int_box.get_content()  # OK
y: str = str_box.get_content()  # OK
# z: int = str_box.get_content()  # Type error!
````


**Multiple Type Variables**

You can use multiple type variables for more complex generic types:


````python
from typing import TypeVar, Generic, Dict

K = TypeVar('K')
V = TypeVar('V')

class Mapping(Generic[K, V]):
    def __init__(self):
        self.items: Dict[K, V] = {}

    def set_item(self, key: K, value: V) -> None:
        self.items[key] = value

    def get_item(self, key: K) -> V:
        return self.items[key]

# Usage with specific types
user_scores: Mapping[str, int] = Mapping()
user_scores.set_item("Alice", 42)
score: int = user_scores.get_item("Alice")
````


**Bounded Type Variables**

You can restrict the types that can be used with a type variable:


````python
from typing import TypeVar, List

# T can only be a str or a bytes
T = TypeVar('T', str, bytes)

def process_text(text: T) -> T:
    # Process and return the same type
    return text

# U must be a subclass of Exception
U = TypeVar('U', bound=Exception)

def handle_exception(ex: U) -> U:
    # Log and return the exception
    print(f"Handling {type(ex).__name__}: {ex}")
    return ex
````


**Generic Functions**

You can also create generic functions without defining a class:


````python
from typing import TypeVar, List, Sequence

T = TypeVar('T')

def first_element(seq: Sequence[T]) -> T:
    if not seq:
        raise ValueError("Sequence is empty")
    return seq[0]

# Usage
x: int = first_element([1, 2, 3])  # x is inferred as int
y: str = first_element(["a", "b", "c"])  # y is inferred as str
````


### 🔍 TypeVar and Protocols

Python's type system includes additional advanced features that provide more flexibility and expressiveness.

**Using TypeVar with Constraints**

Beyond basic generics, `TypeVar` can have constraints and bounds:


````python
from typing import TypeVar, List, Union

# T can only be str or bytes
StringLike = TypeVar('StringLike', str, bytes)

def to_upper(text: StringLike) -> StringLike:
    # Works with either str or bytes
    if isinstance(text, str):
        return text.upper()
    else:  # bytes
        return text.upper()

# U must be a subclass of Exception
ExceptionType = TypeVar('ExceptionType', bound=Exception)

def log_exception(ex: ExceptionType) -> ExceptionType:
    print(f"Exception: {ex}")
    return ex
````


**Protocols and Structural Typing**

Protocols provide a way to define interfaces and enable structural typing (focusing on what methods an object has, rather than its class):


````python
from typing import Protocol, List, Iterator

# Define a protocol (interface)
class Drawable(Protocol):
    def draw(self) -> None:
        ...

# Classes that implement the protocol (implicitly)
class Circle:
    def __init__(self, radius: float):
        self.radius = radius

    def draw(self) -> None:
        print(f"Drawing circle with radius {self.radius}")

class Square:
    def __init__(self, side: float):
        self.side = side

    def draw(self) -> None:
        print(f"Drawing square with side {self.side}")

# Function that accepts any Drawable
def draw_shapes(shapes: List[Drawable]) -> None:
    for shape in shapes:
        shape.draw()

# Usage - both Circle and Square work, even though
# they don't explicitly inherit from Drawable
shapes: List[Drawable] = [Circle(5.0), Square(4.0)]
draw_shapes(shapes)
````


**Runtime Checkable Protocols**

With `@runtime_checkable`, you can check protocol compliance at runtime:


````python
from typing import Protocol, runtime_checkable, List

@runtime_checkable
class Sized(Protocol):
    def __len__(self) -> int:
        ...

def process_sized_items(items: List[object]) -> List[int]:
    result = []
    for item in items:
        if isinstance(item, Sized):  # Runtime check
            result.append(len(item))
    return result

# Usage
lengths = process_sized_items(["hello", [1, 2, 3], 42, {1, 2}])
print(lengths)  # [5, 3, 2] (42 is not Sized)
````


#### ⚠️ Common Mistakes with User-Defined and Generic Types

* **Overly restricting generic types**: Too many constraints can make your generic types less useful
* **Forgetting forward references**: When classes reference themselves or each other
* **Using class instance instead of class**: `x: User()` instead of the correct `x: User = User()`
* **Not considering inheritance**: Functions should typically accept the most general type possible

## **Chapter 1.5:** Static Type Checking

### 🔍 Introduction to Mypy

Python's type annotations by themselves don't enforce type correctness—they're just hints. To actually check your code for type errors, you need a static type checker. **Mypy** is the most popular type checker for Python, created by the same people who developed Python's typing system.

**What Mypy Does**

Mypy analyzes your Python code without running it and reports potential type errors:
- Arguments with incorrect types
- Return values that don't match annotations
- Incompatible variable assignments
- Accessing attributes or methods that don't exist
- And many other potential issues

**Installing Mypy**

You can install mypy using pip:

```bash
pip install mypy
```

**Basic Usage**

To check a Python file:

```bash
mypy your_file.py
```

For a more complete analysis, you can check an entire package:

```bash
mypy your_package/
```

**Example of Mypy Errors**

Consider this code with type annotations:


````python
# typed_example.py
def greet(name: str) -> str:
    return f"Hello, {name}!"

def process_age(age: int) -> str:
    return f"You are {age} years old."

# Type error: greet expects a str but gets an int
greeting = greet(42)

# Type error: process_age expects an int but gets a str
result = process_age("twenty")
````


Running mypy on this file:

```bash
$ mypy typed_example.py

typed_example.py:11: error: Argument 1 to "process_age" has incompatible type "str"; expected "int"
```

Mypy catches both errors without executing the code, helping you fix issues before they cause runtime errors.

**Configuration**

You can configure mypy using a `mypy.ini` or `setup.cfg` file. This allows you to customize the type checking behavior for your project:

```ini

```

Common configuration options include:
- `disallow_untyped_defs`: Require all functions to have type annotations
- `disallow_incomplete_defs`: Require all function parameters to be annotated

- `warn_redundant_casts`: Warn about unnecessary type casts
- `warn_unused_ignores`: Warn about unused `# type: ignore` comments

### 🔍 Finding and Fixing Type Errors

When mypy reports errors, you'll need to understand and fix them. Here are common error types and their solutions:

**Incompatible Type Arguments**


````python
def process_items(items: list[str]) -> None:
    ...

# Error: Argument 1 to "process_items" has incompatible type "List[int]"
process_items([1, 2, 3])
````


Solutions:
1. Change the function call to provide the expected type: `process_items(["1", "2", "3"])`
2. Change the function signature if it should accept more types: `def process_items(items: list[Union[str, int]]) -> None:`
3. Create a converter function to transform the input: `process_items([str(x) for x in [1, 2, 3]])`

**Missing Return Annotations**


````python
# Error: Function is missing a return type annotation
def calculate_total(prices):
    return sum(prices)
````


Solution: Add a return type annotation:


````python
def calculate_total(prices: list[float]) -> float:
    return sum(prices)
````


**Incompatible Return Type**


````python
def get_user_id() -> int:
    user_id = get_id_from_database()
    return user_id  # Error if user_id could be None
````


Solutions:
1. Make the return type more accurate: `def get_user_id() -> Optional[int]:`
2. Add validation to ensure the correct type is returned:

````python
   def get_user_id() -> int:
       user_id = get_id_from_database()
       if user_id is None:
           return 0  # Default value
       return user_id
````


**Optional Types and None Checks**


````python
def process_name(name: Optional[str]) -> str:
    return name.upper()  # Error: Item "None" of "Optional[str]" has no attribute "upper"
````


Solution: Add a None check:


````python
def process_name(name: Optional[str]) -> str:
    if name is None:
        return ""
    return name.upper()  # Now safe
````


**Dealing with Dynamic Code**

Some Python patterns are hard to type correctly. For these cases, you can use `# type: ignore`:


````python
# A dynamic attribute access that mypy can't verify
getattr(obj, dynamic_attribute)()  # type: ignore
````


Use these sparingly and include comments explaining why the type checker should ignore the line.


Type stubs provide type information without modifying the original code.

#### ⚠️ Common Mistakes with Static Type Checking

* **Overreliance on `# type: ignore`**: Use sparingly and with explanatory comments
* **Forgetting that runtime behavior still matters**: Type checking happens before runtime
* **Making code overly complex to please the type checker**: Sometimes a simple cast or ignore comment is better
* **Not checking your entire codebase**: Isolated type checking can miss cross-module issues

## 📚 Key Terms

- **Type Hint/Annotation**: Syntax for specifying the expected type of a variable, parameter, or return value
- **Static Typing**: A typing discipline where variable types are checked at compile time
- **Dynamic Typing**: A typing discipline where variable types are checked at runtime
- **Gradual Typing**: A typing approach that combines static and dynamic typing features
- **Type Variable (TypeVar)**: A placeholder for a type that can be filled in later
- **Generic Type**: A type that can work with different data types while maintaining type safety
- **Optional Type**: A type that can be either a specified type or None
- **Union Type**: A type that can be any one of several specified types
- **Protocol**: An interface definition focusing on what methods an object has rather than its class hierarchy
- **Type Alias**: A name for a complex type definition
- **Mypy**: A static type checker for Python that verifies type hint correctness
