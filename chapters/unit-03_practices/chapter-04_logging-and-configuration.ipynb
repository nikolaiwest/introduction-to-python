## **Chapter 4.1:** Introduction to Logging

### 🔍 Why Use Logging Instead of Print

When you're first learning to program, the `print()` function is often your go-to tool for understanding what's happening in your code. Want to see a variable's value? `print(variable)`. Need to confirm a function was called? `print("Function called")`. While `print()` statements are quick and straightforward, they become problematic as your projects grow in complexity.

Let's examine why professional developers use dedicated logging systems instead of `print()` statements:

| Feature | print() | logging module |
|---------|---------|----------------|
| Severity levels | No built-in way to categorize | DEBUG, INFO, WARNING, ERROR, CRITICAL levels |
| Output control | Appears in stdout; difficult to redirect | Can send to files, network, email, etc. |
| Runtime control | Always executes | Can enable/disable based on configuration |
| Formatting | Manual formatting required | Consistent formatting with timestamps, levels, etc. |
| Persistence | Not persisted unless manually redirected | Easily saved to files for later analysis |
| Context | No built-in context | Includes timestamp, module name, line numbers |
| Performance | Can impact performance in tight loops | Can be optimized and disabled in production |

Consider this simple example of the difference between `print()` and logging:


````python
# Using print
print("Starting the function")
result = perform_calculation(data)
if result < 0:
    print("Error: Got a negative result")
print("Function completed")

# Using logging
import logging
logging.info("Starting the function")
result = perform_calculation(data)
if result < 0:
    logging.error("Got a negative result: %s", result)
logging.info("Function completed")
````


While they might look similar, the logging version provides several advantages:
- It categorizes messages by severity (info vs. error)
- It can include timestamps and source information automatically
- It can be easily configured to write to a file instead of the console
- It can be enabled/disabled without changing the code

### 🔍 Logging Levels

The Python logging module defines five standard levels of severity, allowing you to categorize messages based on their importance:

| Level | Numeric Value | Function | When to Use |
|-------|--------------|----------|-------------|
| DEBUG | 10 | logging.debug() | Detailed information for diagnosing problems |
| INFO | 20 | logging.info() | Confirmation that things are working as expected |
| WARNING | 30 | logging.warning() | Indication that something unexpected happened, but the program is still working |
| ERROR | 40 | logging.error() | Due to a more serious problem, the software couldn't perform some function |
| CRITICAL | 50 | logging.critical() | A very serious error, indicating the program may be unable to continue running |

By default, the logging system is configured to display only WARNING and above levels. This means DEBUG and INFO messages won't appear unless you explicitly configure the system to show them.

Here's an example demonstrating all five levels:


````python
import logging

logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
````


If you run this code with the default configuration, you'll only see:

```
WARNING:root:This is a warning message
ERROR:root:This is an error message
CRITICAL:root:This is a critical message
```

The DEBUG and INFO messages are suppressed because they're below the default threshold.

### 🔍 Basic Logging Setup

To start using the logging module effectively, you typically need to configure it first. The most basic setup involves setting the logging level:


````python
import logging

# Set the logging level to INFO, so DEBUG messages will be suppressed
logging.basicConfig(level=logging.INFO)

logging.debug("This won't be displayed")
logging.info("This will be displayed")
logging.warning("This will also be displayed")
````


The `basicConfig()` function allows you to configure several aspects of the logging system:


````python
import logging

# Configure logging to write to a file, include timestamps, and show INFO level and above
logging.basicConfig(
    filename='app.log',           # Write logs to this file
    filemode='w',                 # 'w' to overwrite, 'a' to append
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',  # Message format
    datefmt='%Y-%m-%d %H:%M:%S',  # Date/time format
    level=logging.INFO            # Show INFO and above
)

logging.info("Application started")
````


This will create a log file named 'app.log' with entries that look like:

```
2023-08-15 14:30:12 - root - INFO - Application started
```

#### ⚠️ Common Mistakes with Basic Logging

* **Forgetting that basicConfig() only works the first time it's called**: After the logging system is configured, subsequent calls to basicConfig() have no effect. This can be confusing if you're trying to change the configuration later in your program.

* **Not setting a logging level**: If you don't specify a level in basicConfig(), the default is WARNING, which means your DEBUG and INFO messages won't be displayed.

* **Using % formatting incorrectly**: When including variables in log messages, use the built-in string formatting capabilities of the logging functions rather than formatting the string yourself:


````python
# Incorrect (less efficient):
name = "Alice"
logging.info("User " + name + " logged in")  # String concatenation happens regardless of log level

# Correct (more efficient):
logging.info("User %s logged in", name)  # String formatting only happens if the message is logged
````

### 📘 Exploring the Logging Module Further

Let's look at some additional aspects of basic logging:

#### Using String Formatting in Log Messages

The logging functions support string formatting similar to the `%` operator:


````python
user_id = 12345
items_count = 5
logging.info("User %s purchased %d items", user_id, items_count)
````


This is more efficient than using string concatenation or f-strings because the string formatting only occurs if the message will actually be logged.

#### Logging Variables and Data Structures

When debugging complex data structures, you can include them in your log messages:


````python
user_data = {
    "id": 12345,
    "name": "Alice",
    "permissions": ["read", "write"]
}
logging.debug("Processing user data: %s", user_data)
````


For more complex or large data structures, consider using the `pprint` module for better formatting:


````python
import logging
from pprint import pformat

complex_data = {"users": [{"id": 1, "name": "Alice"}, {"id": 2, "name": "Bob"}], "status": "active"}
logging.debug("Complex data:\n%s", pformat(complex_data))
````


#### Conditional Logging

Sometimes you might want to log only under certain conditions. While you could use an if statement, the logging module provides a more elegant solution:


````python
# Instead of:
if debug_enabled:
    logging.debug("Complex calculation result: %s", calculate_result())

# You can use:
logging.debug("Complex calculation result: %s", calculate_result() if debug_enabled else "skipped")
````


For even better performance, especially if the calculation is expensive, use the `isEnabledFor` method:


````python
if logging.getLogger().isEnabledFor(logging.DEBUG):
    result = calculate_result()  # Only performed if DEBUG logging is enabled
    logging.debug("Complex calculation result: %s", result)
````

## **Chapter 4.2:** Configuring the Logging System

### 🔍 Logging Components

To use Python's logging system effectively, it's important to understand its core components and how they work together. The logging system is built around four main components:

1. **Loggers**: The main entry points for logging events in your application
2. **Handlers**: Determine where log messages go (console, file, network, etc.)
3. **Formatters**: Define how log messages are displayed or structured
4. **Filters**: Provide additional control over which log messages are output

Let's examine each of these components in detail:

#### Loggers

Loggers are the objects you interact with directly in your code. Each logger has:
- A name (typically the module name)
- A severity level
- Handlers that process its messages
- (Optional) parent-child relationships with other loggers

Here's how to create and use a logger:


````python
import logging

# Create a logger
logger = logging.getLogger('my_application')

# Set its level
logger.setLevel(logging.INFO)

# Use the logger
logger.debug("This won't be processed due to level setting")
logger.info("This will be processed")
````


The root logger is the ancestor of all loggers. When you use the module-level functions like `logging.info()`, you're actually using the root logger:


````python
# These two lines are equivalent
logging.info("Using the root logger")
logging.getLogger().info("Also using the root logger")
````


#### Handlers

Handlers determine where log messages are sent. Common destinations include:
- Console (StreamHandler)
- Files (FileHandler)
- Rotating files (RotatingFileHandler)
- Timed rotating files (TimedRotatingFileHandler)
- Email (SMTPHandler)
- HTTP (HTTPHandler)

A logger can have multiple handlers, allowing you to send the same log messages to different destinations:


````python
import logging
import sys

# Create a logger
logger = logging.getLogger('my_app')
logger.setLevel(logging.DEBUG)

# Create handlers
console_handler = logging.StreamHandler(sys.stdout)
console_handler.setLevel(logging.INFO)  # Only INFO and above to console

file_handler = logging.FileHandler('debug.log')
file_handler.setLevel(logging.DEBUG)    # All messages to file

# Add handlers to logger
logger.addHandler(console_handler)
logger.addHandler(file_handler)

# Test
logger.debug("This goes only to the file")
logger.info("This goes to both console and file")
logger.error("This also goes to both")
````


Each handler can have its own level, allowing you to filter messages at both the logger and handler levels.

#### Formatters

Formatters determine how log messages are presented. They convert the LogRecord (the internal representation of a log event) into a string or other format:


````python
import logging

# Create a formatter
formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# Create a handler and set the formatter
handler = logging.StreamHandler()
handler.setFormatter(formatter)

# Create a logger and add the handler
logger = logging.getLogger('formatter_example')
logger.setLevel(logging.INFO)
logger.addHandler(handler)

# Log a message
logger.info("This message will be formatted nicely")
````


The output will look something like:
```
2023-08-15 15:30:45,123 - formatter_example - INFO - This message will be formatted nicely
```

Formatters use a string with placeholders for various log record attributes. Common attributes include:

| Format | Description | Example |
|--------|-------------|---------|
| %(asctime)s | Timestamp | 2023-08-15 15:30:45,123 |
| %(name)s | Logger name | main |
| %(levelname)s | Level name | INFO |
| %(levelno)d | Level number | 20 |
| %(pathname)s | Full path to source file | /home/user/app.py |
| %(filename)s | Filename | app.py |
| %(module)s | Module name | app |
| %(funcName)s | Function name | process_data |
| %(lineno)d | Line number | 42 |
| %(process)d | Process ID | 12345 |
| %(thread)d | Thread ID | 67890 |
| %(message)s | The log message | User logged in |

#### Filters

Filters provide more sophisticated control over which log records are output. While levels provide a simple threshold-based filtering, custom filters can implement more complex logic:


````python
import logging

class SpecificModuleFilter(logging.Filter):
    """Only allow logs from a specific module."""

    def __init__(self, module_name):
        super().__init__()
        self.module_name = module_name

    def filter(self, record):
        return record.module == self.module_name

# Create logger, handler, etc.
logger = logging.getLogger('filtered_logger')
logger.setLevel(logging.DEBUG)

handler = logging.StreamHandler()
handler.setLevel(logging.DEBUG)

# Create and add the filter to the handler
module_filter = SpecificModuleFilter('my_module')
handler.addFilter(module_filter)

logger.addHandler(handler)
````


This filter will only allow log records from the 'my_module' module to be processed.

### 🔍 Configuring Handlers

Let's dive deeper into handlers, which are crucial for directing your log messages to appropriate destinations.

#### Console Handler (StreamHandler)

The StreamHandler sends output to streams such as sys.stdout or sys.stderr:


````python
import logging
import sys

logger = logging.getLogger('console_example')
logger.setLevel(logging.INFO)

# Create a handler that writes to stdout
stdout_handler = logging.StreamHandler(sys.stdout)
stdout_handler.setLevel(logging.INFO)

# Create a formatter and add it to the handler
formatter = logging.Formatter('%(levelname)s: %(message)s')
stdout_handler.setFormatter(formatter)

# Add the handler to the logger
logger.addHandler(stdout_handler)

# Test
logger.info("This message goes to stdout")
````


#### File Handler

The FileHandler writes log messages to a specified file:


````python
import logging

logger = logging.getLogger('file_example')
logger.setLevel(logging.DEBUG)

# Create a file handler that writes to a file
file_handler = logging.FileHandler('app.log')
file_handler.setLevel(logging.DEBUG)

# Create a formatter and add it to the handler
formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
file_handler.setFormatter(formatter)

# Add the handler to the logger
logger.addHandler(file_handler)

# Test
logger.debug("This debug message is saved to the file")
logger.info("This info message is also saved to the file")
````


#### Multiple Handlers

You can add multiple handlers to a logger to send logs to different destinations:


````python
import logging
import sys

logger = logging.getLogger('multi_handler')
logger.setLevel(logging.DEBUG)

# Console handler for INFO and above
console_handler = logging.StreamHandler(sys.stdout)
console_handler.setLevel(logging.INFO)
console_formatter = logging.Formatter('%(levelname)s: %(message)s')
console_handler.setFormatter(console_formatter)

# File handler for DEBUG and above
file_handler = logging.FileHandler('debug.log')
file_handler.setLevel(logging.DEBUG)
file_formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
file_handler.setFormatter(file_formatter)

# Add both handlers to the logger
logger.addHandler(console_handler)
logger.addHandler(file_handler)

# Test - notice how messages are filtered differently for each handler
logger.debug("Debug message - file only")
logger.info("Info message - console and file")
logger.warning("Warning message - console and file")
````


This setup sends:
- DEBUG level messages only to the file
- INFO and above to both console and file

#### NullHandler

The NullHandler is a special handler that does nothing with log events. It's useful for library code:


````python
import logging

# In a library module
logger = logging.getLogger('my_library')
logger.addHandler(logging.NullHandler())

def my_function():
    logger.info("This won't be displayed unless the application configures logging")
````


This pattern allows libraries to use logging internally without producing output unless the application configures logging.

#### ⚠️ Common Mistakes with Handlers

* **Forgetting to add handlers**: If you create a logger but don't add any handlers, log messages may not appear where you expect. The messages might be propagated up to parent loggers, depending on your configuration.

* **Not setting a level on handlers**: Each handler can have its own level, separate from the logger level. If you don't set a handler level, it defaults to NOTSET, which means it processes all messages that reach it.

* **Adding handlers multiple times**: In long-running applications or during development with repeated executions, you might accidentally add the same handler type multiple times to a logger, resulting in duplicate log messages.

### 🔍 Creating Custom Formatters

While the default formatter is flexible, you might want to create custom formatters for specific needs:

#### Advanced Formatting Options

You can use more sophisticated formatting with the Formatter class:


````python
import logging
import time

# Custom formatter with milliseconds and custom date format
formatter = logging.Formatter(
    fmt='%(asctime)s.%(msecs)03d - %(levelname)8s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

# Create handler and logger
handler = logging.StreamHandler()
handler.setFormatter(formatter)

logger = logging.getLogger('custom_format')
logger.setLevel(logging.INFO)
logger.addHandler(handler)

# Test
logger.info("This message has custom formatting")
````


This produces output like:


#### JSON Formatter

For applications that aggregate logs in systems like ELK (Elasticsearch, Logstash, Kibana) or other log analysis tools, JSON formatting is often preferred:


````python
import logging
import json
import datetime

class JsonFormatter(logging.Formatter):
    """Format log records as JSON strings."""

    def format(self, record):
        # Create a dictionary with log record attributes
        log_record = {
            "timestamp": datetime.datetime.fromtimestamp(record.created).isoformat(),
            "level": record.levelname,
            "name": record.name,
            "message": record.getMessage(),
            "module": record.module,
            "function": record.funcName,
            "line": record.lineno
        }

        # Add exception info if present
        if record.exc_info:
            log_record["exception"] = self.formatException(record.exc_info)

        return json.dumps(log_record)

# Use the custom formatter
handler = logging.StreamHandler()
handler.setFormatter(JsonFormatter())

logger = logging.getLogger('json_logger')
logger.setLevel(logging.INFO)
logger.addHandler(handler)

# Test
logger.info("This message will be formatted as JSON")
try:
    1/0
except ZeroDivisionError:
    logger.error("Division by zero", exc_info=True)
````


This produces output like:
```
{"timestamp": "2023-08-15T15:50:30.123456", "level": "INFO", "name": "json_logger", "message": "This message will be formatted as JSON", "module": "__main__", "function": "<module>", "line": 25}
{"timestamp": "2023-08-15T15:50:30.124789", "level": "ERROR", "name": "json_logger", "message": "Division by zero", "module": "__main__", "function": "<module>", "line": 28, "exception": "Traceback (most recent call last):\n  File \"<stdin>\", line 1, in <module>\nZeroDivisionError: division by zero"}
```

#### Custom Formatter with Colors

For console output, colorized log messages can improve readability:


````python
import logging
import sys

class ColoredFormatter(logging.Formatter):
    """Add colors to log messages based on level."""

    COLORS = {
        'DEBUG': '\033[94m',     # Blue
        'INFO': '\033[92m',      # Green
        'WARNING': '\033[93m',   # Yellow
        'ERROR': '\033[91m',     # Red
        'CRITICAL': '\033[91m\033[1m',  # Bold Red
        'RESET': '\033[0m'       # Reset
    }

    def format(self, record):
        # Format the message using the base formatter
        formatted_message = super().format(record)

        # Add color codes if the level has a color
        color = self.COLORS.get(record.levelname, self.COLORS['RESET'])
        return f"{color}{formatted_message}{self.COLORS['RESET']}"

# Create a formatter with colors
formatter = ColoredFormatter('%(levelname)s: %(message)s')

# Create a handler and set the formatter
handler = logging.StreamHandler(sys.stdout)
handler.setFormatter(formatter)

# Create a logger and add the handler
logger = logging.getLogger('colored_logs')
logger.setLevel(logging.DEBUG)
logger.addHandler(handler)

# Test all levels
logger.debug("This is a debug message")
logger.info("This is an info message")
logger.warning("This is a warning message")
logger.error("This is an error message")
logger.critical("This is a critical message")
````


Note: Colored output will only work in terminals that support ANSI color codes. It won't work in all environments.

### 🔍 Using Filters

Filters provide fine-grained control over which log records are processed. Let's look at some practical examples:

#### Filtering by Content

This filter allows only log messages containing specific text:


````python
import logging

class KeywordFilter(logging.Filter):
    """Only allow logs that contain a specified keyword."""

    def __init__(self, keyword):
        super().__init__()
        self.keyword = keyword

    def filter(self, record):
        return self.keyword in record.getMessage()

# Set up logger and handler
logger = logging.getLogger('keyword_filter')
logger.setLevel(logging.DEBUG)

handler = logging.StreamHandler()
handler.setLevel(logging.DEBUG)

# Add the filter to the handler
keyword_filter = KeywordFilter('user')
handler.addFilter(keyword_filter)

logger.addHandler(handler)

# Test
logger.info("Processing user data")  # Will be shown (contains 'user')
logger.info("System startup complete")  # Will not be shown (no 'user')
````


#### Filtering by Frequency

This filter limits how often similar log messages are output to prevent log flooding:


````python
import logging
import time

class RateLimitFilter(logging.Filter):
    """Limit the rate of similar log messages."""

    def __init__(self, rate=1):
        super().__init__()
        self.rate = rate  # Maximum messages per second
        self.last_log = {}
        self.log_counts = {}

    def filter(self, record):
        now = time.time()
        msg = record.getMessage()

        # If we've never seen this message or it's been more than 1/rate seconds
        if msg not in self.last_log or now - self.last_log[msg] >= 1/self.rate:
            self.last_log[msg] = now
            if msg in self.log_counts:
                record.message = f"{msg} (repeated {self.log_counts[msg]} times)"
                self.log_counts[msg] = 0
            return True
        else:
            # Count this message for later
            self.log_counts[msg] = self.log_counts.get(msg, 0) + 1
            return False

# Set up logger with rate limiting
logger = logging.getLogger('rate_limited')
logger.setLevel(logging.INFO)

handler = logging.StreamHandler()
handler.addFilter(RateLimitFilter(rate=0.5))  # Allow only one message per 2 seconds

formatter = logging.Formatter('%(asctime)s - %(message)s')
handler.setFormatter(formatter)
logger.addHandler(handler)

# Test with a loop
for i in range(10):
    logger.info("This is a repeating message")
    time.sleep(0.1)  # Sleep for 100ms between messages
````


This filter will limit the "This is a repeating message" log to appear at most once every 2 seconds, and will append a count of repeated messages.

### 🔍 Hierarchical Logging

One of the most powerful features of Python's logging system is its support for hierarchical loggers. This allows you to organize loggers in a parent-child relationship, creating a tree-like structure that mirrors your application's module organization.

#### Logger Hierarchy Basics

Loggers are organized in a hierarchy based on their names, with periods (.) separating levels:

```
root
├── app
│   ├── app.models
│   │   ├── app.models.user
│   │   └── app.models.product
│   └── app.views
└── database
    └── database.connection
```

In this hierarchy:
- The root logger is the parent of all loggers
- 'app' is a child of root and parent of 'app.models' and 'app.views'
- 'app.models' is a parent of 'app.models.user' and 'app.models.product'

#### Property Inheritance

Child loggers inherit certain properties from their parents:

1. **Level inheritance**: If a logger doesn't have a level explicitly set, it inherits its parent's level
2. **Handler propagation**: By default, log records are passed to the handlers of ancestor loggers

Here's an example of how this works:


````python
import logging

# Create parent logger
parent_logger = logging.getLogger('app')
parent_logger.setLevel(logging.WARNING)

# Add a handler to the parent
parent_handler = logging.StreamHandler()
parent_formatter = logging.Formatter('%(name)s - %(levelname)s - %(message)s')
parent_handler.setFormatter(parent_formatter)
parent_logger.addHandler(parent_handler)

# Create child logger (no handlers added directly to it)
child_logger = logging.getLogger('app.module')
# Note: We don't set a level, so it inherits from parent

# Log messages using both loggers
parent_logger.info("This won't be displayed (INFO < WARNING)")
parent_logger.warning("Parent warning")

child_logger.info("This also won't be displayed")
child_logger.warning("Child warning - handled by parent's handler")
````


The output will be:
```
app - WARNING - Parent warning
app.module - WARNING - Child warning - handled by parent's handler
```

Note that the child logger's message shows the correct logger name ('app.module') but is processed by the parent's handler.

#### Controlling Propagation

Sometimes you want to prevent log records from propagating up the hierarchy. You can control this with the `propagate` attribute:


````python
import logging

# Create parent logger with handler
parent_logger = logging.getLogger('app')
parent_logger.setLevel(logging.INFO)
parent_handler = logging.StreamHandler()
parent_handler.setFormatter(logging.Formatter('PARENT: %(message)s'))
parent_logger.addHandler(parent_handler)

# Create child logger with its own handler
child_logger = logging.getLogger('app.module')
child_logger.setLevel(logging.INFO)
child_handler = logging.StreamHandler()
child_handler.setFormatter(logging.Formatter('CHILD: %(message)s'))
child_logger.addHandler(child_handler)

# Log a message
child_logger.info("This message")  # Shows up twice - once from child, once from parent

# Disable propagation
child_logger.propagate = False
child_logger.info("This message only appears once")  # Only handled by child's handler
````


This is useful when you want different formatting or destinations for logs from different components of your application.

#### Effective Levels

The effective level of a logger depends on both its explicitly set level and its position in the hierarchy:


````python
import logging

# Root logger defaults to WARNING level
root = logging.getLogger()

# Parent logger with explicit level
parent = logging.getLogger('app')
parent.setLevel(logging.INFO)

# Child with no explicit level (inherits from parent)
child1 = logging.getLogger('app.module1')

# Child with explicit level
child2 = logging.getLogger('app.module2')
child2.setLevel(logging.DEBUG)

# Add a handler to the root to see all messages
handler = logging.StreamHandler()
handler.setFormatter(logging.Formatter('%(name)s - %(levelname)s - %(message)s'))
root.addHandler(handler)

# Test effective levels
print(f"Root effective level: {logging.getLevelName(root.getEffectiveLevel())}")
print(f"Parent effective level: {logging.getLevelName(parent.getEffectiveLevel())}")
print(f"Child1 effective level: {logging.getLevelName(child1.getEffectiveLevel())}")
print(f"Child2 effective level: {logging.getLevelName(child2.getEffectiveLevel())}")

# Test message propagation
child1.debug("Child1 debug message")  # Won't display (INFO level inherited from parent)
child1.info("Child1 info message")    # Will display
child2.debug("Child2 debug message")  # Will display (has DEBUG level)
````


Understanding the hierarchical nature of loggers helps you structure your application's logging system in a clean, organized way.

### 🔍 Logging in Different Modules

In larger applications, your code is usually split across multiple modules (files), and you'll want to implement logging in each of these modules while maintaining a consistent, unified logging system.

#### Recommended Approach

The recommended pattern for logging in a module is:


````python
# mymodule.py
import logging

# Create a module-level logger
logger = logging.getLogger(__name__)

def some_function():
    logger.info("Function called")
    # ...

class MyClass:
    def __init__(self):
        self.logger = logging.getLogger(__name__ + '.MyClass')
        self.logger.info("Instance created")
````


The `__name__` variable contains the module's name, which creates a hierarchical logger structure that mirrors your package structure. This approach has several benefits:

1. Each module's logs are automatically identified by module name
2. The hierarchy respects your application's structure
3. You can control logging levels per module
4. Configuration can be centralized in the main application

#### Configuring Logging in the Main Application

In your main application file, you configure the logging system for all modules:


````python
# main.py
import logging
import logging.config
import mymodule

# Configure the logging system
logging.config.dictConfig({
    'version': 1,
    'formatters': {
        'standard': {
            'format': '%(asctime)s - %(name)s - %(levelname)s - %(message)s'
        },
    },
    'handlers': {
        'console': {
            'class': 'logging.StreamHandler',
            'level': 'INFO',
            'formatter': 'standard',
            'stream': 'ext://sys.stdout'
        },
        'file': {
            'class': 'logging.FileHandler',
            'level': 'DEBUG',
            'formatter': 'standard',
            'filename': 'app.log',
            'mode': 'a',
        }
    },
    'loggers': {
        '': {  # Root logger
            'handlers': ['console'],
            'level': 'WARNING',
        },
        'mymodule': {  # Specific module logger
            'handlers': ['file'],
            'level': 'DEBUG',
            'propagate': True,
        },
    }
})

# Now mymodule's logging is configured
result = mymodule.some_function()
````


This sets up a logging configuration where:
- The root logger outputs WARNING and above to the console
- The 'mymodule' logger outputs DEBUG and above to a file
- The 'mymodule' logger also propagates to the root logger, so its WARNING+ messages appear on the console too

#### Module-level vs. Instance Loggers

In the example above, we had both module-level loggers and loggers specific to class instances. Here's when to use each:

- **Module-level loggers** (`logger = logging.getLogger(__name__)`) are best for logging from functions and class methods that don't need to track specific instances.

- **Instance loggers** (`self.logger = logging.getLogger(__name__ + '.ClassName')`) are useful when you want to track which specific object instance a log comes from.

For simple classes, using the module-level logger is usually sufficient:


````python
# simpler_module.py
import logging

logger = logging.getLogger(__name__)

class SimpleClass:
    def process(self, data):
        logger.info("Processing data with %s", self.__class__.__name__)
        # ...
````


For more complex scenarios where you want to track specific instances, use instance loggers:


````python
# complex_module.py
import logging
import uuid

class DatabaseConnection:
    def __init__(self, host, port):
        self.id = uuid.uuid4().hex[:8]  # Generate a unique ID for this instance
        self.host = host
        self.port = port
        self.logger = logging.getLogger(f"{__name__}.DatabaseConnection.{self.id}")
        self.logger.info("Created connection to %s:%s", host, port)

    def execute(self, query):
        self.logger.debug("Executing query: %s", query)
        # ... execute query ...
        self.logger.info("Query executed successfully")
````


This allows you to trace exactly which connection executed which query in your logs.

### 🔍 Capturing Stack Traces

When exceptions occur, capturing stack traces in your logs is crucial for debugging. Python's logging module provides built-in support for this.

#### Basic Exception Logging

To include exception information in your logs, use the `exc_info` parameter:


````python
import logging
logger = logging.getLogger(__name__)

try:
    result = 1 / 0
except ZeroDivisionError:
    logger.error("Division by zero error", exc_info=True)
````


This will log the exception type, message, and full traceback. In addition to `exc_info=True`, which captures the current exception, you can also pass a specific exception tuple:


````python
try:
    result = 1 / 0
except ZeroDivisionError as e:
    # Store the exception for later
    exception_info = (type(e), e, e.__traceback__)

# Later in the code...
logger.error("An error occurred earlier", exc_info=exception_info)
````


#### Using exception() Helper Method

There's also a shortcut `exception()` method that automatically sets `exc_info=True`:


````python
try:
    result = 1 / 0
except Exception as e:
    logger.exception("An error occurred")  # Equivalent to logger.error(..., exc_info=True)
````


The `exception()` method always logs at ERROR level, so only use it in except blocks.

#### Formatting Exception Messages

For more control over how exceptions are formatted, you can extract parts of the exception information:


````python
import logging
import traceback

logger = logging.getLogger(__name__)

try:
    result = 1 / 0
except Exception as e:
    # Format the exception as a string
    exception_message = str(e)
    exception_traceback = traceback.format_exc()

    # Log just the exception message
    logger.error("Error occurred: %s", exception_message)

    # Log detailed traceback at debug level for developers
    logger.debug("Detailed traceback:\n%s", exception_traceback)
````


This approach gives you more control over what gets logged at each level.

#### Best Practices for Exception Logging

1. **Log exceptions where they're handled**: Log the exception at the point where you catch and handle it, rather than re-raising it to be logged elsewhere.

2. **Include context**: Add relevant context information that helps understand why the exception occurred:


````python
def process_user(user_id):
    logger = logging.getLogger(__name__)
    try:
        user = get_user(user_id)
        result = perform_operation(user)
        return result
    except UserNotFoundError:
        logger.error("User not found during processing, user_id=%s", user_id)
        raise
    except OperationError as e:
        logger.exception("Operation failed for user_id=%s, operation=%s",
                        user_id, e.operation_name)
        return None
````


3. **Consider log levels carefully**: Fatal errors should be CRITICAL, handled exceptions often should be ERROR or WARNING depending on impact.

### 🔍 Rotating File Handlers

Log files can grow very large over time, potentially filling up disk space. The solution is to use rotating file handlers that limit file size and/or age.

#### Size-Based Rotation

The `RotatingFileHandler` rotates log files when they reach a certain size:


````python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger('my_app')
logger.setLevel(logging.INFO)

# Create a rotating file handler
handler = RotatingFileHandler(
    filename='app.log',      # Base file name
    maxBytes=1024*1024,      # 1 MB per file
    backupCount=5,           # Keep 5 backup files
    encoding='utf-8'
)

handler.setFormatter(logging.Formatter('%(asctime)s - %(levelname)s - %(message)s'))
logger.addHandler(handler)

# Test the rotating handler
for i in range(100000):
    logger.info(f"Test log message {i}")
````


This creates:
- app.log (current log file)
- app.log.1 (most recent backup)
- app.log.2 (second most recent backup)
- ...
- app.log.5 (oldest backup)

When app.log reaches 1 MB, it's renamed to app.log.1, and a new app.log is created. If app.log.1 already exists, it's renamed to app.log.2, and so on. The oldest log file (app.log.5) is deleted if it exists.

#### Time-Based Rotation

The `TimedRotatingFileHandler` rotates logs based on time periods:


````python
import logging
from logging.handlers import TimedRotatingFileHandler

logger = logging.getLogger('my_app')
logger.setLevel(logging.INFO)

# Create a time-rotating file handler
handler = TimedRotatingFileHandler(
    filename='app.log',
    when='midnight',       # Rotate at midnight
    interval=1,            # Every day
    backupCount=7,         # Keep 7 days of logs
    encoding='utf-8'
)

handler.setFormatter(logging.Formatter('%(asctime)s - %(levelname)s - %(message)s'))
logger.addHandler(handler)

logger.info("Application started")
````


This rotates logs at midnight every day, keeping 7 days of history. The rotated files are named with date suffixes like:
- app.log (current log)
- app.log.2023-08-15 (yesterday's log)
- app.log.2023-08-14 (2 days ago)

The `when` parameter accepts various time periods:
- 'S': Seconds
- 'M': Minutes
- 'H': Hours
- 'D': Days
- 'midnight': Roll over at midnight
- 'W0'-'W6': Weekday (0=Monday)

#### Hybrid Approach: WatchedFileHandler

If your application runs for extended periods, you might want to use `WatchedFileHandler` which detects when a log file has been rotated by an external tool (like logrotate):


````python
import logging
from logging.handlers import WatchedFileHandler

logger = logging.getLogger('my_app')
logger.setLevel(logging.INFO)

handler = WatchedFileHandler(
    filename='app.log',
    encoding='utf-8'
)

handler.setFormatter(logging.Formatter('%(asctime)s - %(levelname)s - %(message)s'))
logger.addHandler(handler)

logger.info("Application started")
````


This works well with system utilities like `logrotate` on Linux, which can handle compression and other advanced features.

#### ⚠️ Common Mistakes with Advanced Logging

* **Overusing LogRecord attributes**: Adding too many attributes to log records can make logs harder to read and analyze. Focus on information that's genuinely useful for debugging.

* **Inconsistent logger naming**: Make sure your logger naming follows a consistent pattern, typically based on module structure, to make it easier to configure and filter logs.

* **Not planning for production scale**: Logs that seem reasonable during development can become overwhelming in production. Plan your logging strategy with production volumes in mind.

* **Logging sensitive information**: Be careful not to log sensitive data such as passwords, personal information, or access tokens. Consider implementing redaction for sensitive fields.

### 🔍 Configuration Files

Configuration files provide a way to externalize settings from your code, making it possible to modify application behavior without changing the code itself. This separation of configuration from code is crucial for maintaining flexibility and following the principle of separation of concerns.

#### Python's Built-in configparser

The `configparser` module is part of Python's standard library and provides a way to work with INI-style configuration files:

```ini
# config.ini
```

````python
import configparser

# Create a config parser
config = configparser.ConfigParser()

# Read the configuration file
config.read('config.ini')

# Access configuration values
debug_mode = config.getboolean('development', 'debug')
log_level = config['development']['log_level']
timeout = config.getint('development', 'timeout')

print(f"Debug mode: {debug_mode}")
print(f"Log level: {log_level}")
print(f"Timeout: {timeout}")
````


This will output:
```
Debug mode: True
Log level: DEBUG
Timeout: 30
```

Note that the `timeout` value was inherited from the `DEFAULT` section since it wasn't explicitly set in the `development` section.

#### Integrating Command-Line Arguments with Logging

You can use command-line arguments to configure your logging system:


````python
import argparse
import logging

# Create an argument parser
parser = argparse.ArgumentParser(description='Sample application with logging')
parser.add_argument('--log-level', choices=['DEBUG', 'INFO', 'WARNING', 'ERROR', 'CRITICAL'],
                    default='INFO', help='Set the logging level')
parser.add_argument('--log-file', help='Log to a file instead of stdout')

# Parse the arguments
args = parser.parse_args()

# Set up logging based on arguments
log_level = getattr(logging, args.log_level)

if args.log_file:
    logging.basicConfig(
        level=log_level,
        format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
        filename=args.log_file,
        filemode='a'
    )
else:
    logging.basicConfig(
        level=log_level,
        format='%(levelname)s: %(message)s'
    )

# Get a logger
logger = logging.getLogger(__name__)

# Use the logger
logger.debug("This is a debug message")
logger.info("This is an info message")
logger.warning("This is a warning message")
````

#### Using click for More User-Friendly CLI

For more complex command-line interfaces, the `click` library offers a more user-friendly approach:


````python
import click
import logging

@click.command()
@click.option('--debug/--no-debug', default=False, help='Enable debug mode')
@click.option('--log-level', type=click.Choice(['DEBUG', 'INFO', 'WARNING', 'ERROR', 'CRITICAL']),
              default='INFO', help='Set the logging level')
@click.option('--log-file', help='Log to a file instead of stdout')
@click.argument('input_file')
def process_file(debug, log_level, log_file, input_file):
    """Process the given input file."""
    # Set up logging
    log_level_num = getattr(logging, log_level)

    if log_file:
        logging.basicConfig(
            level=log_level_num,
            format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
            filename=log_file,
            filemode='a'
        )
    else:
        logging.basicConfig(
            level=log_level_num,
            format='%(levelname)s: %(message)s'
        )

    logger = logging.getLogger(__name__)

    # Enable debug mode if requested
    if debug:
        logger.setLevel(logging.DEBUG)
        logger.debug("Debug mode enabled")

    # Process the file
    logger.info(f"Processing file: {input_file}")
    # ... process the file ...
    logger.info("Processing complete")

if __name__ == '__main__':
    process_file()
````

````python
import os
import argparse
import yaml
from dotenv import load_dotenv
import logging
import logging.config

def setup_config():
    """Set up application configuration from multiple sources."""
    # Step 1: Set default values
    config = {
        'debug': False,
        'log_level': 'INFO',
        'database': {
            'url': 'sqlite:///default.db'
        },
        'api_key': None,
        'log_file': None
    }

    # Step 2: Load from configuration file
    parser = argparse.ArgumentParser(add_help=False)
    parser.add_argument('--config', default='config.yaml')
    args, _ = parser.parse_known_args()

    try:
        with open(args.config, 'r') as file:
            file_config = yaml.safe_load(file)
            if file_config:
                deep_update(config, file_config)
    except FileNotFoundError:
        print(f"Warning: Configuration file {args.config} not found. Using defaults.")

    # Step 3: Load from environment variables
    load_dotenv()  # Load .env file if available

    if 'DEBUG' in os.environ:
        config['debug'] = os.environ['DEBUG'].lower() in ('true', 'yes', '1')

    if 'LOG_LEVEL' in os.environ:
        config['log_level'] = os.environ['LOG_LEVEL']

    if 'DATABASE_URL' in os.environ:
        config['database']['url'] = os.environ['DATABASE_URL']

    if 'API_KEY' in os.environ:
        config['api_key'] = os.environ['API_KEY']

    if 'LOG_FILE' in os.environ:
        config['log_file'] = os.environ['LOG_FILE']

    # Step 4: Parse command-line arguments
    parser = argparse.ArgumentParser(description='Sample application')
    parser.add_argument('--debug', action='store_true', help='Enable debug mode')
    parser.add_argument('--log-level', choices=['DEBUG', 'INFO', 'WARNING', 'ERROR', 'CRITICAL'],
                      help='Set the logging level')
    parser.add_argument('--database-url', help='Database connection URL')
    parser.add_argument('--api-key', help='API key for external services')
    parser.add_argument('--log-file', help='Log to a file instead of stdout')

    args = parser.parse_args()

    # Override with command-line arguments if provided
    if args.debug:
        config['debug'] = True

    if args.log_level:
        config['log_level'] = args.log_level

    if args.database_url:
        config['database']['url'] = args.database_url

    if args.api_key:
        config['api_key'] = args.api_key

    if args.log_file:
        config['log_file'] = args.log_file

    # Configure logging
    setup_logging(config)

    return config

def deep_update(target, source):
    """Recursively update a dictionary."""
    for key, value in source.items():
        if isinstance(value, dict) and key in target and isinstance(target[key], dict):
            deep_update(target[key], value)
        else:
            target[key] = value

def setup_logging(config):
    """Configure the logging system based on application configuration."""
    log_level = getattr(logging, config['log_level'])

    if config['log_file']:
        logging.basicConfig(
            level=log_level,
            format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
            filename=config['log_file'],
            filemode='a'
        )
    else:
        logging.basicConfig(
            level=log_level,
            format='%(levelname)s: %(message)s'
        )

    # Enable debug logging if debug mode is enabled
    if config['debug']:
        logging.getLogger().setLevel(logging.DEBUG)
        logging.debug("Debug mode enabled")

# Load configuration
config = setup_config()

# Use configuration in your application
logger = logging.getLogger(__name__)
logger.info("Application starting")
logger.debug(f"Configuration: {config}")

# Use configuration values
database_url = config['database']['url']
logger.info(f"Connecting to database: {database_url}")
````

#### ⚠️ Common Mistakes with Configuration Management

* **Hardcoding sensitive information**: Never hardcode sensitive information like passwords, API keys, or tokens in your code.

* **Not validating configuration**: Always validate configuration values to ensure they're of the correct type and within expected ranges.

* **Inconsistent configuration structure**: Maintain a consistent structure across different configuration files and environments to avoid confusion.

* **Overcomplex configuration**: Keep your configuration system as simple as possible. An overly complex system can be difficult to understand and maintain.

* **Exposing too much in logs**: Be careful about what configuration values you log, especially in production environments where logs might be accessible to multiple people.

````python
# 1. Create a YAML configuration file (config.yaml)
"""
# This would be saved as config.yaml
default:
  application:
    name: MyApp
    version: 1.0
    timeout: 30
  logging:
    level: INFO
    format: '%(asctime)s - %(name)s - %(levelname)s - %(message)s'
    file: null  # Log to console by default

development:
  application:
    debug: true
  logging:
    level: DEBUG
  database:
    url: sqlite:///dev.db
    echo: true

production:
  application:
    debug: false
  logging:
    level: WARNING
    file: app.log
  database:
    url: postgresql://user:password@localhost/prod
    echo: false
"""

# 2-3. Load configuration and set up logging
import os
import yaml
import logging

# Load configuration file
def load_config():
    # Determine environment
    env = os.environ.get('APP_ENV', 'development')
    print(f"Loading configuration for {env} environment")

    # Load configuration file
    with open('config.yaml', 'r') as file:
        config_data = yaml.safe_load(file)

    # Merge default with environment-specific settings
    config = {**config_data['default']}

    # Deep merge for nested dictionaries
    if env in config_data:
        for key, value in config_data[env].items():
            if isinstance(value, dict) and key in config and isinstance(config[key], dict):
                config[key] = {**config[key], **value}
            else:
                config[key] = value

    return config

# Configure logging based on config
def setup_logging(config):
    log_config = config['logging']

    # Get log level
    log_level = getattr(logging, log_config['level'])

    # Configure logging
    if log_config.get('file'):
        logging.basicConfig(
            level=log_level,
            format=log_config['format'],
            filename=log_config['file'],
            filemode='a'
        )
    else:
        logging.basicConfig(
            level=log_level,
            format=log_config['format']
        )

    logger = logging.getLogger(__name__)
    logger.info("Logging configured successfully")
    return logger

# Main function
def main():
    # Load configuration
    config = load_config()

    # Set up logging
    logger = setup_logging(config)

    # Log application information
    app_config = config['application']
    logger.info(f"Starting {app_config['name']} v{app_config['version']}")
    logger.debug(f"Debug mode: {app_config.get('debug', False)}")

    # Log database information if available
    if 'database' in config:
        db_config = config['database']
        logger.info(f"Database URL: {db_config['url']}")
        logger.debug(f"SQL Echo: {db_config.get('echo', False)}")

    # Log full configuration in debug mode
    logger.debug(f"Full configuration: {config}")

if __name__ == "__main__":
    main()
````


*Combining configuration sources with validation:*


````python
# 8-9. Create a unified configuration system with validation
import os
import argparse
from dotenv import load_dotenv
import yaml
import logging
import sys
from typing import Any, Dict, List, Optional, Union

class ConfigError(Exception):
    """Exception raised for configuration errors."""
    pass

class Config:
    """Unified configuration system with validation."""

    def __init__(self, config_file: str = 'config.yaml'):
        self.config_file = config_file
        self.config = {}
        self.logger = None

    def load(self) -> Dict[str, Any]:
        """Load configuration from all sources with proper precedence."""
        # Step 1: Parse command-line arguments (some might be needed early)
        args = self._parse_args()

        # Use specified config file if provided
        if args.config:
            self.config_file = args.config

        # Step 2: Load configuration file
        try:
            file_config = self._load_config_file()
        except FileNotFoundError:
            print(f"Error: Configuration file '{self.config_file}' not found.")
            sys.exit(1)
        except yaml.YAMLError as e:
            print(f"Error parsing YAML configuration: {e}")
            sys.exit(1)

        # Step 3: Determine environment
        env = args.env or os.environ.get('APP_ENV', 'development')

        # Step 4: Merge default with environment-specific settings
        self.config = self._merge_config(file_config, env)

        # Step 5: Override with environment variables
        self._apply_environment_variables()

        # Step 6: Override with command-line arguments (highest priority)
        self._apply_command_line_args(args)

        # Step 7: Validate configuration
        try:
            self._validate_config()
        except ConfigError as e:
            print(f"Configuration error: {e}")
            sys.exit(1)

        return self.config

    def _parse_args(self) -> argparse.Namespace:
        """Parse command-line arguments."""
        parser = argparse.ArgumentParser(description='Application configuration')

        # Basic arguments
        parser.add_argument('--config', help='Path to configuration file')
        parser.add_argument('--env', choices=['development', 'production', 'testing'],
                          help='Application environment')

        # Logging arguments
        parser.add_argument('--log-level', choices=['DEBUG', 'INFO', 'WARNING', 'ERROR', 'CRITICAL'],
                          help='Set the logging level')
        parser.add_argument('--log-file', help='Log to a file instead of stdout')
        parser.add_argument('--verbose', action='store_true', help='Enable verbose output (DEBUG level)')

        # Database arguments
        parser.add_argument('--database-url', help='Database connection URL')

        # Application arguments
        parser.add_argument('--debug', action='store_true', help='Enable debug mode')

        return parser.parse_args()

    def _load_config_file(self) -> Dict[str, Any]:
        """Load configuration from file."""
        with open(self.config_file, 'r') as file:
            return yaml.safe_load(file)

    def _merge_config(self, config_data: Dict[str, Any], env: str) -> Dict[str, Any]:
        """Merge default config with environment-specific config."""
        if 'default' not in config_data:
            raise ConfigError("Configuration file must contain a 'default' section")

        # Start with default configuration
        result = self._deep_copy(config_data['default'])

        # Merge with environment-specific configuration if available
        if env in config_data:
            self._deep_update(result, config_data[env])

        return result

    def _deep_copy(self, obj: Any) -> Any:
        """Create a deep copy of a dictionary or list."""
        if isinstance(obj, dict):
            return {k: self._deep_copy(v) for k, v in obj.items()}
        elif isinstance(obj, list):
            return [self._deep_copy(v) for v in obj]
        else:
            return obj

    def _deep_update(self, target: Dict[str, Any], source: Dict[str, Any]) -> None:
        """Recursively update a dictionary."""
        for key, value in source.items():
            if isinstance(value, dict) and key in target and isinstance(target[key], dict):
                self._deep_update(target[key], value)
            else:
                target[key] = value

    def _apply_environment_variables(self) -> None:
        """Apply configuration from environment variables."""
        # Load .env file
        load_dotenv()

        # Application settings
        if 'DEBUG' in os.environ:
            debug_value = os.environ['DEBUG'].lower() in ('true', 'yes', '1')
            self._ensure_dict(self.config, 'application')
            self.config['application']['debug'] = debug_value

        # Logging settings
        if 'LOG_LEVEL' in os.environ:
            self._ensure_dict(self.config, 'logging')
            self.config['logging']['level'] = os.environ['LOG_LEVEL']

        if 'LOG_FILE' in os.environ:
            self._ensure_dict(self.config, 'logging')
            self.config['logging']['file'] = os.environ['LOG_FILE']

        # Database settings
        if 'DATABASE_URL' in os.environ:
            self._ensure_dict(self.config, 'database')
            self.config['database']['url'] = os.environ['DATABASE_URL']

        # API settings
        if 'API_KEY' in os.environ:
            self._ensure_dict(self.config, 'api')
            self.config['api']['key'] = os.environ['API_KEY']

    def _apply_command_line_args(self, args: argparse.Namespace) -> None:
        """Apply configuration from command-line arguments."""
        # Logging settings
        if args.log_level:
            self._ensure_dict(self.config, 'logging')
            self.config['logging']['level'] = args.log_level

        if args.log_file:
            self._ensure_dict(self.config, 'logging')
            self.config['logging']['file'] = args.log_file

        if args.verbose:
            self._ensure_dict(self.config, 'logging')
            self.config['logging']['level'] = 'DEBUG'

        # Database settings
        if args.database_url:
            self._ensure_dict(self.config, 'database')
            self.config['database']['url'] = args.database_url

        # Application settings
        if args.debug:
            self._ensure_dict(self.config, 'application')
            self.config['application']['debug'] = True

    def _ensure_dict(self, parent: Dict[str, Any], key: str) -> None:
        """Ensure that a dictionary key exists and contains a dictionary."""
        if key not in parent:
            parent[key] = {}
        elif not isinstance(parent[key], dict):
            parent[key] = {}

    def _validate_config(self) -> None:
        """Validate the configuration."""
        # Validate logging configuration
        if 'logging' in self.config:
            log_config = self.config['logging']

            # Validate log level
            if 'level' in log_config:
                level = log_config['level']
                valid_levels = ['DEBUG', 'INFO', 'WARNING', 'ERROR', 'CRITICAL']
                if level not in valid_levels:
                    raise ConfigError(f"Invalid log level: {level}. Must be one of {valid_levels}")

            # Validate log file if specified
            if 'file' in log_config and log_config['file']:
                log_file = log_config['file']
                log_dir = os.path.dirname(log_file)
                if log_dir and not os.path.exists(log_dir):
                    raise ConfigError(f"Log directory does not exist: {log_dir}")

        # Validate database configuration
        if 'database' in self.config:
            db_config = self.config['database']

            # Check required database URL
            if 'url' not in db_config:
                raise ConfigError("Database URL is required in database configuration")

            # Validate URL format (simple check)
            url = db_config['url']
            if '://' not in url:
                raise ConfigError(f"Invalid database URL format: {url}")

        # Validate application configuration
        if 'application' in self.config:
            app_config = self.config['application']

            # Validate application name if present
            if 'name' in app_config and not isinstance(app_config['name'], str):
                raise ConfigError("Application name must be a string")

            # Validate timeout if present
            if 'timeout' in app_config:
                timeout = app_config['timeout']
                if not isinstance(timeout, (int, float)) or timeout <= 0:
                    raise ConfigError(f"Invalid timeout value: {timeout}. Must be a positive number")

    def setup_logging(self) -> logging.Logger:
        """Configure logging based on configuration."""
        if 'logging' not in self.config:
            # Default configuration if not specified
            logging.basicConfig(
                level=logging.INFO,
                format='%(levelname)s: %(message)s'
            )
            self.logger = logging.getLogger(__name__)
            return self.logger

        log_config = self.config['logging']

        # Get log level
        log_level_name = log_config.get('level', 'INFO')
        log_level = getattr(logging, log_level_name)

        # Get log format
        log_format = log_config.get('format', '%(asctime)s - %(name)s - %(levelname)s - %(message)s')

        # Configure logging
        if 'file' in log_config and log_config['file']:
            logging.basicConfig(
                level=log_level,
                format=log_format,
                filename=log_config['file'],
                filemode='a'
            )
        else:
            logging.basicConfig(
                level=log_level,
                format=log_format
            )

        self.logger = logging.getLogger(__name__)
        self.logger.info("Logging configured successfully")

        # Log debug info if in debug mode
        if 'application' in self.config and self.config['application'].get('debug', False):
            self.logger.setLevel(logging.DEBUG)
            self.logger.debug("Debug mode enabled")

        return self.logger

# Sample usage:
def main():
    # Initialize configuration
    config_manager = Config()
    config = config_manager.load()

    # Setup logging
    logger = config_manager.setup_logging()

    # Use configuration in application
    logger.info("Application started with unified configuration")
    if config['application'].get('debug', False):
        logger.debug("Configuration: %s", config)

    # Connect to database
    if 'database' in config:
        database_url = config['database']['url']
        logger.info("Connecting to database: %s", database_url)
        # Actual database connection code would go here

    # More application logic...
    logger.info("Application running normally")

if __name__ == "__main__":
    main()
````

## 📚 Key Terms

- **Logger:** The primary interface for capturing log records in Python's logging system, typically corresponding to a module or component.

- **Handler:** A component that determines where log messages are sent (console, file, network, etc.) and can filter based on severity.

- **Formatter:** A component that defines the structure and content of log messages, controlling how they appear in the output.

- **Log Level:** A numeric value or name (DEBUG, INFO, WARNING, ERROR, CRITICAL) that indicates the severity or importance of a log message.

- **Filter:** A component that provides additional control over which log records are output, beyond the simple level-based filtering.

- **Propagation:** The process by which log records are passed from child loggers to parent loggers in the hierarchy.

- **Rotating File Handler:** A log handler that limits file size by creating new log files when the current one reaches a size limit or time threshold.

- **Configuration File:** An external file (like INI, YAML, or JSON) that stores application settings separate from the code.

- **Environment Variable:** A variable in the system environment that can be accessed by applications, often used for configuration.

- **Command-Line Argument:** A value passed to a program when it is invoked from the command line, typically used to control program behavior.

- **Hierarchical Configuration:** A nested structure of configuration settings that organizes related settings in groups.

- **Configuration Precedence:** The order in which different configuration sources (defaults, files, environment, command line) are applied to determine the final settings.
