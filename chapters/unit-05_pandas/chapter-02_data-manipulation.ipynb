## **Chapter 2.1:** Data Importing

### 🔍 Reading from Files

Data rarely starts in a Pandas DataFrame. Instead, it typically exists in files, databases, or web services. Pandas provides a comprehensive suite of functions to import data from various sources directly into DataFrames, making it easy to begin your analysis.

The most common data formats you'll encounter include CSV (Comma-Separated Values), Excel spreadsheets, JSON (JavaScript Object Notation), and database tables. Pandas offers specialized functions to handle each of these formats efficiently.

### 🧩 Important Components

* **read_csv()**: Import data from CSV and other delimited text files
* **read_excel()**: Import data from Excel spreadsheets
* **read_json()**: Import data from JSON files or strings
* **read_sql()**: Import data from SQL databases
* **read_html()**: Import HTML tables from web pages

    | Function | Purpose | Common Parameters |
    |----------|---------|-------------------|
    | `pd.read_csv()` | Read delimited text files | `delimiter`, `header`, `skiprows`, `usecols` |
    | `pd.read_excel()` | Read Excel files | `sheet_name`, `header`, `usecols` |
    | `pd.read_json()` | Read JSON data | `orient`, `lines`, `typ` |
    | `pd.read_sql()` | Read SQL query or database table | `con` (connection), `sql`, `index_col` |
    | `pd.read_html()` | Read HTML tables | `match`, `header`, `flavor` |

### 📘 Main Content

#### Importing CSV Files

`CSV` (*Comma-Separated Values*) is one of the most common formats for storing tabular data. Pandas makes it easy to import CSV files with the `read_csv()` function:

```python
import pandas as pd

# Basic usage - import a CSV file
df = pd.read_csv('data.csv')

# Handling different delimiters
df_semicolon = pd.read_csv('data.csv', delimiter=';')

# Specifying column names
df_custom_names = pd.read_csv('data.csv', names=['A', 'B', 'C'])

# Setting a specific column as the index
df_indexed = pd.read_csv('data.csv', index_col='ID')

# Reading only specific columns
df_selected_cols = pd.read_csv('data.csv', usecols=['Name', 'Age', 'Salary'])

# Skipping rows
df_skip_rows = pd.read_csv('data.csv', skiprows=3)  # Skip first 3 rows
```


The `read_csv()` function is highly versatile, with dozens of parameters to handle various CSV file formats and quirks. Some of the most useful parameters include:

- `delimiter`: Specify the character used to separate values (e.g., `,`, `;`, `\t`)
- `header`: Specify which row(s) to use as column names
- `index_col`: Specify which column(s) to use as row indices
- `usecols`: Specify which columns to import (others will be ignored)
- `skiprows`: Specify rows to skip at the beginning of the file
- `nrows`: Specify the number of rows to read
- `na_values`: Specify strings to recognize as NA/NaN

#### Working with Excel Files

Excel spreadsheets are widely used in business contexts. Pandas can read these files directly with the `read_excel()` function:

```python
import pandas as pd

# Basic usage - import an Excel file
df = pd.read_excel('data.xlsx')

# Specify a particular sheet
df_sheet2 = pd.read_excel('data.xlsx', sheet_name='Sheet2')

# Read multiple sheets (returns a dictionary of DataFrames)
all_sheets = pd.read_excel('data.xlsx', sheet_name=None)

# Access a specific sheet from the dictionary
sheet1_df = all_sheets['Sheet1']
```


#### Working with JSON Data

JSON (JavaScript Object Notation) is a common format for web APIs and config files. Pandas can read JSON with the `read_json()` function:



````python
import pandas as pd

# Basic usage - read a JSON file
df = pd.read_json('data/products.json')
````


The structure of JSON can vary significantly. Pandas attempts to parse JSON into a tabular format, but sometimes you may need to preprocess the JSON data or use additional parameters like `orient` to specify the JSON structure.

#### Importing from Databases

Pandas can connect to various database systems and import data using the `read_sql()` function:



````python
import pandas as pd
import sqlite3  # Or other database connector

# Create a connection to a SQLite database
conn = sqlite3.connect('data/database.db')

# Read a table
df = pd.read_sql('SELECT * FROM employees', conn)

# Read with a WHERE clause
df_filtered = pd.read_sql('SELECT * FROM employees WHERE department = "Sales"', conn)

# Close the connection when done
conn.close()
````


To connect to different database systems, you'll need the appropriate Python connector (e.g., `psycopg2` for PostgreSQL, `mysql-connector` for MySQL).

#### Reading HTML Tables

Pandas can scrape tables from HTML web pages using the `read_html()` function:



````python
import pandas as pd

# Read all tables from a webpage
tables = pd.read_html('https://en.wikipedia.org/wiki/List_of_countries_by_population')

# The result is a list of DataFrames, one for each table
first_table = tables[0]

# You can specify which tables to read with the match parameter
gdp_tables = pd.read_html('https://en.wikipedia.org/wiki/List_of_countries_by_GDP_(nominal)')
````


The `read_html()` function uses libraries like `lxml`, `html5lib`  and `beautifulsoup4` to parse HTML content. You may need to install these dependencies:


### ⚠️ Common Mistakes with Data Importing

* **Ignoring encoding issues**: Different files may use different character encodings (e.g., UTF-8, Latin-1). Specify the `encoding` parameter if you encounter strange characters.
* **Forgetting to handle missing values**: Files often represent missing values in various ways (empty strings, "NA", "NULL"). Use the `na_values` parameter to specify these.
* **Assuming the first row is headers**: Some files may have data in the first row or multiple header rows. Use the `header` parameter to specify the header row(s).
* **Not checking data types**: Pandas tries to infer data types, but sometimes gets it wrong. Check the types after importing and use `dtype` parameter if needed.

## **Chapter 2.2:** Data Exploration

### 🔍 Understanding Your Data

Once you've imported your data into a Pandas DataFrame, the next step is to explore and understand its structure, content, and characteristics. This exploratory phase is crucial for identifying patterns, anomalies, and potential issues that may need to be addressed before conducting more advanced analysis.

Pandas provides numerous methods to help you quickly examine your data from different angles, giving you a comprehensive view of what you're working with. These exploration tools allow you to preview the data, check its dimensions, view summary statistics, and understand the data types of each column.

### 🧩 Important Components

* **Previewing Methods**: Functions like `head()`, `tail()`, and `sample()` for looking at subsets of your data
* **Information Functions**: Methods like `info()` and `dtypes` for understanding data structure
* **Statistical Summaries**: The `describe()` method for generating summary statistics
* **Size and Shape**: Attributes like `shape`, `size`, and `ndim` for dimensional information
* **Unique Values**: Methods like `unique()` and `value_counts()` for categorical analysis

    | Method/Attribute | Purpose | Example |
    |------------------|---------|---------|
    | `head(n)` | View first n rows | `df.head(10)` |
    | `tail(n)` | View last n rows | `df.tail(5)` |
    | `sample(n)` | View n random rows | `df.sample(3)` |
    | `info()` | Summary of DataFrame | `df.info()` |
    | `describe()` | Statistical summary | `df.describe()` |
    | `shape` | Dimensions (rows, cols) | `df.shape` |
    | `dtypes` | Data types of columns | `df.dtypes` |

### 📘 Main Content

#### Previewing Your Data

The first step in exploring a dataset is to look at a sample of the records to get a feel for its content and structure. Pandas provides several methods for this:



````python
import pandas as pd

# Create a sample DataFrame
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Heidi', 'Ivan', 'Judy'],
    'age': [24, 42, 18, 33, 57, 36, 29, 41, 30, 45],
    'city': ['New York', 'Boston', 'Chicago', 'Boston', 'Chicago', 'New York', 'Chicago', 'New York', 'Boston', 'Chicago'],
    'salary': [65000, 77000, 48000, 90000, 102000, 67000, 88000, 99000, 73000, 91000],
    'hired_date': pd.date_range(start='2020-01-01', periods=10, freq='M')
}

df = pd.DataFrame(data)
````

````python
# View the first 5 rows
print("First 5 rows:")
df.head()
````

````python
# View the last 3 rows
print("Last 3 rows:")
df.tail(3)
````

````python
# View a random sample of 2 rows
print("Random sample of 2 rows:")
df.sample(2)
````


These preview methods are particularly useful for large datasets where viewing the entire DataFrame would be impractical. By default, `head()` and `tail()` show 5 rows, but you can specify a different number as a parameter.

#### Understanding DataFrame Structure

To get a comprehensive overview of your DataFrame's structure, the `info()` method is invaluable:



````python
# Get information about the DataFrame
print("DataFrame information:")
df.info()
````


The `info()` method provides:
- The number of rows and columns
- Column names and data types
- The number of non-null values in each column
- Memory usage information

You can also check specific attributes of the DataFrame:



````python
# Check dimensions
print(f"Shape (rows, columns): {df.shape}")
print(f"Total elements: {df.size}")
print(f"Number of dimensions: {df.ndim}")
````

````python
# Check column names and index
print(f"Columns: {df.columns.tolist()}")
print(f"Index: {df.index}")
````

````python
# Check data types
print("Data types:")
df.dtypes
````


#### Generating Summary Statistics

The `describe()` method provides a statistical summary of your numerical data:



````python
# Generate summary statistics for numerical columns
print("Summary statistics:")
df.describe()
````

````python
# Include all columns, not just numerical ones
print("Summary with all columns:")
df.describe(include='all')
````


By default, `describe()` only includes numerical columns, but you can use the `include='all'` parameter to include statistics for all data types. The summary includes:
- Count (number of non-null values)
- Mean, standard deviation, minimum, and maximum
- Quartiles (25th, 50th/median, and 75th percentiles)
- For non-numerical columns: count, unique values, top (most frequent), and frequency

#### Exploring Categorical Data

For categorical columns, you often want to understand the distribution of values:



````python
# Check unique values in a categorical column
print(f"Unique cities: {df['city'].unique()}")
````

````python
# Count occurrences of each value
print("City counts:")
df['city'].value_counts()
````

````python
# Get relative frequencies (proportions)
print("City proportions:")
df['city'].value_counts(normalize=True)
````


These methods help you understand the distribution of categorical variables, which is essential for many analyses and visualizations.

#### Cross-tabulations

To explore relationships between categorical variables, cross-tabulations are useful:



````python
# Create a binary feature for demonstration
df['high_salary'] = df['salary'] > 80000
````

````python
# Create a cross-tabulation
print("Cross-tabulation of city and high_salary:")
pd.crosstab(df['city'], df['high_salary'])
````

````python
# With percentages
print("\nRow percentages:")
pd.crosstab(df['city'], df['high_salary'], normalize='index')
````


Cross-tabulations show the frequency of combinations of values across two or more categorical variables, helping you identify potential patterns or relationships.

### ⚠️ Common Mistakes with Data Exploration

* **Skipping the exploration phase**: Diving into complex analysis without understanding your data can lead to incorrect conclusions or missed insights.
* **Not checking for missing values**: Missing data can significantly impact your analysis; always check for and handle null values.
* **Ignoring data types**: Incorrect data types can cause unexpected behavior in calculations or visualizations.
* **Overlooking categorical distributions**: Imbalanced categories can affect statistical models and interpretations.

## **Chapter 2.3:** Data Indexing

### 🔍 Selecting Subsets of Data

Indexing in Pandas refers to the process of selecting specific rows and columns from a DataFrame. This is a fundamental skill for data analysis, as it allows you to focus on particular subsets of your data that are relevant to your analysis questions.

Pandas provides multiple methods for indexing, each with its own use cases and advantages. Mastering these methods will give you precise control over your data selection, making your analysis more efficient and your code more readable.

### 🧩 Important Components

* **Basic Selection**: Using brackets to select columns or rows
* **loc**: Label-based indexing for selecting rows and columns using labels
* **iloc**: Integer position-based indexing for selecting by numerical position
* **Boolean Indexing**: Using conditions to filter rows that meet specific criteria
* **ix**: Mixed indexing (deprecated in newer versions)

    | Indexing Method | Selection Type | Example |
    |-----------------|----------------|---------|
    | `df['column']` | Single column | `df['age']` |
    | `df[['col1', 'col2']]` | Multiple columns | `df[['name', 'age']]` |
    | `df[start:end]` | Rows by position | `df[5:10]` |
    | `df.loc[label]` | Row by label | `df.loc['row_label']` |
    | `df.loc[:, 'col']` | All rows, specific column | `df.loc[:, 'age']` |
    | `df.iloc[pos]` | Row by position | `df.iloc[3]` |
    | `df.iloc[:, pos]` | All rows, column by position | `df.iloc[:, 2]` |
    | `df[condition]` | Rows meeting condition | `df[df['age'] > 30]` |

### 📘 Main Content

#### Basic Column Selection

The simplest form of indexing in Pandas is selecting columns using bracket notation:


````python
import pandas as pd

# Create a sample DataFrame
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'age': [24, 42, 18, 33, 57],
    'city': ['New York', 'Boston', 'Chicago', 'Boston', 'Chicago'],
    'salary': [65000, 77000, 48000, 90000, 102000]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
df
````

````python
# Select a single column (returns a Series)
names = df['name']
print("Just the 'name' column (Series):")
names
````

````python
# Select multiple columns (returns a DataFrame)
name_and_age = df[['name', 'age']]
print("Name and age columns (DataFrame):")
name_and_age
````


Note that selecting a single column returns a Series, while selecting multiple columns returns a DataFrame. This distinction is important because Series and DataFrames have slightly different methods and behaviors.

#### Basic Row Selection with Slicing

You can select rows using slice notation:



````python
# Select rows by position using slicing
first_three_rows = df[0:3]  # Rows 0, 1, and 2
print("First three rows:")
first_three_rows
````

````python
# Select from a specific row to the end
from_second_row = df[1:]
print("From second row to end:")
from_second_row
````

````python
# Select in steps (every other row)
alternate_rows = df[::2]
print("Every other row:")
alternate_rows
````

It's important to note that slice notation with DataFrames selects rows, not columns. This is different from NumPy arrays, where the first dimension in slicing refers to rows and the second to columns.

#### Label-Based Indexing with loc

The `loc` accessor is a powerful way to select data based on labels rather than numerical positions. It follows the pattern `df.loc[row_labels, column_labels]` and allows for intuitive and precise data selection.

````python
df
````

````python
# Select a single row by its label (index)
row_by_label = df.loc[2]  # Row with index 2
print("Row with index 2:")
print(row_by_label)
````

````python
# Select multiple rows by label
selected_rows = df.loc[[0, 2, 4]]  # Rows with indices 0, 2, and 4
print("Rows with indices 0, 2, and 4:")
selected_rows
````

````python
# Select a range of rows by label
row_range = df.loc[1:3]  # Rows with indices 1, 2, and 3 (inclusive!)
print("Rows with indices 1 through 3 (inclusive):")
row_range
````

````python
# Select specific rows and columns
subset = df.loc[[1, 3], ['name', 'salary']]
print("Specific rows and columns:")
subset
````

````python
# Select all rows but only specific columns
all_names_ages = df.loc[:, ['name', 'age']]
print("All rows, name and age columns:")
all_names_ages
````

The key characteristic of `loc` is that it's inclusive of the end label in ranges. This is different from most Python slicing, which excludes the end value.

#### Integer Position-Based Indexing with iloc

The `iloc` accessor selects data based on integer positions, similar to NumPy array indexing:



````python
# Select a single row by its position
row_by_position = df.iloc[2]  # Third row
print("Third row (position 2):")
row_by_position
````

````python
# Select multiple rows by position
selected_positions = df.iloc[[0, 2, 4]]  # First, third, and fifth rows
print("Rows at positions 0, 2, and 4:")
selected_positions
````

````python
# Select a range of rows by position
position_range = df.iloc[1:3]  # Second to third rows (exclusive!)
print("Rows from position 1 to 2 (exclusive of 3):")
position_range
````

````python
# Select specific rows and columns by position
subset_by_position = df.iloc[[1, 3], [0, 3]]  # Rows 1 & 3, columns 0 & 3
print("Specific rows and columns by position:")
subset_by_position
````

````python
# Select all rows but only specific columns by position
cols_by_position = df.iloc[:, [1, 2]]  # All rows, columns 1 & 2
print("All rows, columns at positions 1 and 2:")
cols_by_position
````


Unlike `loc`, `iloc` follows Python's standard 0-based indexing and exclusive end bounds. This means `df.iloc[1:3]` includes positions 1 and 2, but not 3.

#### Boolean Indexing

Boolean indexing allows you to select rows based on conditions:



````python
# Select rows where age is greater than 30
older_than_30 = df[df['age'] > 30]
print("People older than 30:")
older_than_30
````

````python
# Select rows with multiple conditions (using & for AND, | for OR)
from_boston = df[df['city'] == 'Boston']
print("People from Boston:")
from_boston
````

````python
# Combine conditions
boston_high_salary = df[(df['city'] == 'Boston') & (df['salary'] > 80000)]
print("People from Boston with salary > 80000:")
boston_high_salary
````

````python
# Use the query method for more readable conditions
chicago_query = df.query("city == 'Chicago'")
print("People from Chicago (using query):")
chicago_query
````


Boolean indexing is particularly powerful for data analysis, as it allows you to filter data based on complex conditions. Note that when combining multiple conditions, each condition must be enclosed in parentheses and joined with the bitwise operators `&` (AND) and `|` (OR), not the logical operators `and` and `or`.

#### Combining Different Indexing Methods

You can combine these methods for more complex selections:



````python
# Use boolean indexing to filter rows, then select specific columns
high_earners_cities = df[df['salary'] > 70000][['name', 'city']]
print("Names and cities of high earners:")
high_earners_cities
````

````python
# Use loc with boolean condition
young_names = df.loc[df['age'] < 30, 'name']
print("Names of people under 30:")
young_names
````

````python
# Use iloc to select rows, then filter by condition
subset_high_salary = df.iloc[:3][df.iloc[:3]['salary'] > 60000]
print("High salary from first three people:")
subset_high_salary
````


These combinations give you extreme flexibility in how you select and filter your data.

### ⚠️ Common Mistakes with Data Indexing

* **Confusing loc and iloc**: Remember that `loc` uses labels and is inclusive of the endpoint, while `iloc` uses integer positions and excludes the endpoint.
* **Forgetting parentheses in boolean conditions**: When combining conditions, each condition needs parentheses: `(condition1) & (condition2)`.
* **Using Python's and/or instead of &/|**: Boolean indexing requires the bitwise operators `&` and `|`, not the logical operators `and` and `or`.
* **Chain indexing**: Avoid chained indexing like `df['A']['B']` as it can lead to unpredictable behavior; use `df.loc[:, ('A', 'B')]` instead.

This critical step ensures that your analysis is based on high-quality, reliable data.

Common data quality issues include missing values, duplicate records, inconsistent formatting, and incorrect data types. Pandas provides powerful tools to identify and address these issues efficiently.

* **Duplicates**: Functions for identifying and handling duplicate records
* **Data Type Conversion**: Tools for changing column data types
* **Text Cleaning**: String methods for standardizing and transforming text data

### 📘 Main Content

#### Handling Missing Values

Missing values can significantly impact your analysis. Pandas represents missing values primarily with `NaN` (Not a Number) from NumPy, though `None` is also used for object data types.

First, let's look at how to detect missing values:

````python
import pandas as pd
import numpy as np

# Create a DataFrame with missing values
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': ['a', 'b', 'c', None, 'e']
}

df = pd.DataFrame(data)
print("DataFrame with missing values:")
df
````

````python
# Check for missing values
print("Missing value indicators (True where NaN/None):")
df.isna()
````

````python
# Check if any values are missing in each column
print("Any missing values per column?")
df.isna().any()
````

````python
# Count missing values per column
print("Count of missing values per column:")
df.isna().sum()
````

````python
# Count missing values in entire DataFrame
print(f"Total missing values: {df.isna().sum().sum()}")
````

````python
# Check for non-missing values
print("Non-missing value indicators:")
df.notna()
````

Once you've identified missing values, you have several options:

1. **Drop rows or columns with missing values**:



````python
# Drop rows with any missing values
df_dropped_rows = df.dropna()
print("After dropping rows with any NaN:")
df_dropped_rows
````

````python
# Drop only rows where all values are missing
df_dropped_all_na = df.dropna(how='all')
print("After dropping rows where all values are NaN:")
df_dropped_all_na
````

````python
# Drop columns with at least 3 missing values
df_dropped_cols = df.dropna(axis=1, thresh=3)
print("After dropping columns with at least 3 NaNs:")
df_dropped_cols
````


2. **Fill missing values**:



````python
# Fill all missing values with a specific value
df_filled = df.fillna(0)
print("After filling all NaN with 0:")
df_filled
````

````python
# Fill with different values for each column
fill_values = {'A': 0, 'B': 1, 'C': 2, 'D': 'missing'}
df_filled_dict = df.fillna(fill_values)
print("After filling with column-specific values:")
df_filled_dict
````

````python
# Fill with previous or next valid value
df_ffill = df.ffill()  # Forward fill (use previous value)
print("After forward fill:")
df_ffill
````

````python
# For backward fill, use:
df_bfill = df.bfill()  # Backward fill (use next value)
print("After backward fill:")

df_bfill
````


3. **Fill with calculated values**:



````python
# Fill with column mean
df_mean_filled = df.copy()
for column in ['A', 'B', 'C']:  # Exclude 'D' as it's non-numeric
    df_mean_filled[column] = df_mean_filled[column].fillna(df_mean_filled[column].mean())
    
print("After filling numeric columns with their means:")

df_mean_filled
````

````python
# First infer object types, then interpolate
df_interp = df.infer_objects(copy=False).interpolate()
print("After interpolation:")

df_interp

# Previously, .interpolate() was used to fill missing values with linear interpolation.
````


#### Removing Duplicate Records

Duplicate records can skew your analysis and waste computational resources. Pandas makes it easy to identify and remove them:



````python
# Create a DataFrame with duplicate rows
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Alice', 'Bob'],
    'age': [25, 30, 35, 25, 30],
    'city': ['New York', 'Boston', 'Chicago', 'New York', 'Boston']
}

df = pd.DataFrame(data)
print("DataFrame with duplicates:")
print(df)

# Check for duplicate rows
print("\nDuplicate row indicators (True for duplicates):")
print(df.duplicated())

# Count duplicate rows
print(f"\nNumber of duplicate rows: {df.duplicated().sum()}")

# Remove duplicate rows (keep first occurrence)
df_no_dupes = df.drop_duplicates()
print("\nAfter removing duplicates (keeping first):")
print(df_no_dupes)

# Remove duplicates based on specific columns
df_no_name_dupes = df.drop_duplicates(subset=['name'])
print("\nAfter removing duplicates based on 'name' column:")
print(df_no_name_dupes)

# Keep last occurrence instead of first
df_keep_last = df.drop_duplicates(keep='last')
print("\nAfter removing duplicates (keeping last):")
print(df_keep_last)
````


You can also mark duplicates without removing them, which is useful for further investigation:



````python
# Mark duplicates with a new column
df['is_duplicate'] = df.duplicated()
print("\nWith duplicate flag column:")
print(df)
````


#### Converting Data Types

Ensuring your data is stored with the correct data types is essential for both memory efficiency and proper analysis:



````python
# Create a DataFrame with mixed types
data = {
    'id': ['1', '2', '3', '4', '5'],
    'value': ['10.5', '20.0', '30.2', 'N/A', '50.8'],
    'category': ['A', 'B', 'A', 'C', 'B'],
    'date_str': ['2021-01-01', '2021-02-15', '2021-03-30', '2021-04-10', '2021-05-22']
}

df = pd.DataFrame(data)
print("DataFrame with string data types:")
print(df.dtypes)

# Convert 'id' to integer
df['id'] = df['id'].astype(int)

# Convert 'value' to float (handling errors)
df['value'] = pd.to_numeric(df['value'], errors='coerce')  # 'N/A' becomes NaN

# Convert 'category' to categorical type
df['category'] = df['category'].astype('category')

# Convert 'date_str' to datetime
df['date'] = pd.to_datetime(df['date_str'])

print("\nData types after conversion:")
print(df.dtypes)
print("\nConverted DataFrame:")
print(df)
````


The `astype()` method is useful for straightforward conversions, while specialized functions like `to_numeric()`, `to_datetime()`, and `to_timedelta()` provide more control for complex conversions.

For categorical data, using the 'category' dtype can significantly improve memory usage and performance, especially for columns with a limited number of unique values.

#### String Cleaning and Transformation

Text data often requires cleaning and standardization. Pandas provides a `.str` accessor with numerous string methods:



````python
# Create a DataFrame with messy string data
data = {
    'name': [' Alice ', 'BOB', 'charlie ', ' Dave', 'Eve'],
    'email': ['alice@example.com', 'bob@example.COM', 'charlie@gmail.com', 'dave@example.com', 'eve123@yahoo.com'],
    'id': ['ID-001', 'ID_002', 'id:003', 'ID#004', 'id-005']
}

df = pd.DataFrame(data)
print("DataFrame with messy string data:")
print(df)

# Strip whitespace
df['name'] = df['name'].str.strip()

# Standardize case
df['name'] = df['name'].str.title()  # Capitalize first letter of each word
df['email'] = df['email'].str.lower()

# Extract domain from email
df['domain'] = df['email'].str.extract(r'@([^.]+)')

# Standardize ID format
df['id'] = df['id'].str.replace(r'[^a-zA-Z0-9]', '-', regex=True)  # Replace non-alphanumeric with '-'
df['id'] = df['id'].str.upper()  # Convert to uppercase

print("\nCleaned DataFrame:")
print(df)
````


The `.str` accessor provides numerous methods for string manipulation, including:
- `strip()`, `lstrip()`, `rstrip()` for removing whitespace
- `upper()`, `lower()`, `title()`, `capitalize()` for case conversion
- `replace()` for substituting substrings
- `extract()`, `contains()`, `find()` for pattern matching
- `split()`, `join()`, `slice()` for string splitting and combining

#### Handling Outliers

Outliers are extreme values that significantly differ from other observations. They can distort analyses and should be identified and handled appropriately:



````python
import matplotlib.pyplot as plt

# Create a DataFrame with outliers
data = {'value': [10, 12, 13, 15, 18, 20, 22, 25, 30, 40, 100]}
df = pd.DataFrame(data)

# Visualize the data (optional)
# df['value'].plot(kind='box')
# plt.show()

# Calculate summary statistics
print("Summary statistics:")
print(df['value'].describe())

# Identify outliers using the IQR method
Q1 = df['value'].quantile(0.25)
Q3 = df['value'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"\nIQR: {IQR}")
print(f"Lower bound: {lower_bound}")
print(f"Upper bound: {upper_bound}")

# Flag outliers
outliers = df[(df['value'] < lower_bound) | (df['value'] > upper_bound)]
print("\nIdentified outliers:")
print(outliers)

# Handle outliers - option 1: remove them
df_no_outliers = df[(df['value'] >= lower_bound) & (df['value'] <= upper_bound)]
print("\nDataFrame with outliers removed:")
print(df_no_outliers)

# Handle outliers - option 2: cap them
df_capped = df.copy()
df_capped['value'] = df_capped['value'].clip(lower=lower_bound, upper=upper_bound)
print("\nDataFrame with outliers capped:")
print(df_capped)
````


There are multiple approaches to handling outliers, and the appropriate method depends on your specific data and analysis goals:
- **Removal**: Delete outlier rows if they're erroneous or not relevant
- **Capping**: Set outliers to a maximum/minimum threshold
- **Transformation**: Apply transformations (e.g., log) to compress the range
- **Separate analysis**: Analyze outliers separately to understand their cause

### ⚠️ Common Mistakes with Data Cleaning

* **Dropping missing values without understanding their pattern**: Missing values might indicate something meaningful; investigate their distribution before removing them.
* **Using mean imputation when data is skewed**: For skewed distributions, median might be a better replacement value.
* **Over-cleaning data**: Be careful not to remove important signals or patterns by being too aggressive with outlier removal.
* **Not documenting cleaning steps**: Keep track of all transformations for reproducibility and transparency.
* **Assuming cleaning is a one-time task**: Data cleaning is often iterative and may need revision as you understand your data better.

````python
# Create a copy of the original data
df_clean = df.copy()

# Handle missing values
# For numerical columns, we could fill with mean, median, or remove rows
# The choice depends on the amount and pattern of missing data
if missing_values.sum() > 0:
    print("\nHandling missing values...")
    # Example approach - fill numerical columns with median
    numerical_cols = df_clean.select_dtypes(include=['number']).columns
    for col in numerical_cols:
        if df_clean[col].isna().sum() > 0:
            df_clean[col] = df_clean[col].fillna(df_clean[col].median())
    
    # For categorical columns, fill with mode
    categorical_cols = df_clean.select_dtypes(include=['object']).columns
    for col in categorical_cols:
        if df_clean[col].isna().sum() > 0:
            df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
    
    # Verify missing values are handled
    print("Missing values after handling:")
    print(df_clean.isna().sum().sum())

# Remove duplicates if any
if duplicate_count > 0:
    print("\nRemoving duplicate rows...")
    df_clean = df_clean.drop_duplicates()
    print(f"Rows after removing duplicates: {df_clean.shape[0]}")

# Convert data types as needed
# Example: Convert string representations of numbers to float
for col in df_clean.columns:
    # Try to convert object columns to numeric if they contain numbers
    if df_clean[col].dtype == 'object':
        try:
            df_clean[col] = pd.to_numeric(df_clean[col])
            print(f"Converted {col} to numeric type")
        except:
            # If conversion fails, it's likely a genuine categorical column
            pass
````


#### **Step 4: Initial Observations**

````python
# Check correlations between numerical variables
numeric_df = df_clean.select_dtypes(include=['number'])
correlation_matrix = numeric_df.corr()

# Display the most strongly correlated features
print("\nTop correlations:")
# Flatten the correlation matrix and sort
corr_pairs = correlation_matrix.unstack()
# Remove self-correlations and duplicates
corr_pairs = corr_pairs[corr_pairs < 1.0]
top_correlations = corr_pairs.abs().sort_values(ascending=False)[:10]
print(top_correlations)

# Summarize what we've learned
print("\nData Summary:")
print(f"- Dataset contains {df_clean.shape[0]} rows and {df_clean.shape[1]} columns")
print(f"- Data types: {df_clean.dtypes.value_counts().to_dict()}")
if 'quality' in df_clean.columns:
    quality_counts = df_clean['quality'].value_counts()
    print(f"- Quality distribution: {quality_counts.to_dict()}")

print("\nNext Steps:")
print("1. Further explore relationships between process parameters and quality")
print("2. Identify key factors that affect product quality")
print("3. Prepare data for building predictive models in later challenges")
````

## 📚 Key Terms

- **DataFrame**: Pandas' primary 2D data structure with labeled axes (rows and columns)
- **Series**: Pandas' 1D labeled array, often representing a single column of a DataFrame
- **Missing Values**: Data points represented as NaN (Not a Number) or None
- **loc**: Label-based indexing for selecting data by row and column labels
- **iloc**: Position-based indexing for selecting data by integer position
- **Boolean Indexing**: Using conditions to filter data (e.g., df[df['age'] > 30])
- **Data Cleaning**: Process of identifying and correcting errors and inconsistencies in data
- **Data Type**: The classification of data (e.g., integer, float, string) that determines how values are stored and manipulated
- **Duplicates**: Multiple identical records in a dataset
- **Outliers**: Data points that significantly differ from other observations
