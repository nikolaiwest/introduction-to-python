## **Chapter 3.1:** Filtering and Sorting

### 🔍 Focusing on Relevant Data

Data analysis rarely requires examining every single data point. More often, you'll want to focus on specific subsets of your data that are relevant to your current question or hypothesis. Filtering allows you to extract these subsets based on conditions, while sorting helps you organize data in ways that highlight patterns and extremes.

These operations are fundamental to data analysis, as they help you navigate large datasets efficiently and focus your attention on the most relevant information. Mastering filtering and sorting in Pandas will significantly enhance your ability to explore data and extract insights effectively.

### 🧩 Important Components

* **Boolean Indexing**: Using conditions to filter rows that meet specific criteria
* **Logical Operators**: Combining multiple conditions with `&` (AND), `|` (OR), and `~` (NOT)
* **Query Method**: A more readable alternative to boolean indexing for complex conditions
* **Sort Values**: Arranging data in ascending or descending order based on one or more columns
* **Sort Index**: Ordering data based on index values rather than column values

| Operation | Method | Example |
|-----------|--------|---------|
| Simple Filtering | Boolean indexing | `df[df['column'] > value]` |
| Complex Filtering | Combined conditions | `df[(df['col1'] > x) & (df['col2'] == y)]` |
| Query-style Filtering | `.query()` method | `df.query("col1 > x and col2 == y")` |
| Sort by Column | `.sort_values()` | `df.sort_values('column', ascending=False)` |
| Sort by Multiple Columns | `.sort_values()` with list | `df.sort_values(['col1', 'col2'])` |
| Sort by Index | `.sort_index()` | `df.sort_index(ascending=False)` |

### 📘 Main Content

#### Filtering with Boolean Indexing

Boolean indexing is the most common method for filtering data in Pandas. It involves creating a boolean mask (a Series of True/False values) and using it to select rows:



````python
import pandas as pd
import numpy as np

# Create a sample DataFrame
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Hannah'],
    'age': [25, 32, 18, 47, 22, 35, 29, 41],
    'city': ['New York', 'Boston', 'Chicago', 'Boston', 'Chicago', 'New York', 'Boston', 'Chicago'],
    'salary': [72000, 85000, 45000, 120000, 55000, 92000, 76000, 110000],
    'department': ['Marketing', 'IT', 'Sales', 'IT', 'Marketing', 'Sales', 'IT', 'Marketing']
}

df = pd.DataFrame(data)

print("Original DataFrame:")
df
````

````python
# Simple filtering: employees older than 30
older_than_30 = df[df['age'] > 30]
print("\nEmployees older than 30:")
older_than_30
````

````python
# Filtering with exact matching: employees from Chicago
from_chicago = df[df['city'] == 'Chicago']
print("\nEmployees from Chicago:")
from_chicago
````

````python
# Filtering with .isin(): employees in IT or Sales departments
departments_of_interest = ['IT', 'Sales']
it_or_sales = df[df['department'].isin(departments_of_interest)]
print("\nEmployees in IT or Sales departments:")
it_or_sales
````

````python
# Filtering with negation: employees NOT from Boston
not_from_boston = df[df['city'] != 'Boston']
# Alternative: not_from_boston = df[~(df['city'] == 'Boston')]
print("\nEmployees NOT from Boston:")
not_from_boston
````


#### Combining Multiple Conditions

For more complex filtering, you can combine multiple conditions using logical operators:



````python
# Combining conditions with & (AND): Marketing employees with salary > 60000
high_paid_marketing = df[(df['department'] == 'Marketing') & (df['salary'] > 60000)]
print("\nHigh-paid Marketing employees:")
print(high_paid_marketing)

# Combining conditions with | (OR): employees from Chicago OR in the IT department
chicago_or_it = df[(df['city'] == 'Chicago') | (df['department'] == 'IT')]
print("\nEmployees from Chicago OR in IT:")
print(chicago_or_it)

# Complex condition: young IT employees from Boston
young_boston_it = df[(df['age'] < 30) & (df['city'] == 'Boston') & (df['department'] == 'IT')]
print("\nYoung IT employees from Boston:")
print(young_boston_it)
````


It's important to note that when combining conditions, each condition must be enclosed in parentheses and connected using the bitwise operators `&` (AND) and `|` (OR), not the Python logical operators `and` and `or`.

#### Using the Query Method

For complex filtering conditions, the `query()` method provides a more readable alternative to boolean indexing:



````python
# Simple query
high_salary = df.query("salary > 80000")
print("\nEmployees with salary > 80000 (using query):")
print(high_salary)

# Complex query with multiple conditions
complex_query = df.query("age > 25 and department == 'IT' or salary >= 100000")
print("\nEmployees matching complex criteria (using query):")
print(complex_query)

# Using variables in queries
min_age = 30
target_city = 'Chicago'
filtered = df.query("age >= @min_age and city == @target_city")
print(f"\nEmployees from {target_city} who are at least {min_age} years old:")
print(filtered)
````


The `query()` method offers a more intuitive syntax, especially for complex conditions, and allows you to reference Python variables using the `@` prefix.

#### Sorting Data

Sorting is another essential operation that helps you organize data for analysis and presentation:



````python
# Sort by a single column (ascending by default)
sorted_by_age = df.sort_values('age')
print("\nEmployees sorted by age (ascending):")
print(sorted_by_age)

# Sort in descending order
sorted_by_salary_desc = df.sort_values('salary', ascending=False)
print("\nEmployees sorted by salary (descending):")
print(sorted_by_salary_desc)

# Sort by multiple columns
sorted_by_dept_and_salary = df.sort_values(['department', 'salary'], ascending=[True, False])
print("\nEmployees sorted by department, then by salary (descending) within each department:")
print(sorted_by_dept_and_salary)

# Sort by index
df_indexed = df.set_index('name')
sorted_by_index = df_indexed.sort_index()
print("\nData sorted by index (name):")
print(sorted_by_index)
````


When sorting by multiple columns, you can specify different sort orders for each column using a list of boolean values for the `ascending` parameter.

#### Filtering, Then Sorting

Often, filtering and sorting are used together to focus on a specific subset of data and arrange it in a meaningful order:



````python
# Filter high-earning employees and sort by age
high_earners = df[df['salary'] > 80000].sort_values('age')
print("\nHigh-earning employees, sorted by age:")
print(high_earners)

# Filter employees from specific cities and sort by salary in descending order
target_cities = ['New York', 'Boston']
selected_employees = df[df['city'].isin(target_cities)].sort_values('salary', ascending=False)
print("\nEmployees from New York or Boston, sorted by salary (descending):")
print(selected_employees)
````


#### Ranking Data

Pandas also provides the `rank()` method for assigning ranks to values:



````python
# Add salary rank column (higher salary = lower rank number)
df['salary_rank'] = df['salary'].rank(ascending=False)
print("\nEmployees with salary rank:")
print(df)

# Rank with different tie-breaking methods
df['salary_rank_avg'] = df['salary'].rank(method='average', ascending=False)
df['salary_rank_min'] = df['salary'].rank(method='min', ascending=False)
df['salary_rank_dense'] = df['salary'].rank(method='dense', ascending=False)
print("\nDifferent ranking methods:")
print(df[['name', 'salary', 'salary_rank', 'salary_rank_avg', 'salary_rank_min', 'salary_rank_dense']])
````


The `rank()` method assigns ranks to values in a Series, with different methods for handling ties. This is useful for creating competitions, finding top performers, or computing percentiles.

### ⚠️ Common Mistakes with Filtering and Sorting

* **Forgetting parentheses in complex conditions**: Each condition must be enclosed in parentheses when using `&` or `|`.
* **Using `and`/`or` instead of `&`/`|`**: Boolean indexing requires the bitwise operators `&` and `|`, not the logical operators `and` and `or`.
* **Comparing with a single `=`**: Use `==` for equality comparison, not `=` (which is assignment).
* **Not resetting index after filtering**: Filtered DataFrames retain the original indices unless you use `reset_index()`.
* **Overlooking NaN values**: Be careful when filtering numeric columns, as rows with NaN values will be excluded from the result.

## **Chapter 3.2:** Grouping and Aggregating

### 🔍 Finding Patterns in Groups

Grouping and aggregation are at the heart of data analysis, allowing you to segment your data into meaningful groups and calculate summary statistics for each group. This process helps uncover patterns, relationships, and insights that might be hidden when looking at the data as a whole.

The `groupby()` operation in Pandas follows the "split-apply-combine" pattern:
1. **Split**: The data is divided into groups based on one or more keys
2. **Apply**: A function is applied to each group independently
3. **Combine**: The results are combined into a new data structure

This powerful paradigm enables a wide range of analyses, from simple summaries to complex transformations, making it an essential tool for data analysis.

### 🧩 Important Components

* **GroupBy Object**: The initial result of the `groupby()` operation
* **Aggregation**: Calculating summary statistics for each group
* **Transformation**: Applying functions that return data with the same shape as the input
* **Filtration**: Selecting groups based on group-wise properties
* **Multiple Aggregation**: Applying different functions to different columns

| Operation | Method | Example |
|-----------|--------|---------|
| Basic Grouping | `groupby()` | `df.groupby('category')` |
| Single Aggregation | `aggregation func` | `df.groupby('category')['sales'].sum()` |
| Multiple Aggregations | `agg()` | `df.groupby('category').agg({'sales': 'sum', 'price': 'mean'})` |
| Group Transformation | `transform()` | `df.groupby('category')['price'].transform('mean')` |
| Group Filtering | `filter()` | `df.groupby('category').filter(lambda x: x['sales'].mean() > 100)` |

### 📘 Main Content

#### Creating Groups with GroupBy

The first step in aggregation is to create groups using the `groupby()` method:



````python
import pandas as pd
import numpy as np

# Create a sample sales DataFrame
data = {
    'date': pd.date_range(start='2023-01-01', periods=20),
    'product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C', 'A', 
                'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C', 'A', 'B'],
    'category': ['Electronics', 'Electronics', 'Clothing', 'Electronics', 'Electronics', 
                'Clothing', 'Electronics', 'Electronics', 'Clothing', 'Electronics',
                'Clothing', 'Clothing', 'Electronics', 'Clothing', 'Clothing', 
                'Electronics', 'Clothing', 'Clothing', 'Electronics', 'Clothing'],
    'store': ['North', 'South', 'North', 'South', 'North', 'South', 'North', 'South', 'North', 'South',
             'North', 'South', 'North', 'South', 'North', 'South', 'North', 'South', 'North', 'South'],
    'sales': np.random.randint(100, 1000, 20),
    'units': np.random.randint(1, 10, 20),
    'price': np.random.uniform(50, 500, 20).round(2),
    'discount': np.random.choice([0, 0.1, 0.2, 0.3], 20)
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df.head())

# Create a GroupBy object by a single column
grouped_by_product = df.groupby('product')
print("\nGrouped by product - type:", type(grouped_by_product))

# GroupBy objects are not DataFrames - they need aggregation to produce output
# We can see the groups
print("\nGroup keys:")
print(list(grouped_by_product.groups.keys()))

# We can also get the indices for each group
print("\nIndices for group 'A':")
print(grouped_by_product.groups['A'])
````


A GroupBy object is an intermediate object that doesn't display as a table. It's a collection of groups waiting for some operation to be applied.

#### Basic Aggregations

The most common operations on GroupBy objects are aggregations, which compute summary statistics for each group:



````python
# Calculate total sales by product
sales_by_product = df.groupby('product')['sales'].sum()
print("Total sales by product:")
print(sales_by_product)

# Calculate average price by category
avg_price_by_category = df.groupby('category')['price'].mean()
print("\nAverage price by category:")
print(avg_price_by_category)

# Count number of items in each store
items_by_store = df.groupby('store').size()
print("\nNumber of items by store:")
print(items_by_store)

# Multiple statistics for a column
sales_stats_by_product = df.groupby('product')['sales'].agg(['min', 'max', 'mean', 'sum'])
print("\nSales statistics by product:")
print(sales_stats_by_product)

# Count number of unique products by category
unique_products_by_category = df.groupby('category')['product'].nunique()
print("\nNumber of unique products by category:")
print(unique_products_by_category)
````


These basic aggregations provide quick insights into your data, showing totals, averages, counts, and other statistics grouped by categories of interest.

#### Grouping by Multiple Columns

You can create hierarchical groups by specifying multiple grouping columns:



````python
# Group by category and store
grouped_by_cat_store = df.groupby(['category', 'store'])

# Calculate total sales for each category-store combination
sales_by_cat_store = grouped_by_cat_store['sales'].sum()
print("\nTotal sales by category and store:")
print(sales_by_cat_store)

# The result has a MultiIndex - we can unstack it for a more table-like view
print("\nUnstacked view (category × store):")
print(sales_by_cat_store.unstack())

# We can also reset the index to convert to a regular DataFrame
print("\nReset index to DataFrame:")
print(sales_by_cat_store.reset_index())

# To get more control over the output format, we can use as_index=False
flat_sales_by_cat_store = df.groupby(['category', 'store'], as_index=False)['sales'].sum()
print("\nFlat format with as_index=False:")
print(flat_sales_by_cat_store)
````


Grouping by multiple columns creates hierarchical indices, which can be reshaped with methods like `unstack()` and `reset_index()` to produce the desired output format.

#### Aggregating Multiple Columns

You can apply different aggregation functions to different columns using the `agg()` method:



````python
# Apply different aggregations to different columns
agg_result = df.groupby('category').agg({
    'sales': 'sum',           # Sum of sales
    'units': 'sum',           # Sum of units
    'price': 'mean',          # Average price
    'discount': 'mean'        # Average discount
})
print("\nMultiple aggregations by category:")
print(agg_result)

# Apply multiple aggregations to the same column
multi_agg = df.groupby('product').agg({
    'sales': ['sum', 'mean', 'std'],
    'price': ['min', 'max', 'mean']
})
print("\nMultiple aggregations per column:")
print(multi_agg)

# Access specific results from multi-level columns
print("\nAverage sales for product A:", multi_agg.loc['A', ('sales', 'mean')])
````


The `agg()` method provides great flexibility in choosing which aggregations to apply to which columns, allowing for highly customized summary tables.

#### Named Aggregations

For clearer column names in the result, you can use named aggregations:



````python
# Named aggregations for more readable column names
named_agg = df.groupby('category').agg(
    total_sales=('sales', 'sum'),
    avg_price=('price', 'mean'),
    total_units=('units', 'sum'),
    count=('product', 'size')
)
print("\nNamed aggregations:")
print(named_agg)
````


#### Custom Aggregation Functions

You can also define custom aggregation functions for more specialized calculations:



````python
# Define a custom aggregation function
def sales_per_unit(x):
    return x['sales'].sum() / x['units'].sum()

# Apply the custom function
custom_agg = df.groupby('category').apply(sales_per_unit)
print("\nCustom aggregation (sales per unit):")
print(custom_agg)

# Multiple custom functions with lambda expressions
lambda_agg = df.groupby('product').agg({
    'price': [
        ('avg', 'mean'),
        ('range', lambda x: x.max() - x.min())
    ],
    'sales': [
        ('total', 'sum'),
        ('per_unit', lambda x: x.sum() / df.loc[x.index, 'units'].sum())
    ]
})
print("\nAggregation with lambda functions:")
print(lambda_agg)
````


Custom functions provide unlimited flexibility for your aggregations, allowing you to calculate any metric you need from your grouped data.

#### Group Transformation

While aggregation reduces each group to a single value, transformation returns a result with the same shape as the input. This is useful for operations like normalization and filling missing values:



````python
# Calculate the percentage of total sales for each product
total_sales = df['sales'].sum()
df['pct_of_total'] = df.groupby('product')['sales'].transform(lambda x: x.sum() / total_sales * 100)

# Calculate how each product's price compares to the average price for its category
df['price_vs_category_avg'] = df['price'] / df.groupby('category')['price'].transform('mean')

print("\nDataFrame with transformed columns:")
print(df[['product', 'category', 'sales', 'pct_of_total', 'price', 'price_vs_category_avg']].head(10))

# Fill missing values with group averages (if we had any NaN values)
df_with_nan = df.copy()
df_with_nan.loc[0:5, 'sales'] = np.nan
df_with_nan['sales_filled'] = df_with_nan['sales'].fillna(df_with_nan.groupby('category')['sales'].transform('mean'))

print("\nFilling missing values with category average:")
print(df_with_nan[['category', 'sales', 'sales_filled']].head(10))
````


Transformations are powerful because they create new columns that contain group-level information while maintaining the original row structure of your data.

#### Filtering Groups

The `filter()` method allows you to select entire groups based on a condition applied to each group:



````python
# Keep only groups (products) with total sales > 2000
high_sales_products = df.groupby('product').filter(lambda x: x['sales'].sum() > 2000)
print("\nFiltered to products with sales > 2000:")
print(high_sales_products['product'].unique())

# Keep only category-store combinations with at least 4 items
large_groups = df.groupby(['category', 'store']).filter(lambda x: len(x) >= 4)
print("\nFiltered to category-store groups with >= 4 items:")
print(large_groups.groupby(['category', 'store']).size())
````


Filtering is useful for focusing on significant groups and removing groups that don't meet certain criteria from your analysis.

#### Applying Functions to Groups

The `apply()` method is the most flexible way to work with groups, allowing you to perform almost any operation on each group:



````python
# Calculate the top 2 sales days for each product
top_sales_days = df.groupby('product').apply(lambda x: x.nlargest(2, 'sales'))
print("\nTop 2 sales days for each product:")
print(top_sales_days)

# Calculate correlation between price and sales for each category
price_sales_corr = df.groupby('category').apply(lambda x: x['price'].corr(x['sales']))
print("\nPrice-sales correlation by category:")
print(price_sales_corr)

# More complex example: calculate cumulative sales within each store
def add_cumulative_sales(group):
    group = group.sort_values('date')
    group['cumulative_sales'] = group['sales'].cumsum()
    return group

df_with_cumulative = df.groupby('store').apply(add_cumulative_sales)
print("\nDataFrame with cumulative sales by store:")
print(df_with_cumulative[['store', 'date', 'sales', 'cumulative_sales']].head(10))
````


The `apply()` method is a catch-all for complex group operations, trading some performance for maximum flexibility.

### ⚠️ Common Mistakes with Grouping and Aggregating

* **Forgetting to specify an aggregation function**: Just calling `df.groupby('column')` without an aggregation function returns a GroupBy object, not a result.
* **Using the wrong aggregation function**: Make sure the function makes sense for your data (e.g., don't take the average of categorical data).
* **Ignoring the index structure**: Groupby operations often return Series or DataFrames with hierarchical indices, which require special handling.
* **Performance issues with large datasets**: For very large datasets, certain groupby operations can be slow - consider using more efficient methods or packages like Dask for big data.
* **Failing to handle NaN values**: NaN values in grouping columns create their own group, which might not be intended.

### 🔄 Quick Check: Grouping and Aggregating

What's the difference between `transform()` and `agg()` when used with `groupby()`?

<details>
   <summary>View Answer</summary>

   The key difference lies in the shape of the output:
   
   - `agg()` reduces each group to a single value, resulting in a DataFrame with one row per group.
   - `transform()` returns a result with the same shape as the input DataFrame, where each row receives the result of the transformation function applied to its group.
   
   For example:
   ```python
   # agg() - one row per product
   df.groupby('product')['sales'].agg('mean')  # Series with product as index
   
   # transform() - same number of rows as original
   df.groupby('product')['sales'].transform('mean')  # Series with same index as df
   ```
   
   This difference makes `transform()` useful for operations like normalization or creating features that compare individual values to group statistics.
</details>

How would you calculate the percentage contribution of each row to its group's total?

<details>
   <summary>View Answer</summary>

   You can use the `transform()` method to calculate each group's total, then divide each value by its group total:
   
   ```python
   # Calculate percentage of group total
   df['pct_of_group'] = df['sales'] / df.groupby('category')['sales'].transform('sum') * 100
   ```
   
   This creates a new column 'pct_of_group' containing each row's sales as a percentage of the total sales in its category.
   
   This technique is valuable for understanding the relative contribution of individual items within categories, which is useful for analyses like Pareto (80/20) analysis or identifying key contributors.
</details>

````python
# Sales concentration by category
def sales_concentration(x):
    max_sale = x.max()
    avg_sale = x.mean()
    if avg_sale == 0:
        return 0
    return max_sale / avg_sale

category_concentration = sales_df.groupby('category')['total'].agg(sales_concentration)
print("Sales concentration by category (max/avg ratio):")
print(category_concentration)

# Date of highest sales for each region
def date_of_highest_sales(group):
    if group.empty or group['total'].isna().all():
        return pd.NaT
    idx = group['total'].idxmax()
    return group.loc[idx, 'date']

highest_sales_dates = sales_df.groupby('region').apply(date_of_highest_sales)
print("\nDate of highest sales for each region:")
print(highest_sales_dates)

# Discount effectiveness (total sales / potential sales without discount)
def discount_effectiveness(group):
    actual_sales = group['total'].sum()
    potential_sales = (group['quantity'] * group['unit_price']).sum()
    if potential_sales == 0:
        return 0
    return actual_sales / potential_sales

category_discount_effect = sales_df.groupby('category').apply(discount_effectiveness)
print("\nDiscount effectiveness by category (actual/potential ratio):")
print(category_discount_effect)
````

````python
# Identify the top 3 products by total sales in each category
top_products = sales_df.groupby(['category', 'product_id'])['total'].sum().reset_index()
top_products_by_category = top_products.sort_values(['category', 'total'], ascending=[True, False])
top_3_products = top_products_by_category.groupby('category').head(3)
print("Top 3 products by sales in each category:")
print(top_3_products)

# Calculate month-over-month sales growth percentage for each category
# First, get monthly sales by category
monthly_sales = sales_df.groupby(['category', 'month'])['total'].sum().reset_index()
monthly_sales = monthly_sales.sort_values(['category', 'month'])

# Calculate growth percentage
monthly_sales['prev_month_sales'] = monthly_sales.groupby('category')['total'].shift(1)
monthly_sales['growth_pct'] = ((monthly_sales['total'] - monthly_sales['prev_month_sales']) / 
                              monthly_sales['prev_month_sales'] * 100).round(2)
print("\nMonth-over-month sales growth by category:")
print(monthly_sales[monthly_sales['growth_pct'].notna()].head(15))

# Find which payment method has the highest average order value in each region
payment_aov_by_region = sales_df.groupby(['region', 'payment_method'])['total'].mean().reset_index()
highest_aov_payment = payment_aov_by_region.sort_values(['region', 'total'], ascending=[True, False])
highest_aov_payment = highest_aov_payment.groupby('region').first()
print("\nPayment method with highest average order value in each region:")
print(highest_aov_payment)

# Create a "customer value" metric
customer_analysis = sales_df.groupby('customer_id').agg(
    order_count=('order_id', 'nunique'),
    avg_order_value=('total', 'mean'),
    total_spend=('total', 'sum')
)
# Create composite score: normalize each metric and sum them
for col in ['order_count', 'avg_order_value', 'total_spend']:
    # Min-max scaling to 0-100 range
    min_val = customer_analysis[col].min()
    max_val = customer_analysis[col].max()
    range_val = max_val - min_val
    if range_val == 0:
        customer_analysis[f'{col}_score'] = 0
    else:
        customer_analysis[f'{col}_score'] = ((customer_analysis[col] - min_val) / range_val) * 100

# Calculate composite score (weighted average)
customer_analysis['customer_value'] = (
    0.3 * customer_analysis['order_count_score'] +
    0.3 * customer_analysis['avg_order_value_score'] +
    0.4 * customer_analysis['total_spend_score']
).round(2)

print("\nTop 10 customers by customer value score:")
print(customer_analysis.sort_values('customer_value', ascending=False).head(10))

# Analyze relationship between discount level and quantity purchased
discount_analysis = sales_df.groupby('discount').agg(
    avg_quantity=('quantity', 'mean'),
    avg_order_value=('total', 'mean'),
    order_count=('order_id', 'count')
)
print("\nRelationship between discount level and purchase behavior:")
print(discount_analysis)
````

## **Chapter 3.3:** Pivot Tables and Cross-Tabulation

### 🔍 Reshaping Data for Multidimensional Analysis

Pivot tables and cross-tabulations are powerful tools for transforming your data into summary tables that facilitate multidimensional analysis. These techniques allow you to restructure your data to reveal relationships and patterns that might be difficult to see in the original format.

Originally popularized by spreadsheet applications like Excel, pivot tables in Pandas provide an intuitive way to aggregate data across multiple dimensions, making it easier to answer complex analytical questions. Cross-tabulations, a related concept, are especially useful for examining the relationship between categorical variables by counting their co-occurrence.

### 🧩 Important Components

* **pivot_table()**: Creates a spreadsheet-style pivot table as a DataFrame
* **crosstab()**: Computes a cross-tabulation of two or more factors
* **pivot()**: Reshapes data based on column values (without aggregation)
* **stack/unstack**: Methods for pivoting between wide and long formats
* **melt()**: Converts a DataFrame from wide to long format

| Method | Primary Purpose | Key Parameters |
|--------|----------------|----------------|
| `pivot_table()` | Summarize data with aggregation | `values`, `index`, `columns`, `aggfunc` |
| `crosstab()` | Count co-occurrences of categorical variables | `index`, `columns`, `values`, `aggfunc` |
| `pivot()` | Reshape without aggregation | `index`, `columns`, `values` |
| `melt()` | Convert from wide to long format | `id_vars`, `value_vars`, `var_name`, `value_name` |
| `stack()` / `unstack()` | Convert between levels of a hierarchical index | `level`, `fill_value` |

### 📘 Main Content

#### Creating Pivot Tables

Pivot tables are a powerful way to summarize data across multiple dimensions. Let's explore how to create and customize them:



````python
import pandas as pd
import numpy as np

# Create a sample sales DataFrame
data = {
    'date': pd.date_range(start='2023-01-01', periods=100),
    'product': np.random.choice(['Laptop', 'Phone', 'Tablet', 'Monitor'], 100),
    'region': np.random.choice(['North', 'South', 'East', 'West'], 100),
    'sales': np.random.randint(500, 2000, 100),
    'units': np.random.randint(1, 10, 100),
    'category': np.random.choice(['Electronics', 'Accessories'], 100)
}

df = pd.DataFrame(data)
df['month'] = df['date'].dt.month
df['quarter'] = df['date'].dt.quarter
print("Original DataFrame (first few rows):")
print(df.head())

# Basic pivot table: Sales by product and region
pivot1 = pd.pivot_table(df, values='sales', index='product', columns='region')
print("\nSales by product and region:")
print(pivot1)

# More complex pivot table with multiple values and aggregation functions
pivot2 = pd.pivot_table(
    df, 
    values=['sales', 'units'], 
    index=['category', 'product'], 
    columns='region',
    aggfunc={'sales': 'sum', 'units': 'sum'},
    fill_value=0  # Replace NaN with 0
)
print("\nSales and units by category, product, and region:")
print(pivot2)

# Pivot table with multiple aggregation functions for the same value
pivot3 = pd.pivot_table(
    df,
    values='sales',
    index='product',
    columns='month',
    aggfunc=['sum', 'mean', 'count'],
    margins=True,  # Add row and column totals
    margins_name='Total'
)
print("\nSales by product and month with multiple aggregations and totals:")
print(pivot3)
````


The `pivot_table()` function is highly flexible, allowing you to specify which columns to use as row indices, column indices, and values, as well as how to aggregate the values.

#### Customizing Pivot Tables

You can further customize pivot tables to make them more useful for your analysis:



````python
# Customize display format
pd.options.display.float_format = '${:,.2f}'.format

# Pivot table with custom aggregation functions
def sales_per_unit(x):
    return x.sum() / df.loc[x.index, 'units'].sum()

pivot4 = pd.pivot_table(
    df,
    values='sales',
    index='category',
    columns='quarter',
    aggfunc={
        'sales': ['sum', 'mean', sales_per_unit]
    }
)
print("\nSales by category and quarter with custom aggregation:")
print(pivot4)

# Reset display format
pd.reset_option('display.float_format')

# Pivot table with hierarchical columns
pivot5 = pd.pivot_table(
    df,
    values=['sales', 'units'],
    index=['category'],
    columns=['quarter', 'month'],
    aggfunc='sum',
    fill_value=0
)
print("\nSales and units by category, quarter, and month:")
print(pivot5)

# Flatten a hierarchical pivot table
pivot5_flat = pivot5.reset_index()
print("\nFlattened pivot table (first few rows):")
print(pivot5_flat.head())
````


Pivot tables often have hierarchical indices and/or columns, which can be manipulated using methods like `reset_index()`, `stack()`, and `unstack()`.

#### Cross-Tabulation

Cross-tabulation (crosstab) is a special case of pivot table that counts the frequency of combinations of values between two or more categorical variables:



````python
# Create a sample dataset for cross-tabulation
survey_data = {
    'gender': np.random.choice(['Male', 'Female', 'Other'], 100),
    'age_group': np.random.choice(['18-24', '25-34', '35-44', '45-54', '55+'], 100),
    'response': np.random.choice(['Yes', 'No', 'Maybe'], 100),
    'satisfaction': np.random.randint(1, 6, 100)  # 1-5 scale
}
survey_df = pd.DataFrame(survey_data)

# Basic cross-tabulation: Gender vs. Response
ct1 = pd.crosstab(survey_df['gender'], survey_df['response'])
print("\nCross-tabulation of Gender vs. Response:")
print(ct1)

# Cross-tabulation with percentages (normalized)
ct2 = pd.crosstab(
    survey_df['gender'], 
    survey_df['response'],
    normalize='index'  # Row percentages
)
print("\nCross-tabulation with row percentages:")
print(ct2.round(2))

# Cross-tabulation with multiple indices/columns
ct3 = pd.crosstab(
    [survey_df['gender'], survey_df['age_group']],  # Row indices
    survey_df['response'],  # Column index
    margins=True,  # Add row and column totals
    margins_name='Total'
)
print("\nCross-tabulation with multiple row indices and totals:")
print(ct3)

# Cross-tabulation with aggregation
ct4 = pd.crosstab(
    survey_df['gender'],
    survey_df['age_group'],
    values=survey_df['satisfaction'],
    aggfunc='mean'
)
print("\nAverage satisfaction by gender and age group:")
print(ct4.round(2))
````


Cross-tabulation is particularly useful for survey data analysis, customer segmentation, and any scenario where you need to understand the relationship between categorical variables.

#### Data Reshaping with Pivot and Melt

While pivot tables aggregate data, sometimes you simply want to reshape your data without aggregation. The `pivot()` and `melt()` functions help with this:



````python
# Create a sample dataset for reshaping
student_scores = {
    'student_id': [1, 1, 1, 2, 2, 2, 3, 3, 3],
    'subject': ['Math', 'Science', 'English', 'Math', 'Science', 'English', 'Math', 'Science', 'English'],
    'score': [85, 90, 80, 70, 80, 75, 95, 85, 90]
}
scores_df = pd.DataFrame(student_scores)
print("\nOriginal student scores:")
print(scores_df)

# Pivot: Convert from long to wide format
scores_wide = scores_df.pivot(index='student_id', columns='subject', values='score')
print("\nScores in wide format:")
print(scores_wide)

# Add a mean score column
scores_wide['mean_score'] = scores_wide.mean(axis=1)
print("\nWide format with mean score:")
print(scores_wide)

# Melt: Convert from wide back to long format
scores_long = pd.melt(
    scores_wide.reset_index(),
    id_vars='student_id',
    value_vars=['Math', 'Science', 'English', 'mean_score'],
    var_name='subject',
    value_name='score'
)
print("\nBack to long format with melt:")
print(scores_long)
````


`pivot()` transforms data from a long format (multiple rows per entity) to a wide format (one row per entity), while `melt()` does the opposite.

#### Stack and Unstack for Hierarchical Indices

The `stack()` and `unstack()` methods provide another way to reshape data with hierarchical indices:



````python
# Create a MultiIndex DataFrame
multi_idx = pd.MultiIndex.from_product([['A', 'B'], ['X', 'Y']], names=['first', 'second'])
values = np.random.randint(0, 10, 4)
hier_df = pd.DataFrame({'value': values}, index=multi_idx)
print("\nDataFrame with hierarchical index:")
print(hier_df)

# Unstack: Move the second level of index to columns
unstacked = hier_df.unstack(level='second')
print("\nUnstacked DataFrame (second level as columns):")
print(unstacked)

# Stack: Move columns to a new index level
restacked = unstacked.stack()
print("\nRe-stacked DataFrame:")
print(restacked)

# Using stack/unstack with pivot tables
pivot6 = pd.pivot_table(
    df,
    values='sales',
    index=['category', 'product'],
    columns='region',
    aggfunc='sum'
)
print("\nPivot table with hierarchical index:")
print(pivot6)

# Unstack the first level of the index
unstacked_pivot = pivot6.unstack(level=0)
print("\nUnstacked pivot table (category level as columns):")
print(unstacked_pivot)

# Stack all column levels into the index
fully_stacked = pivot6.stack()
print("\nFully stacked pivot table:")
print(fully_stacked.head(8))
````


`stack()` and `unstack()` are particularly useful for working with the hierarchical indices that often result from pivot tables and groupby operations.

#### Advanced Pivot Table Techniques

Let's explore some more advanced techniques with pivot tables:



````python
# Create a more detailed pivot table
advanced_pivot = pd.pivot_table(
    df,
    values=['sales', 'units'],
    index=['category', 'product'],
    columns=['quarter', 'month'],
    aggfunc={'sales': 'sum', 'units': 'sum'},
    fill_value=0
)
print("\nAdvanced pivot table:")
print(advanced_pivot)

# Access specific values from the pivot table
q1_laptop_sales = advanced_pivot.loc[('Electronics', 'Laptop'), ('sales', 1)]
print(f"\nQ1 Laptop sales: {q1_laptop_sales}")

# Slice the pivot table
q1_sales = advanced_pivot.xs(1, level='quarter', axis=1)['sales']
print("\nQ1 sales for all products:")
print(q1_sales)

# Calculate derived metrics
advanced_pivot['price_per_unit'] = advanced_pivot['sales'] / advanced_pivot['units']
print("\nPivot table with price per unit:")
print(advanced_pivot['price_per_unit'].head())

# Calculate growth rates
sales_by_month = pd.pivot_table(
    df,
    values='sales',
    index='product',
    columns='month',
    aggfunc='sum',
    fill_value=0
)

sales_by_month['growth_1_to_2'] = (sales_by_month[2] - sales_by_month[1]) / sales_by_month[1] * 100
print("\nSales by month with growth rate:")
print(sales_by_month)
````


These techniques demonstrate how to extract specific data from pivot tables, add calculated fields, and derive new insights from the pivoted data.

### ⚠️ Common Mistakes with Pivot Tables and Cross-Tabulation

* **Forgetting to specify an aggregation function**: When multiple values exist for a combination of index and columns, you need an aggregation function to resolve the conflict.
* **Using pivot() when pivot_table() is needed**: If your data has multiple values for the same combination of index and columns, `pivot()` will raise an error - use `pivot_table()` instead.
* **Not handling missing values**: Pivot tables often contain NaN values by default, which can affect further calculations if not handled properly.
* **Losing track of hierarchical indices**: The resulting tables often have multiple index or column levels, which require special methods for access and manipulation.
* **Incorrect normalization in crosstab**: Be careful with the `normalize` parameter - 'index' normalizes rows, 'columns' normalizes columns, and 'all' normalizes the entire table.

````python
# Count of sales transactions by region and channel
txn_count = pd.crosstab(df['region'], df['channel'])
print("Transaction count by region and channel:")
print(txn_count)

# Percentage distribution of sales across channels for each region (normalize by row)
channel_pct_by_region = pd.crosstab(
    df['region'], 
    df['channel'],
    normalize='index'
).round(2) * 100  # Convert to percentage
print("\nPercentage distribution of channels by region:")
print(channel_pct_by_region)

# Percentage distribution of sales across regions for each channel (normalize by column)
region_pct_by_channel = pd.crosstab(
    df['region'], 
    df['channel'],
    normalize='columns'
).round(2) * 100  # Convert to percentage
print("\nPercentage distribution of regions by channel:")
print(region_pct_by_channel)

# Relationship between promotions and customer segments, with totals
promo_segment_ct = pd.crosstab(
    df['promotion'], 
    df['customer_segment'],
    margins=True,
    margins_name='Total'
)
print("\nPromotion usage by customer segment:")
print(promo_segment_ct)
````


*Multiple Aggregations:*



````python
# For each product, calculate sum, mean, min, and max of sales by region
product_region_agg = pd.pivot_table(
    df,
    values='sales',
    index='product',
    columns='region',
    aggfunc=['sum', 'mean', 'min', 'max']
)
print("Multiple sales aggregations by product and region:")
print(product_region_agg)

# For each month and channel, calculate sum of sales and count of transactions
month_channel_agg = pd.pivot_table(
    df,
    values='sales',
    index='month',
    columns='channel',
    aggfunc=['sum', 'count']
)
print("\nSales sum and count by month and channel:")
print(month_channel_agg)

# For each category and customer segment, calculate mean satisfaction and sum of returns
cat_segment_agg = pd.pivot_table(
    df,
    values=['satisfaction', 'returns'],
    index='category',
    columns='customer_segment',
    aggfunc={'satisfaction': 'mean', 'returns': 'sum'}
)
print("\nSatisfaction and returns by category and customer segment:")
print(cat_segment_agg)

# For each region, calculate sum of sales, count of transactions, and mean profit margin
region_multi_agg = pd.pivot_table(
    df,
    values=['sales', 'profit_margin'],
    index='region',
    aggfunc={'sales': ['sum', 'count'], 'profit_margin': 'mean'}
)
print("\nMultiple metrics by region:")
print(region_multi_agg)
````


*Hierarchical Indices:*



````python
# Pivot table with products and categories as hierarchical row indices
hier_product_pivot = pd.pivot_table(
    df,
    values='sales',
    index=['category', 'product'],
    columns='region',
    aggfunc='sum'
)
print("Sales by category, product, and region (hierarchical):")
print(hier_product_pivot)

# Pivot table with month and quarter as hierarchical column indices
hier_time_pivot = pd.pivot_table(
    df,
    values='sales',
    index='channel',
    columns=['quarter', 'month'],
    aggfunc='sum'
)
print("\nSales by channel, quarter, and month (hierarchical):")
print(hier_time_pivot)

# Extract all data for a specific product using xs()
product_a_data = hier_product_pivot.xs('Product A', level='product')
print("\nAll data for Product A:")
print(product_a_data)

# Use .loc[] with a tuple to select specific hierarchical data
category1_product_b = hier_product_pivot.loc[('Category 1', 'Product B')]
print("\nData for Product B in Category 1:")
print(category1_product_b)
````


*Data Reshaping with pivot() and melt():*



````python
# First, create a subset of data for easier visualization
subset_df = df[['product', 'region', 'sales']].head(30)

# Use pivot() to convert from long to wide format
# This only works if there are no duplicate product-region combinations
# Let's ensure unique combinations by taking the first occurrence of each
unique_subset = subset_df.drop_duplicates(['product', 'region'])
wide_format = unique_subset.pivot(index='product', columns='region', values='sales')
print("Data in wide format:")
print(wide_format)

# Use melt() to convert back to long format
long_format = pd.melt(
    wide_format.reset_index(),
    id_vars='product',
    value_vars=['North', 'South', 'East', 'West'],
    var_name='region',
    value_name='sales'
)
print("\nData back in long format:")
print(long_format.head(10))

# Create a wide format table of satisfaction scores by product and channel
satisfaction_wide = df.pivot_table(
    values='satisfaction',
    index='product',
    columns='channel',
    aggfunc='mean'
).round(1)
print("\nSatisfaction scores by product and channel (wide format):")
print(satisfaction_wide)

# Monthly sales for each product as a time series
monthly_sales_by_product = df.pivot_table(
    values='sales',
    index='month',
    columns='product',
    aggfunc='sum'
)
print("\nMonthly sales for each product (time series):")
print(monthly_sales_by_product)
````


*Working with stacked and unstacked formats:*



````python
# Create a pivot table with hierarchical index
hier_pivot = pd.pivot_table(
    df,
    values='sales',
    index=['category', 'product'],
    columns=['channel', 'quarter'],
    aggfunc='sum'
)
print("Hierarchical pivot table:")
print(hier_pivot.head())

# Unstack one level of the index (product)
unstacked_product = hier_pivot.unstack(level='product')
print("\nPivot table with product level unstacked:")
print(unstacked_product.head())

# Stack one level of the columns (quarter)
stacked_quarter = hier_pivot.stack(level='quarter')
print("\nPivot table with quarter level stacked:")
print(stacked_quarter.head())

# Unstack to a completely flat wide-format table
fully_unstacked = hier_pivot.unstack(level=['category', 'product'])
print("\nFully unstacked table (flat wide format):")
print(fully_unstacked.head())

# Stack all levels to get a fully stacked Series
fully_stacked = hier_pivot.stack(level=['channel', 'quarter'])
print("\nFully stacked Series:")
print(fully_stacked.head(10))
````


*Calculated Fields in Pivot Tables:*



````python
# Pivot table of sales by product and month, then add month-over-month growth
monthly_product_sales = pd.pivot_table(
    df,
    values='sales',
    index='product',
    columns='month',
    aggfunc='sum'
)

# Calculate month-over-month growth
for month in range(2, 13):  # Assuming months 1-12
    if month in monthly_product_sales.columns:
        prev_month = month - 1
        if prev_month in monthly_product_sales.columns:
            col_name = f"{month-1}_to_{month}_growth"
            monthly_product_sales[col_name] = ((monthly_product_sales[month] - monthly_product_sales[prev_month]) / 
                                              monthly_product_sales[prev_month] * 100).round(1)

print("Monthly sales with growth rates:")
# Select a subset of columns for cleaner display
display_cols = [1, 2, '1_to_2_growth', 3, '2_to_3_growth']
print(monthly_product_sales[display_cols])

# Pivot table of units and sales by category, then add avg unit price
category_units_sales = pd.pivot_table(
    df,
    values=['units', 'sales'],
    index='category',
    aggfunc='sum'
)
category_units_sales['avg_unit_price'] = (category_units_sales['sales'] / category_units_sales['units']).round(2)
print("\nCategory summary with calculated avg unit price:")
print(category_units_sales)

# Pivot table of sales by region and channel, then add column percentages
region_channel_sales = pd.pivot_table(
    df,
    values='sales',
    index='region',
    columns='channel',
    aggfunc='sum'
)
# Calculate each region's percentage contribution to channel total
for channel in region_channel_sales.columns:
    channel_total = region_channel_sales[channel].sum()
    col_name = f"{channel}_pct"
    region_channel_sales[col_name] = (region_channel_sales[channel] / channel_total * 100).round(1)

print("\nSales by region and channel with percentage contribution:")
print(region_channel_sales)

# Pivot table of profits by product and quarter, add profitability ratio
product_quarter_profit = pd.pivot_table(
    df,
    values='profit',
    index='product',
    columns='quarter',
    aggfunc='sum'
)
# Calculate average profit across products for each quarter
avg_profit_by_quarter = product_quarter_profit.mean()
# Calculate profitability ratio
for quarter in product_quarter_profit.columns:
    ratio_col = f"Q{quarter}_ratio"
    product_quarter_profit[ratio_col] = (product_quarter_profit[quarter] / avg_profit_by_quarter[quarter]).round(2)

print("\nProduct profitability by quarter with relative ratios:")
print(product_quarter_profit)
````

## **Chapter 3.4:** Coding Challenge: Quality Analysis 2/3

### 🏆 Analyzing Quality with Aggregation Methods

We are going to use the dataset from the previous coding challenge. It contains metrics from an injection molding process, including melt temperature, mold temperature, cycle times, forces, pressures, volumes, and a quality indicator. The goal is to analyze the dataset to uncover insights into the manufacturing process and quality outcomes.

**Challenge objectives:**

1. Step: Filter the dataset to include only cycles where the `'Melt temperature'` is greater than the average melt temperature.

2. Step: Slice the dataset to include only the columns related to temperatures (`'Melt temperature'`, `'Mold temperature'`) and `quality`.

3. Step: Group the dataset by `'quality'` and calculate the average `'Mold temperature'` and `'Melt temperature'` for each quality group.

4. Step: For each `'quality'` group, find the maximum `'Specific injection pressure peak value'` (APVs).

5. Step: Create a pivot table showing the average `'Cycle time' (ZUx - Cycle time) for each combination of 'quality'` and `'Mold temperature'` (rounded to the nearest integer).

6. Step: Use cross-tabulation to summarize the count of cycles by `'quality'` and the `'Screw position at the end of hold pressure'` (CPn) binned into categories (e.g., <9, 9-10, >10).



````python
# Step 0: Import a dataset into a Pandas DataFrame

# Sample code template
import pandas as pd
import numpy as np

# Import the dataset
df = pd.read_csv('../data/data.csv', delimiter=";")

df.head()
````

````python
# Step 1: Filter the dataset to include only cycles where the `'Melt temperature'` is greater than the average melt temperature.

avg_melt_temp = df['Melt temperature'].mean()
filtered_df = df[df['Melt temperature'] > avg_melt_temp]

filtered_df.head()
````


This step filters rows based on whether their `'Melt temperature'` is above the column's average. This step focuses on cycles with higher-than-average melt temperatures.



````python
# 2. Step: Slice the dataset to include only the columns related to temperatures (`'Melt temperature'`, `'Mold temperature'`) and `quality`.

sliced_df = filtered_df[['Melt temperature', 'Mold temperature', 'quality']]
sliced_df.head()
````


This step narrows down the dataset to only include the columns that are immediately relevant to the subsequent analysis, enhancing clarity and performance.



````python
# 3. Step: Group the dataset by `'quality'` and calculate the average `'Mold temperature'` and `'Melt temperature'` for each quality group.

avg_temp_by_quality = df.groupby('quality')[['Mold temperature', 'Melt temperature']].mean()

avg_temp_by_quality
````

````python
# 4. Step: For each `'quality'` group, find the maximum `'Specific injection pressure peak value'` (APVs).

max_apvs_by_quality = df.groupby('quality')['APVs - Specific injection pressure peak value'].max()

max_apvs_by_quality
````


Step 3 and Step 4 leverage the `groupby()` method to segment the data by `'quality'`, enabling detailed analysis of temperature averages and maximum injection pressures within each quality category.



````python
# 5. Step: Create a pivot table showing the average `'Cycle time' (ZUx - Cycle time) for each combination of 'quality'` and `'Mold temperature'` (rounded to the nearest integer).

df['Mold temperature rounded'] = df['Mold temperature'].round()
pivot_table = df.pivot_table(values='ZUx - Cycle time', index='quality', columns='Mold temperature rounded', aggfunc='mean')

pivot_table
````


This step creates a pivot table that averages `'Cycle time'` across combinations of `'quality'` and `'Mold temperature'`, offering insights into how these variables interact.



````python
# 6. Step: Use cross-tabulation to summarize the count of cycles by `'quality'` and the `'Screw position at the end of hold pressure'` (CPn) binned into categories (e.g., <9, 9-10, >10).

# Binning the CPn values
df['CPn_category'] = pd.cut(df['CPn - Screw position at the end of hold pressure'], bins=[-np.inf, 9, 10, np.inf], labels=['<9', '9-10', '>10'])

# Cross-tabulation
cross_tab = pd.crosstab(index=df['quality'], columns=df['CPn_category'])

cross_tab
````


This step utilizes `pd.cut()` to categorize `'Screw position at the end of hold pressure'` and `pd.crosstab()` to summarize the distribution of these categories by `'quality'`, providing a clear overview of cycle counts across different operational settings and outcomes.

## 📚 Key Terms

- **Boolean Indexing**: Technique for filtering data using conditions that evaluate to True/False (e.g., df[df['temperature'] > 200])
- **GroupBy**: The operation of splitting data into groups based on criteria and applying functions to each group independently
- **Aggregation Functions**: Methods like sum(), mean(), count() that calculate summary statistics from grouped data
- **Pivot Table**: A data summarization tool that restructures and aggregates data to analyze the relationship between variables
- **Cross-Tabulation**: A special case of pivot table that counts the frequency of combinations between categorical variables
- **Multi-Index**: Hierarchical indexing structure that enables representation of higher-dimensional data in a DataFrame
- **Query Method**: An alternative syntax for boolean indexing with more readable expressions (e.g., df.query("temperature > 200"))
- **Transform**: A GroupBy operation that returns results with the same shape as the input, useful for calculations like percent of total
- **Stack/Unstack**: Methods for pivoting levels of a hierarchical index into columns or vice versa
- **Melt**: Function to convert data from wide to long format, restructuring multiple columns into rows
