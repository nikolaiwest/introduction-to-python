## **Chapter 3.1:** Understanding Figures and Axes

### 🔍 The Structural Framework of Matplotlib

To master customization in Matplotlib, you must first understand its **fundamental organizational structure**. At the heart of this structure are two key concepts: *Figures* and *Axes*. These components form the hierarchical framework upon which all Matplotlib visualizations are built.

This hierarchical approach gives Matplotlib its flexibility and power, allowing for both simple plots and complex, multi-panel visualizations. By understanding this structure, you gain precise control over every element of your visualizations.

### 🧩 Important Components

* **Figure:** The top-level container that holds everything visible in the visualization
* **Axes:** The actual plotting area where data is displayed (a single Figure can contain multiple Axes)
* **Object-Oriented vs. State-Based Interfaces:** Two different approaches to creating and manipulating plots
* **Figure Dimensions:** Controls for size, resolution, and aspect ratio

    | Component | Description | Key Methods |
    |-----------|-------------|-------------|
    | Figure | Overall container for the visualization | `plt.figure()`, `fig.savefig()` |
    | Axes | Individual plotting area with its own coordinate system | `fig.add_subplot()`, `ax.plot()`, `ax.set_title()` |
    | Subplot | A way to create and position multiple Axes in a grid | `plt.subplots()`, `plt.subplot()` |
    | Figure dimensions | Controls for size and resolution | `figsize=(width, height)`, `dpi=resolution` |

### 📘 Main Content

#### Figures and Axes Hierarchy

In Matplotlib, visualizations follow a clear hierarchical structure:

1. The **Figure** is the outermost container. It represents the entire visualization window or page and can contain one or more Axes objects.

2. The **Axes** are the actual plotting areas, each with its own coordinate system. When we talk about "a plot," we're typically referring to a single Axes.

This hierarchical approach allows for complex visualization layouts while maintaining organized control over each component:



```python
import matplotlib.pyplot as plt
import numpy as np

# Create a figure (the container)
fig = plt.figure(figsize=(10, 6))

# Add an axes to the figure (the actual plot)
ax = fig.add_subplot(111)  # 1 row, 1 column, 1st subplot

# Generate some data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Plot the data on our axes
ax.plot(x, y)

# Customize the axes
ax.set_title('A Simple Sine Wave')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')

# Show the figure with its axes
plt.show()
```


This code demonstrates the basic hierarchy: we create a Figure, add an Axes to it, and then plot data on the Axes. This approach gives us explicit control over both the container (Figure) and the plotting area (Axes).

#### The Two Interfaces: State-Based vs. Object-Oriented

Matplotlib offers two interfaces for creating and manipulating plots:

1. **State-Based Interface (pyplot)**: This approach is modeled after MATLAB's plotting functions and maintains an internal state that tracks the "current" Figure and Axes.



```python
# State-based approach (pyplot interface)
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

# All of these commands implicitly use the "current" figure and axes
plt.figure(figsize=(10, 6))
plt.plot(x, y)
plt.title('A Simple Sine Wave')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.grid(True)
plt.show()
```


2. **Object-Oriented Interface**: This approach works directly with Figure and Axes objects, providing more explicit control.



```python
# Object-oriented approach
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create objects explicitly
fig, ax = plt.subplots(figsize=(10, 6))

# Work directly with the axes object
ax.plot(x, y)
ax.set_title('A Simple Sine Wave')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.grid(True)

plt.show()
```


> **Note:** While the state-based interface is convenient for simple plots, the object-oriented interface is recommended for more complex visualizations. It provides clearer code, better control, and avoids potential confusion with the "current" figure state.

#### ⚠️ Common Mistakes with Figures and Axes

* **Mixing interfaces**: Inconsistently switching between state-based and object-oriented approaches can lead to confusion and unexpected results
* **Not understanding the hierarchy**: Attempting to apply Figure methods to Axes or vice versa
* **Overwriting figures**: Creating a new figure without displaying the previous one first
* **Failing to clear state**: Not clearing the state between plots when using the state-based interface in interactive sessions

#### Controlling Figure Dimensions

The physical dimensions of your visualization are crucial for readability and impact. Matplotlib offers several ways to control these dimensions:



```python
# Control figure size (width, height in inches)
fig = plt.figure(figsize=(10, 6))

# Control resolution (dots per inch)
fig = plt.figure(dpi=300)

# Both together
fig = plt.figure(figsize=(10, 6), dpi=300)

# With subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 8), dpi=100)
```


Some key considerations for figure dimensions:

- **Output medium**: Adjust dimensions based on where the visualization will appear (screen, print, presentation)
- **Aspect ratio**: Choose a ratio appropriate for your data (e.g., square for correlation matrices, wider for time series)
- **Resolution**: Use higher DPI (300+) for publication-quality figures, lower (72-150) for screens
- **Balance**: Ensure there's enough space for all elements including titles, labels, and legends

#### Creating and Saving Figures

When creating professional visualizations, you'll often need to save them for reports, publications, or presentations:



```python
# Create a figure with specific dimensions
fig, ax = plt.subplots(figsize=(10, 6))

# Plot some data
x = np.linspace(0, 10, 100)
y = np.sin(x)
ax.plot(x, y)

# Customize as needed
ax.set_title('Sine Wave', fontsize=16)
ax.set_xlabel('X Axis', fontsize=12)
ax.set_ylabel('Y Axis', fontsize=12)

# Save the figure with high resolution
fig.savefig('sine_wave.png', dpi=300, bbox_inches='tight')

# Save in different formats
fig.savefig('sine_wave.pdf')  # PDF for publication
fig.savefig('sine_wave.svg')  # SVG for web or editing
```


The `bbox_inches='tight'` parameter is particularly useful as it automatically adjusts the figure boundaries to include all elements, preventing labels or titles from being cut off.

*Basic Figures and Axes:*



```python
# 1-3. Create a figure, add axes, and plot y = x²
import numpy as np
import matplotlib.pyplot as plt

# Create figure with specific dimensions
fig = plt.figure(figsize=(10, 6))

# Add an axes to the figure
ax = fig.add_subplot(111)

# Generate data and plot
x = np.linspace(-5, 5, 100)
y = x**2
ax.plot(x, y)

# Add title and labels
ax.set_title('Quadratic Function', fontsize=14)
ax.set_xlabel('X Axis', fontsize=12)
ax.set_ylabel('Y Axis', fontsize=12)

plt.show()
```


*Comparing Interfaces:*



```python
# 4. Same plot using pyplot interface
plt.figure(figsize=(10, 6))
x = np.linspace(-5, 5, 100)
y = x**2
plt.plot(x, y)
plt.title('Quadratic Function (pyplot interface)', fontsize=14)
plt.xlabel('X Axis', fontsize=12)
plt.ylabel('Y Axis', fontsize=12)
plt.show()

# 5. Same plot using plt.subplots()
fig, ax = plt.subplots(figsize=(10, 6))
x = np.linspace(-5, 5, 100)
y = x**2
ax.plot(x, y)
ax.set_title('Quadratic Function (subplots)', fontsize=14)
ax.set_xlabel('X Axis', fontsize=12)
ax.set_ylabel('Y Axis', fontsize=12)
plt.show()
```


*Advanced Figure Control:*



```python
# 6. High-resolution figure saved as PNG
fig, ax = plt.subplots(figsize=(10, 6), dpi=300)
x = np.linspace(-5, 5, 100)
y = x**2
ax.plot(x, y)
ax.set_title('High Resolution Plot', fontsize=14)
ax.set_xlabel('X Axis', fontsize=12)
ax.set_ylabel('Y Axis', fontsize=12)
fig.savefig('high_res_plot.png', dpi=300, bbox_inches='tight')
plt.show()

# 7. Different figure sizes and aspect ratios
for dimensions in [(8, 6), (6, 6), (12, 4)]:
    fig, ax = plt.subplots(figsize=dimensions)
    ax.plot(x, y)
    ax.set_title(f'Figure size: {dimensions}', fontsize=14)
    ax.set_xlabel('X Axis', fontsize=12)
    ax.set_ylabel('Y Axis', fontsize=12)
    plt.show()
```


*Multi-Component Visualization:*



```python
# 8-10. Two horizontally arranged axes with figure title
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Generate data
x = np.linspace(0, 2*np.pi, 100)
sin_y = np.sin(x)
cos_y = np.cos(x)

# Plot on the first axes
ax1.plot(x, sin_y, 'b-')
ax1.set_title('Sine Function')
ax1.set_xlabel('X Axis')
ax1.set_ylabel('sin(x)')

# Plot on the second axes
ax2.plot(x, cos_y, 'r-')
ax2.set_title('Cosine Function')
ax2.set_xlabel('X Axis')
ax2.set_ylabel('cos(x)')

# Add a title to the overall figure
fig.suptitle('Trigonometric Functions', fontsize=16)

# Adjust layout
plt.tight_layout()
plt.subplots_adjust(top=0.88)  # Make room for the suptitle

# Save in multiple formats
fig.savefig('trig_functions.png', dpi=300, bbox_inches='tight')
fig.savefig('trig_functions.pdf', bbox_inches='tight')

plt.show()
```

## **Chapter 3.2:** Working with Multiple Figures and Axes

### 🔍 Creating Complex Visualization Layouts

The ability to organize multiple plots in a single visualization is essential for comparative analysis and comprehensive data presentation. Matplotlib provides powerful tools for creating complex layouts through its **subplot system**, which allows you to arrange multiple Axes within a Figure in a variety of configurations.

In this section, we'll explore how to create everything from simple grid layouts to more complex, non-uniform arrangements. We'll also cover techniques for coordinating scales and sharing information across multiple plots to create cohesive, multi-panel visualizations.

### 🧩 Important Components

* **Subplot Creation:** Methods for adding multiple Axes to a Figure
* **Grid Layouts:** Arranging plots in rows and columns
* **Axes Positioning:** Controlling the exact placement and size of each Axes
* **Axes Sharing:** Techniques for coordinating axes limits and scales
* **Layout Adjustment:** Tools for managing spacing and preventing overlaps

| Method | Description | Usage |
|--------|-------------|-------|
| `plt.subplots()` | Create a grid of subplots | `fig, axs = plt.subplots(rows, cols)` |
| `plt.subplot()` | Create a single subplot | `plt.subplot(rows, cols, index)` |
| `plt.subplot2grid()` | Create a subplot in a grid | `plt.subplot2grid((rows, cols), (row, col), rowspan=1, colspan=1)` |
| `GridSpec` | Flexible grid layout | `gs = GridSpec(rows, cols)` |
| `tight_layout()` | Adjust spacing between plots | `plt.tight_layout()` |

### 📘 Main Content

#### Creating Basic Subplot Grids

The most common way to create multiple Axes is using the `plt.subplots()` function, which creates a grid of subplots:



```python
import matplotlib.pyplot as plt
import numpy as np

# Create a 2x2 grid of subplots
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Generate some data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.exp(-x/5)

# Plot on each subplot
axs[0, 0].plot(x, y1)
axs[0, 0].set_title('Sine Wave')

axs[0, 1].plot(x, y2)
axs[0, 1].set_title('Cosine Wave')

axs[1, 0].plot(x, y3)
axs[1, 0].set_title('Tangent Wave')
axs[1, 0].set_ylim(-5, 5)  # Limit y-axis for better visibility

axs[1, 1].plot(x, y4)
axs[1, 1].set_title('Exponential Decay')

# Adjust spacing between subplots
plt.tight_layout()

plt.show()
```


In this example:
- We create a 2×2 grid of subplots using `plt.subplots(2, 2)`
- Each subplot is accessed using the indexing notation `axs[row, column]`
- We plot different functions on each subplot
- `plt.tight_layout()` automatically adjusts the spacing to prevent overlap

#### Working with Flattened Axes Arrays

For more complex grid layouts, it can be convenient to work with a flattened array of Axes:



```python
# Create a 2x3 grid of subplots
fig, axs = plt.subplots(2, 3, figsize=(12, 8))

# Flatten the 2D array of axes for easier indexing
axs = axs.flatten()

# Generate some data
x = np.linspace(0, 10, 100)
functions = [
    np.sin,
    np.cos,
    np.tan,
    lambda x: x**2,
    lambda x: np.sqrt(x),
    lambda x: np.exp(-x/5)
]
titles = ['Sine', 'Cosine', 'Tangent', 'Quadratic', 'Square Root', 'Exponential Decay']

# Plot each function on its respective subplot
for i, (func, title) in enumerate(zip(functions, titles)):
    axs[i].plot(x, func(x))
    axs[i].set_title(title)
    axs[i].grid(True)

plt.tight_layout()
plt.show()
```


Using `flatten()` or `ravel()` converts the 2D array of Axes into a 1D array, allowing us to use a simple loop to populate all subplots. This approach is especially useful for larger grid layouts or when the number of plots might vary.

#### ⚠️ Common Mistakes with Subplots

* **Indexing confusion**: Remember that `axs[0, 1]` refers to the subplot in the first row (0) and second column (1)
* **Forgetting to adjust layout**: Not using `tight_layout()` can result in overlapping titles and labels
* **Not accounting for different scales**: Some plots might need adjusted limits for proper visibility
* **Inconsistent styling**: Maintaining visual consistency across multiple subplots is important for readability

#### Sharing Axes Across Subplots

When comparing related data, it's often helpful to share axes across subplots:



```python
# Create subplots with shared axes
fig, axs = plt.subplots(2, 2, figsize=(10, 8), 
                       sharex=True,    # Share x-axes across columns
                       sharey=True)    # Share y-axes across rows

# Generate some related data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.sin(x + np.pi/4)
y3 = np.sin(x + np.pi/2)
y4 = np.sin(x + 3*np.pi/4)

# Plot on each subplot
axs[0, 0].plot(x, y1)
axs[0, 0].set_title('Phase = 0')

axs[0, 1].plot(x, y2)
axs[0, 1].set_title('Phase = π/4')

axs[1, 0].plot(x, y3)
axs[1, 0].set_title('Phase = π/2')

axs[1, 1].plot(x, y4)
axs[1, 1].set_title('Phase = 3π/4')

# Add a common label for the shared axes
fig.text(0.5, 0.04, 'X Axis (shared)', ha='center')
fig.text(0.04, 0.5, 'Y Axis (shared)', va='center', rotation='vertical')

plt.tight_layout()
plt.show()
```


In this example:
- `sharex=True` means all subplots in the same column share the same x-axis
- `sharey=True` means all subplots in the same row share the same y-axis
- We use `fig.text()` to add common labels for the shared axes

Other sharing options include:
- `sharex='col'` or `sharey='row'` to share axes only within columns or rows
- `sharex=True` or `sharey=True` to share across all subplots

Shared axes offer several advantages:

2. **Space efficiency**: Labels and ticks need only be shown once for the shared dimension
3. **Interactivity**: When using interactive plots, zooming and panning affects all shared axes together

#### Creating Non-Uniform Layouts with GridSpec

For more complex layouts, Matplotlib's `GridSpec` offers greater flexibility:



```python
import matplotlib.gridspec as gridspec

# Create a figure
fig = plt.figure(figsize=(12, 8))

# Create a custom grid layout: 3 rows, 3 columns
gs = gridspec.GridSpec(3, 3)

# Add subplots at specific grid positions
ax1 = fig.add_subplot(gs[0, :])         # Span all columns in first row
ax2 = fig.add_subplot(gs[1, 0:2])       # Span first two columns in second row
ax3 = fig.add_subplot(gs[1:, 2])        # Span last two rows in last column
ax4 = fig.add_subplot(gs[2, 0])         # Single cell
ax5 = fig.add_subplot(gs[2, 1])         # Single cell

# Plot some data on each subplot
ax1.plot(np.random.rand(100))
ax1.set_title('Subplot spanning the entire first row')

ax2.plot(np.random.rand(100))
ax2.set_title('Subplot spanning 2 columns')

ax3.plot(np.random.rand(100))
ax3.set_title('Subplot\nspanning\n2 rows')

ax4.plot(np.random.rand(100))
ax4.set_title('Single cell')

ax5.plot(np.random.rand(100))
ax5.set_title('Single cell')

plt.tight_layout()
plt.show()
```


`GridSpec` allows you to:
- Create grid cells of different sizes
- Span multiple rows or columns with a single subplot
- Create complex, magazine-style layouts
- Precisely control the position and size of each subplot

#### Adjusting Spacing Between Subplots

Fine control over the spacing between subplots is often necessary for professional layouts:



```python
# Create a grid of subplots
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Plot some data
for ax in axs.flat:
    ax.plot(np.random.rand(100))

# Adjust the spacing
plt.subplots_adjust(
    left=0.1,    # Left margin
    right=0.95,  # Right margin
    bottom=0.1,  # Bottom margin
    top=0.9,     # Top margin
    wspace=0.3,  # Width spacing between subplots
    hspace=0.3   # Height spacing between subplots
)

plt.show()
```


While `tight_layout()` works well in many cases, `subplots_adjust()` gives you precise control when needed.

```python
# 4-5. Create a 2×3 grid and use flattened array for trigonometric functions
import numpy as np
import matplotlib.pyplot as plt

# Create a 2×3 grid of subplots
fig, axs = plt.subplots(2, 3, figsize=(15, 8))

# Flatten the array for easier indexing
axs = axs.flatten()

# Generate data
x = np.linspace(0, 2*np.pi, 100)

# Define functions and titles
functions = [
    np.sin,
    np.cos,
    np.tan,
    lambda x: np.sinh(x/2),
    lambda x: np.cosh(x/2),
    lambda x: np.tanh(x/2)
]

titles = [
    'Sine: sin(x)',
    'Cosine: cos(x)',
    'Tangent: tan(x)',
    'Hyperbolic Sine: sinh(x/2)',
    'Hyperbolic Cosine: cosh(x/2)',
    'Hyperbolic Tangent: tanh(x/2)'
]

# Plot each function on its respective subplot
for i, (func, title) in enumerate(zip(functions, titles)):
    # Calculate y values safely (handle potential infinities for tan)
    if func is np.tan:
        y = np.ma.masked_where(np.abs(func(x)) > 10, func(x))
        axs[i].plot(x, y)
        axs[i].set_ylim(-5, 5)
    else:
        axs[i].plot(x, func(x))
    
    # Add title and grid
    axs[i].set_title(title)
    axs[i].set_xlabel('x')
    axs[i].set_ylabel('f(x)')
    axs[i].grid(True)

# Adjust layout
plt.tight_layout()
fig.suptitle('Trigonometric Functions', fontsize=16, y=1.02)

plt.show()
```

* **Text Annotations:** Adding explanatory text to highlight specific data points or regions

* **Styling Elements:** Customizing colors, fonts, line styles, and other visual properties
* **Style Sheets:** Applying pre-defined themes for consistent styling across visualizations

    | Component | Purpose | Example Method |
    |-----------|---------|----------------|
    | Annotations | Add text with optional arrows | `ax.annotate()` |
    | Text | Add text at specific coordinates | `ax.text()` |
    | Arrows | Point to specific features | `ax.arrow()` |
    | Style sheets | Apply predefined themes | `plt.style.use()` |
    | Color cycles | Control default colors | `plt.rcParams['axes.prop_cycle']` |

### 📘 Main Content

#### Adding Text Annotations

Text annotations are crucial for explaining specific features in your data:

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create the plot
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(x, y)

# Add basic text at a specific position (data coordinates)
ax.text(4, 0.5, 'Text at position (4, 0.5)', fontsize=12)

# Add an annotation with an arrow pointing to a specific point
ax.annotate('Local Maximum', 
            xy=(1.5, 1.0),        # Point to annotate (arrow tip)
            xytext=(3, 0.8),      # Text position
            arrowprops=dict(facecolor='black', shrink=0.05, width=1.5),
            fontsize=12)

# Add an annotation with a box around the text
ax.annotate('Local Minimum', 
            xy=(4.7, -1.0),       # Point to annotate
            xytext=(6, -0.5),     # Text position
            arrowprops=dict(facecolor='red', shrink=0.05),
            bbox=dict(boxstyle="round,pad=0.3", fc="yellow", ec="red", alpha=0.8),
            fontsize=12)

ax.set_title('Sine Wave with Annotations')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.grid(True)

plt.show()
```


Key annotation options include:
- `xy`: The point to annotate (where the arrow points)
- `xytext`: The position of the text
- `arrowprops`: Dictionary of arrow properties
- `bbox`: Dictionary of text box properties

#### Using Style Sheets

Matplotlib provides pre-defined style sheets that can transform the look of your visualizations:



```python
import matplotlib.pyplot as plt
import numpy as np

# Available styles
print(plt.style.available)

# Create a simple function to demonstrate styles
def plot_example(style_name):
    with plt.style.context(style_name):
        fig, ax = plt.subplots(figsize=(8, 6))
        
        # Generate sample data
        x = np.linspace(0, 10, 100)
        ax.plot(x, np.sin(x), label='sin(x)')
        ax.plot(x, np.cos(x), label='cos(x)')
        
        # Add elements for styling demonstration
        ax.set_title(f'Style: {style_name}')
        ax.set_xlabel('X Axis')
        ax.set_ylabel('Y Axis')
        ax.legend()
        ax.grid(True)
        
        plt.tight_layout()
        plt.show()

# Demonstrate a few popular styles
for style in ['default', 'seaborn-v0_8', 'ggplot', 'bmh', 'dark_background']:
    plot_example(style)
```

#### Customizing Individual Style Elements

```python
# Set style parameters directly
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman']
plt.rcParams['font.size'] = 12
plt.rcParams['lines.linewidth'] = 2
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.linestyle'] = '--'
plt.rcParams['grid.alpha'] = 0.7

# Create a plot with these custom settings
fig, ax = plt.subplots(figsize=(10, 6))
x = np.linspace(0, 10, 100)
ax.plot(x, np.sin(x))
ax.set_title('Plot with Custom Style Parameters')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')

plt.show()

# Reset to default style
plt.rcdefaults()
```

#### Creating Context-Based Styling

The `with` statement allows you to temporarily apply styles to specific plots:



```python
# Default style for the first plot
plt.figure(figsize=(8, 4))
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.title('Default Style')
plt.tight_layout()
plt.show()

# Temporarily use a different style
with plt.style.context('dark_background'):
    plt.figure(figsize=(8, 4))
    plt.plot(x, np.sin(x), color='yellow')
    plt.title('Dark Style (Temporary)')
    plt.tight_layout()
    plt.show()

# Back to default style automatically
plt.figure(figsize=(8, 4))
plt.plot(x, np.sin(x))
plt.title('Default Style Again')
plt.tight_layout()
plt.show()
```

## 📋 Chapter Summary

- **Figures and Axes** form the hierarchical structure of Matplotlib visualizations, where Figures are the overall container and Axes are the actual plotting areas. This structure provides precise control over every element of your plots.

- The **Object-Oriented Interface** offers greater flexibility and explicit control for complex visualizations compared to the state-based pyplot interface, making it the preferred approach for professional visualization work.

- **Multi-Panel Visualizations** allow for comparative analysis and comprehensive data presentation. Matplotlib provides various tools for creating grid layouts, including `subplots()`, `subplot2grid()`, and `GridSpec`.

- **Axes Sharing** enables consistent scales across multiple plots, making it easier to compare related data while reducing redundant labels and ticks.

- **Text Annotations** add context and explanation to your visualizations, highlighting important features and guiding viewers' attention to key insights in the data.

- **Custom Styling** transforms basic plots into professional visualizations through control of colors, fonts, line styles, and other visual elements. This enhances both the aesthetic appeal and the communicative power of your plots.

- **Style Sheets** provide predefined visual themes that can be applied to ensure consistent styling across visualizations, saving time and ensuring a professional look.

## 📚 Key Terms

- **Figure**: The top-level container that holds everything visible in a visualization
- **Axes**: The actual plotting area where data is displayed, with its own coordinate system
- **Subplot**: A single Axes within a grid layout of multiple plots
- **GridSpec**: A flexible class for creating complex grid layouts for subplots
- **Annotation**: A text element with an optional arrow that highlights a specific data point
- **Style Sheets**: Predefined collections of visual settings that can be applied to plots
- **rcParams**: Runtime configuration parameters that control Matplotlib's default behavior
- **tight_layout()**: A function that automatically adjusts subplot parameters for optimal spacing
