## **Chapter 4.1:** Pie Charts and Polar Plots

### 🔍 Visualizing Proportional and Circular Data

Pie charts and polar plots represent two specialized visualization types that are particularly useful for specific data scenarios. **Pie charts** excel at showing the composition of a whole and the relative sizes of its parts, while **polar plots** use a circular coordinate system that's ideal for directional or cyclical data. Both visualization types leverage the circular form to convey information in ways that traditional Cartesian plots cannot.

In this section, we'll explore how to create these circular visualizations effectively and when they're the most appropriate choice for your data.

### 🧩 Important Components

* **Pie Charts:**
  * **Wedges:** Individual slices representing categorical proportions
  * **Labels:** Text identifying each category
  * **Explode:** Values that "pull out" specific slices for emphasis
  * **Autopct:** Format string for displaying values on wedges

* **Polar Plots:**
  * **Polar Axes:** Circular plotting area with angular and radial dimensions
  * **Theta:** Angular coordinate (in radians or degrees)
  * **Radius:** Distance from the center point
  * **Projection:** Setting that transforms the coordinate system

| Component | Purpose | Example Parameter |
|-----------|---------|-------------------|
| Wedges | Slices of a pie chart | Determined by the data values |
| Labels | Identify pie chart categories | `labels=['A', 'B', 'C']` |
| Explode | Offset specific slices | `explode=(0.1, 0, 0)` |
| Autopct | Format percentages | `autopct='%1.1f%%'` |
| Polar projection | Create polar coordinate system | `projection='polar'` |

### 📘 Main Content

#### Creating Basic Pie Charts

Pie charts are ideal for showing the relative proportions of categorical data that sum up to a meaningful whole. They should be used when you have a limited number of categories (generally fewer than 7) to maintain readability.

Here's how to create a basic pie chart:



```python
import matplotlib.pyplot as plt
import numpy as np

# Data for the pie chart
labels = ['Python', 'Java', 'JavaScript', 'C/C++', 'C#', 'R', 'Others']
sizes = [28.53, 16.04, 8.72, 6.65, 6.63, 4.63, 28.8]

# Create the pie chart
plt.figure(figsize=(10, 6))
plt.pie(sizes, labels=labels)
plt.title('Programming Language Popularity')
plt.axis('equal')  # Equal aspect ratio ensures the pie is circular
plt.show()
```


This simple example creates a basic pie chart showing the relative popularity of programming languages. The `axis('equal')` line ensures the pie is drawn as a circle rather than an ellipse, which is important for accurate visual representation of proportions.

#### Customizing Pie Charts

To create more informative and visually appealing pie charts, Matplotlib offers numerous customization options:



```python
# Enhanced pie chart with customization
labels = ['Python', 'Java', 'JavaScript', 'C/C++', 'C#', 'R', 'Others']
sizes = [28.53, 16.04, 8.72, 6.65, 6.63, 4.63, 28.8]

# Custom colors and highlighting Python slice
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0', '#ffb3e6', '#d9d9d9']
explode = (0.1, 0, 0, 0, 0, 0, 0)  # Explode the Python slice

plt.figure(figsize=(10, 6))
plt.pie(sizes, 
        explode=explode, 
        labels=labels, 
        colors=colors,
        autopct='%1.1f%%',  # Add percentage labels
        shadow=True,        # Add shadow
        startangle=90,      # Rotate start angle
        wedgeprops={'edgecolor': 'black', 'linewidth': 0.5})  # Add edge to wedges

plt.title('Programming Language Popularity', fontsize=14)
plt.axis('equal')

# Add a legend
plt.legend(labels, loc='center left', bbox_to_anchor=(1, 0.5))

plt.tight_layout()
plt.show()
```


In this enhanced example, we've added several customizations:
- `explode` to offset the first slice for emphasis
- Custom `colors` for each slice
- `autopct` to display percentage values on each slice
- `shadow` for a 3D effect
- `startangle` to rotate the pie
- `wedgeprops` to add borders to the slices
- A legend positioned to the right of the pie

#### ⚠️ Common Mistakes with Pie Charts

* **Too many categories:** Pie charts become difficult to read with more than 7 categories
* **Similar sized slices:** When many slices are similar in size, comparisons become challenging
* **No ordering:** Not arranging slices in a meaningful order (e.g., from largest to smallest)
* **Poor color choices:** Using colors that don't provide enough contrast or are visually jarring
* **Missing the total:** Not providing context about what the whole 100% represents

#### Introduction to Polar Coordinates

Before diving into polar plots, it's important to understand the polar coordinate system. Unlike the Cartesian system (x, y), polar coordinates use:
- **Theta (θ):** The angle from the reference direction (typically measured counterclockwise from the positive x-axis)
- **Radius (r):** The distance from the origin

This system is particularly useful for data with cyclical or directional characteristics, such as:
- Wind direction and speed
- Seasonal patterns
- Directional distribution
- Periodic functions

#### Creating Polar Plots

To create a polar plot, you need to specify a projection when creating the axes:



```python
import matplotlib.pyplot as plt
import numpy as np

# Create a figure with polar projection
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw={'projection': 'polar'})

# Generate data
theta = np.linspace(0, 2*np.pi, 100)
radius = np.ones(100)  # Constant radius = circle

# Plot in polar coordinates
ax.plot(theta, radius)
ax.set_title('Basic Polar Plot: Circle', pad=20)

plt.show()
```


This basic example creates a simple circle in polar coordinates. The `projection='polar'` parameter is what transforms the axes to use the polar coordinate system.

#### Creating More Complex Polar Plots

Polar plots can be used to create a variety of interesting and informative visualizations:



```python
# Create a figure with 2x2 polar subplots
fig, axs = plt.subplots(2, 2, figsize=(12, 12), subplot_kw={'projection': 'polar'})
axs = axs.flatten()

# Example 1: Rose curve
theta1 = np.linspace(0, 2*np.pi, 500)
radius1 = 2 * np.cos(4 * theta1)  # Rose curve with 4 petals
axs[0].plot(theta1, radius1)
axs[0].set_title('Rose Curve: r = 2cos(4θ)', pad=20)
axs[0].set_rticks([0.5, 1, 1.5, 2])  # Set radius ticks

# Example 2: Spiral
theta2 = np.linspace(0, 6*np.pi, 500)
radius2 = np.linspace(0, 2, 500)  # Increasing radius
axs[1].plot(theta2, radius2)
axs[1].set_title('Spiral: r = θ/3π', pad=20)

# Example 3: Polar scatter plot (e.g., wind direction and speed)
np.random.seed(42)
theta3 = np.random.uniform(0, 2*np.pi, 100)  # Random angles
radius3 = np.random.uniform(0, 2, 100)       # Random distances
axs[2].scatter(theta3, radius3, c=radius3, cmap='viridis', alpha=0.7)
axs[2].set_title('Polar Scatter Plot', pad=20)

# Example 4: Directional histogram (e.g., wind rose)
theta4 = np.random.normal(np.pi/4, np.pi/6, 1000)  # Normally distributed angles
bins = np.linspace(0, 2*np.pi, 16)
axs[3].hist(theta4, bins=bins, color='skyblue', alpha=0.7)
axs[3].set_title('Directional Histogram', pad=20)

plt.tight_layout()
plt.show()
```


This more complex example demonstrates several types of polar plots:
1. A rose curve showing a mathematical pattern
2. A spiral showing increasing radius with angle
3. A scatter plot that could represent directional data like wind direction and speed
4. A directional histogram showing the distribution of angles

#### Applications of Polar Plots

Polar plots are particularly useful in various fields:

1. **Meteorology**: Wind rose diagrams showing wind direction and speed distribution
2. **Engineering**: Radiation patterns of antennas or speakers
3. **Statistics**: Directional or cyclical data analysis
4. **Mathematics**: Visualization of polar functions like cardioids or rose curves
5. **Biology**: Animal migration patterns or circadian rhythms
6. **Astronomy**: Star positions or planetary orbits



```python
# Example: Wind rose plot
np.random.seed(42)
# Generate random wind directions (0 to 360 degrees) and speeds
directions = np.random.uniform(0, 360, 500)
speeds = np.random.weibull(1.5, 500) * 5  # Weibull distribution common for wind speeds

# Convert to radians for plotting
theta = np.radians(directions)

# Create the figure
fig, ax = plt.subplots(figsize=(10, 10), subplot_kw={'projection': 'polar'})

# Create a scatter plot colored by speed
scatter = ax.scatter(theta, speeds, c=speeds, cmap='viridis', alpha=0.7)

# Customize the plot
ax.set_theta_zero_location('N')  # Set 0 degrees to North
ax.set_theta_direction(-1)       # Clockwise
ax.set_rlabel_position(0)        # Move radial labels
ax.set_title('Wind Rose Diagram: Direction and Speed', pad=20)

# Add a colorbar for speed reference
cbar = plt.colorbar(scatter, ax=ax, pad=0.1)
cbar.set_label('Wind Speed (m/s)')

# Add grid lines
ax.grid(True)

plt.show()
```


This example creates a wind rose diagram, a common application of polar plots in meteorology. The plot shows both wind direction (angle) and speed (radius), with color also encoding speed for additional visual information. Direction is set to follow meteorological conventions (0° at North, clockwise rotation).

## **Chapter 4.2:** Heatmaps and Contour Plots

### 🔍 Visualizing Data Intensity and Level Curves

Heatmaps and contour plots are powerful visualization techniques for representing **continuous data across two dimensions**. They excel at revealing patterns, trends, and relationships that might be difficult to discern in other plot types. Both use color to convey information, but in different ways: heatmaps show intensity directly through color mapping, while contour plots use lines or filled regions to connect points of equal value.

These visualizations are particularly valuable for scientific, engineering, and data science applications where understanding the distribution and variation of values across a 2D space is essential.

### 🧩 Important Components

* **Heatmaps:**
  * **2D Data Matrix:** The values to be visualized as colors
  * **Colormap:** Maps data values to colors
  * **Color Scale:** Indicates how values correspond to colors
  * **Annotations:** Optional text labels for individual cells

* **Contour Plots:**
  * **X-Y Grid:** The 2D spatial coordinates
  * **Z Values:** The height or intensity at each (x, y) point
  * **Level Curves:** Lines connecting points of equal z-value
  * **Level Fills:** Colored regions between contour lines

| Component | Purpose | Example Parameter |
|-----------|---------|-------------------|
| Colormap | Define the color scale | `cmap='viridis'` |
| Normalization | Control value-to-color mapping | `norm=matplotlib.colors.LogNorm()` |
| Colorbar | Legend showing color scale | `plt.colorbar()` |
| Contour levels | Control number/values of contours | `levels=10` or `levels=[0, 5, 10, 15]` |
| Contour labels | Add value labels to contour lines | `plt.clabel()` |

### 📘 Main Content

#### Creating Basic Heatmaps

Heatmaps in Matplotlib are typically created using the `imshow()` function, which displays data as an image where color represents value:



```python
import matplotlib.pyplot as plt
import numpy as np

# Generate some data for the heatmap
data = np.random.rand(10, 12)  # 10x12 matrix of random values

# Create a basic heatmap
plt.figure(figsize=(10, 6))
heatmap = plt.imshow(data, cmap='viridis')

# Add a color bar
plt.colorbar(heatmap, label='Value')

# Add title and labels
plt.title('Basic Heatmap', fontsize=14)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')

plt.tight_layout()
plt.show()
```


This example creates a basic heatmap from random data. The `imshow()` function displays the data as a color-mapped image, and `colorbar()` adds a legend that shows how colors correspond to values.

#### Customizing Heatmaps

To create more informative and visually effective heatmaps, you can customize various aspects:



```python
import matplotlib.pyplot as plt
import numpy as np

# Generate structured data for a more meaningful heatmap
x = np.linspace(-3, 3, 20)
y = np.linspace(-3, 3, 20)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)  # Create a 2D wave pattern

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 8))

# Create an enhanced heatmap
heatmap = ax.imshow(Z, 
                   cmap='coolwarm',       # Use a diverging colormap
                   interpolation='bicubic',  # Smooth the colors
                   aspect='auto',         # Adjust aspect ratio to fill the axes
                   origin='lower',        # Place origin at bottom left
                   extent=[-3, 3, -3, 3]) # Set axis limits to match our data

# Add contour lines over the heatmap for enhanced perception
contour = ax.contour(X, Y, Z, colors='black', linewidths=0.5, alpha=0.7)
ax.clabel(contour, inline=True, fontsize=8)  # Add value labels to contours

# Add a colorbar
cbar = fig.colorbar(heatmap, ax=ax)
cbar.set_label('Z Value', rotation=270, labelpad=20)

# Add title and axis labels
ax.set_title('Customized Heatmap with Contours', fontsize=14)
ax.set_xlabel('X Axis', fontsize=12)
ax.set_ylabel('Y Axis', fontsize=12)

# Add text annotations for specific features
ax.text(0, 0, 'Center Point', color='white', 
        ha='center', va='center', fontweight='bold')

plt.tight_layout()
plt.show()
```


This enhanced example demonstrates several customization options:
- Using a structured data pattern instead of random values
- Applying a diverging colormap (`coolwarm`) that is appropriate for data with positive and negative values
- Setting the `origin` and `extent` to control the coordinate mapping
- Adding contour lines on top of the heatmap for enhanced perception
- Adding text annotations to highlight specific points
- Customizing the colorbar label

#### ⚠️ Common Mistakes with Heatmaps

* **Poor colormap selection:** Using rainbow colormaps like `jet` that can distort perception
* **Missing color bar:** Not including a legend to interpret the colors
* **Inappropriate aspect ratio:** Distorting the visualization by not maintaining the correct proportions
* **Unclear axis labeling:** Not properly indicating what the dimensions represent
* **Improper normalization:** Not adjusting the color scaling for skewed distributions

#### Creating Correlation Matrices as Heatmaps

One common application of heatmaps is visualizing correlation matrices:



```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns  # For better heatmap visualization

# Generate some correlated data
np.random.seed(42)
n = 1000
x = np.random.normal(0, 1, n)
y = x * 0.8 + np.random.normal(0, 0.5, n)
z = x * 0.4 - y * 0.3 + np.random.normal(0, 0.8, n)
w = -y * 0.5 + np.random.normal(0, 0.6, n)
v = x * 0.1 + y * 0.2 + z * 0.1 + np.random.normal(0, 0.7, n)

# Create a DataFrame
df = pd.DataFrame({'A': x, 'B': y, 'C': z, 'D': w, 'E': v})

# Calculate the correlation matrix
corr_matrix = df.corr()

# Create a heatmap of the correlation matrix
plt.figure(figsize=(10, 8))
heatmap = plt.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(heatmap, label='Correlation Coefficient')

# Add labels and ticks
plt.title('Correlation Matrix Heatmap', fontsize=14)
plt.xticks(np.arange(len(corr_matrix.columns)), corr_matrix.columns)
plt.yticks(np.arange(len(corr_matrix.index)), corr_matrix.index)

# Add text annotations
for i in range(len(corr_matrix.columns)):
    for j in range(len(corr_matrix.index)):
        plt.text(j, i, f'{corr_matrix.iloc[i, j]:.2f}',
                ha='center', va='center', 
                color='white' if abs(corr_matrix.iloc[i, j]) > 0.5 else 'black')

plt.tight_layout()
plt.show()

# Alternative using Seaborn (optional but often preferred for correlation matrices)
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, 
            annot_kws={"size": 10}, square=True)
plt.title('Correlation Matrix Heatmap (Seaborn)', fontsize=14)
plt.tight_layout()
plt.show()
```


This example demonstrates two approaches to visualizing a correlation matrix:
1. Using pure Matplotlib's `imshow()` with custom text annotations
2. Using Seaborn's `heatmap()`, which is specifically designed for this purpose and includes built-in annotation options

The correlation coefficient values range from -1 to 1, making a diverging colormap like 'coolwarm' appropriate, as it clearly shows positive correlations (warm colors) and negative correlations (cool colors).

#### Basic Contour Plots

Contour plots display lines connecting points of equal value, similar to topographic maps. They're created with the `contour()` function:



```python
import matplotlib.pyplot as plt
import numpy as np

# Generate data for the contour plot
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

# Create a basic contour plot
plt.figure(figsize=(10, 8))
contour = plt.contour(X, Y, Z, levels=10, cmap='viridis')

# Add contour labels
plt.clabel(contour, inline=True, fontsize=8)

# Add a colorbar
plt.colorbar(contour, label='Z Value')

# Add title and labels
plt.title('Basic Contour Plot', fontsize=14)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.grid(linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

#### Filled Contour Plots

For better visualization of regions between contour lines, filled contour plots color the areas between contours:



```python
import matplotlib.pyplot as plt
import numpy as np

# Generate data for the filled contour plot
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

# Create a figure with two subplots side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# Contour plot on the left
CS1 = ax1.contour(X, Y, Z, levels=10, colors='black', linewidths=0.5)
ax1.clabel(CS1, inline=True, fontsize=8)
ax1.set_title('Contour Lines', fontsize=14)
ax1.set_xlabel('X Axis')
ax1.set_ylabel('Y Axis')
ax1.grid(linestyle='--', alpha=0.7)

# Filled contour plot on the right
CS2 = ax2.contourf(X, Y, Z, levels=10, cmap='viridis', alpha=0.8)
# Optionally add contour lines on top for clarity
CS2_lines = ax2.contour(X, Y, Z, levels=10, colors='black', linewidths=0.5, alpha=0.7)
ax2.clabel(CS2_lines, inline=True, fontsize=8)
ax2.set_title('Filled Contour Plot', fontsize=14)
ax2.set_xlabel('X Axis')
ax2.set_ylabel('Y Axis')
ax2.grid(linestyle='--', alpha=0.7)

# Add a shared colorbar
cbar = fig.colorbar(CS2, ax=[ax1, ax2], label='Z Value')

plt.tight_layout()
plt.show()
```

#### Applications of Contour Plots

Contour plots are valuable in many fields:

1. **Meteorology**: Weather maps showing pressure, temperature, or precipitation
2. **Geography**: Topographic maps showing elevation
3. **Engineering**: Stress or heat distribution analysis
4. **Physics**: Electric or magnetic field visualizations
5. **Statistics**: Density estimation and likelihood functions



```python
# Example: Creating a topographic-like map with contours
import matplotlib.pyplot as plt
import numpy as np

# Generate terrain-like data
def terrain_function(x, y):
    """Generate artificial terrain data"""
    return (3*(1-x)**2 * np.exp(-x**2 - (y+1)**2) 
            - 10*(x/5 - x**3 - y**5) * np.exp(-x**2 - y**2)
            - 1/3 * np.exp(-(x+1)**2 - y**2))

# Create coordinate grid
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = terrain_function(X, Y)

# Create figure
fig, ax = plt.subplots(figsize=(12, 10))

# Create a filled contour plot with many levels for smooth appearance
contourf = ax.contourf(X, Y, Z, levels=50, cmap='terrain')

# Add contour lines for emphasis
contour = ax.contour(X, Y, Z, levels=15, colors='black', linewidths=0.5, alpha=0.7)

# Add labels to some contours
ax.clabel(contour, inline=True, fontsize=8, fmt='%.1f', levels=contour.levels[::3])

# Add colorbar
cbar = fig.colorbar(contourf, ax=ax, label='Elevation')

# Customize plot appearance
ax.set_title('Topographic Map (Simulated Terrain)', fontsize=16)
ax.set_xlabel('X Distance (km)', fontsize=12)
ax.set_ylabel('Y Distance (km)', fontsize=12)
ax.set_aspect('equal')  # Equal scale for x and y

# Add annotations for notable features
max_point = np.unravel_index(np.argmax(Z), Z.shape)
max_x, max_y = X[max_point], Y[max_point]
ax.plot(max_x, max_y, 'r^', markersize=10)  # Red triangle for peak
ax.annotate('Summit', xy=(max_x, max_y), xytext=(max_x+0.5, max_y+0.3),
            arrowprops=dict(facecolor='black', shrink=0.05, width=1.5))

# Add a "valley" annotation
valley_x, valley_y = -0.5, -1.5
ax.plot(valley_x, valley_y, 'bs', markersize=8)  # Blue square for valley
ax.annotate('Valley', xy=(valley_x, valley_y), xytext=(valley_x-1, valley_y-0.5),
            arrowprops=dict(facecolor='black', shrink=0.05, width=1.5))

plt.tight_layout()
plt.show()
```


This example creates a contour plot that resembles a topographic map, with colors representing elevation and contour lines marking specific heights. Annotations highlight notable features like the highest point and a valley.

## **Chapter 4.3:** 3D Visualization

### 🔍 Exploring Data in Three Dimensions

While 2D visualizations are powerful, some datasets and relationships are best understood in their full three-dimensional form. Matplotlib's `mplot3d` toolkit extends the library's capabilities to create **true 3D visualizations** that can be rotated, viewed from different angles, and styled for optimal understanding of complex spatial relationships.

3D visualization is particularly valuable for scientific and engineering data where the relationship between three variables needs to be understood simultaneously. In this section, we'll explore how to create, customize, and interpret various types of 3D plots.

### 🧩 Important Components

* **3D Axes:** Plotting area with x, y, and z dimensions
* **Projections:** Methods for representing 3D space on a 2D surface
* **Surface Plots:** Visualization of a function z = f(x, y)
* **3D Scatter Plots:** Points positioned in 3D space
* **Wireframes:** Transparent 3D surface representations

| Component | Purpose | Example Usage |
|-----------|---------|---------------|
| Axes3D | Creates 3D axes object | `ax = fig.add_subplot(111, projection='3d')` |
| plot_surface | Creates a 3D surface | `ax.plot_surface(X, Y, Z)` |
| scatter3D | Creates 3D scatter plot | `ax.scatter3D(x, y, z)` |
| plot_wireframe | Creates wireframe plot | `ax.plot_wireframe(X, Y, Z)` |
| view_init | Sets viewing angle | `ax.view_init(elev=30, azim=45)` |

#### Creating 3D Surface Plots

One of the most common 3D visualizations is the surface plot, which represents a function z = f(x, y) as a continuous surface:



```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Create figure and 3D axis
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Generate data for a 3D surface
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create the surface plot
surface = ax.plot_surface(X, Y, Z, cmap='viridis', 
                          edgecolor='none', alpha=0.8)

# Add a color bar
fig.colorbar(surface, ax=ax, shrink=0.5, aspect=5, label='Z Value')

# Set labels
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
ax.set_title('3D Surface Plot: z = sin(√(x² + y²))', fontsize=14)

plt.tight_layout()
plt.show()
```


This example creates a surface plot of a "ripple" function. The `plot_surface()` method requires 2D arrays of X, Y, and Z values, which we create using `np.meshgrid()`. The `cmap` parameter specifies the colormap used to color the surface based on height (Z value).

#### Customizing Surface Plots

You can enhance surface plots with various customization options:



```python
# Create an enhanced 3D surface plot
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Generate data
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create the surface with additional customization
surface = ax.plot_surface(X, Y, Z, 
                         cmap='plasma',          # Different colormap
                         rstride=1,              # Row stride for wireframe
                         cstride=1,              # Column stride for wireframe
                         linewidth=0,            # No lines between faces
                         antialiased=True,       # Smooth appearance
                         alpha=0.8)              # Slight transparency

# Add a color bar with custom formatting
cbar = fig.colorbar(surface, ax=ax, shrink=0.5, aspect=10)
cbar.set_label('Height (z)', rotation=270, labelpad=20, fontsize=12)

# Customize labels
ax.set_xlabel('X Axis', fontsize=12, labelpad=10)
ax.set_ylabel('Y Axis', fontsize=12, labelpad=10)
ax.set_zlabel('Z Axis', fontsize=12, labelpad=10)
ax.set_title('Enhanced 3D Surface Plot', fontsize=16, pad=20)

# Customize view angle
ax.view_init(elev=30, azim=45)  # Elevation and azimuth angles

# Customize the background
ax.xaxis.pane.fill = False      # Transparent background
ax.yaxis.pane.fill = False
ax.zaxis.pane.fill = False
ax.xaxis.pane.set_edgecolor('lightgray')
ax.yaxis.pane.set_edgecolor('lightgray')
ax.zaxis.pane.set_edgecolor('lightgray')
ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```


This enhanced example demonstrates several customization options:
- `rstride` and `cstride` control the density of the wireframe grid
- `view_init()` sets the camera angle
- Customizing the background panes and grid
- Formatting the colorbar and labels

#### ⚠️ Common Mistakes with 3D Plots

* **Overplotting:** Including too many data points or surfaces, making the visualization cluttered
* **Poor viewing angle:** Not adjusting the view to highlight important features of the data
* **Missing context:** Forgetting to add axis labels, title, or colorbar
* **Deceptive perspective:** Creating misleading impressions due to foreshortening or occlusion
* **Inappropriate use:** Using 3D when a 2D visualization would be clearer

#### 3D Scatter Plots

3D scatter plots display individual points in three-dimensional space, making them ideal for visualizing discrete data with three variables:



```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Create figure and 3D axis
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Generate random 3D data
n = 100
x = np.random.normal(0, 1, n)
y = np.random.normal(0, 1, n)
z = np.random.normal(0, 1, n)

# Color points based on their distance from origin
colors = np.sqrt(x**2 + y**2 + z**2)

# Create the scatter plot
scatter = ax.scatter(x, y, z, 
                    c=colors,           # Color by distance
                    cmap='viridis',     # Colormap
                    marker='o',         # Marker style
                    s=30,               # Marker size
                    alpha=0.8)          # Transparency

# Add a color bar
fig.colorbar(scatter, ax=ax, label='Distance from Origin')

# Set labels
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
ax.set_title('3D Scatter Plot', fontsize=14)

plt.tight_layout()
plt.show()
```


This example creates a 3D scatter plot of normally distributed random points, with colors representing each point's distance from the origin. The `scatter()` method is similar to its 2D counterpart but places points in 3D space.

#### 3D Line Plots

3D line plots connect points in 3D space, useful for visualizing paths, trajectories, or parametric curves:



```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Create figure and 3D axis
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Generate data for a 3D spiral
t = np.linspace(0, 10*np.pi, 500)
x = np.cos(t)
y = np.sin(t)
z = t/5

# Create the 3D line plot
ax.plot(x, y, z, 
        label='Spiral Curve',
        color='blue',
        linewidth=2)

# Add a few scatter points along the curve for emphasis
points_t = np.linspace(0, 10*np.pi, 10)
points_x = np.cos(points_t)
points_y = np.sin(points_t)
points_z = points_t/5
ax.scatter(points_x, points_y, points_z, color='red', s=50)

# Set labels and title
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
ax.set_title('3D Line Plot: Helix', fontsize=14)

# Add a legend
ax.legend()

plt.tight_layout()
plt.show()
```


This example creates a 3D helix curve using the `plot()` method, with additional scatter points highlighting specific positions along the curve. 3D line plots are valuable for visualizing data with an inherent order or progression.

#### 3D Wireframe Plots

Wireframe plots show a 3D surface as a mesh of lines, which can help visualize the structure while maintaining transparency:



```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Create figure and 3D axis
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Generate data for a 3D surface
x = np.linspace(-5, 5, 30)
y = np.linspace(-5, 5, 30)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create a wireframe plot
wireframe = ax.plot_wireframe(X, Y, Z, 
                            color='black',     # Line color
                            linewidth=0.5,     # Line width
                            rstride=1,         # Row stride
                            cstride=1)         # Column stride

# Set labels and title
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
ax.set_title('3D Wireframe Plot', fontsize=14)

# Customize the view angle
ax.view_init(elev=20, azim=30)

plt.tight_layout()
plt.show()
```


Wireframe plots are useful when you want to see through the surface to understand its structure, or when you want to overlay multiple surfaces.

#### Combining Multiple 3D Visualization Types

You can combine different 3D plot types to create more informative visualizations:



```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Create figure and 3D axis
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Generate data for a 3D surface
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create a surface plot with transparency
surface = ax.plot_surface(X, Y, Z, 
                        cmap='viridis', 
                        alpha=0.7,
                        linewidth=0,
                        antialiased=True)

# Add a wireframe with larger stride to highlight structure
wireframe = ax.plot_wireframe(X, Y, Z,
                            color='black',
                            linewidth=0.5,
                            rstride=5,
                            cstride=5)

# Add scatter points at local maxima and minima
# Find a few interesting points manually for illustration
max_points = [(0, 0, 1), (4, 0, -0.2), (-3, 2, 0.3)]
min_points = [(2.5, 2.5, -0.6), (-3, -3, 0.4)]

max_x, max_y, max_z = zip(*max_points)
min_x, min_y, min_z = zip(*min_points)

ax.scatter(max_x, max_y, max_z, color='red', s=50, label='Maxima')
ax.scatter(min_x, min_y, min_z, color='blue', s=50, label='Minima')

# Set labels and title
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
ax.set_title('Combined 3D Visualization', fontsize=14)

# Add a legend and colorbar
ax.legend()
fig.colorbar(surface, ax=ax, shrink=0.5, aspect=5, label='Z Value')

# Customize view
ax.view_init(elev=30, azim=45)

plt.tight_layout()
plt.show()
```


This combined approach provides multiple perspectives on the same data: the surface shows the overall shape and height using color, the wireframe highlights the structural grid, and the scatter points mark specific features of interest.

```python
# Example: Creating a 3D terrain visualization
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Create figure and 3D axis
fig = plt.figure(figsize=(14, 10))
ax = fig.add_subplot(111, projection='3d')

# Generate terrain-like data
def terrain_function(x, y):
    """Generate artificial terrain data with mountains and valleys"""
    scale = 0.5
    return (2 * np.sin(scale*x) * np.cos(scale*y) +
            np.sin(2*scale*x) * np.cos(2*scale*y)/3 +
            ((x**2 + y**2)/100))

# Create coordinate grid
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(x, y)
Z = terrain_function(X, Y)

# Create a surface plot for the terrain
terrain = ax.plot_surface(X, Y, Z, 
                        cmap='terrain',     # Terrain-specific colormap
                        linewidth=0,
                        antialiased=True,
                        alpha=0.8)

# Add a contour plot beneath for topographic effect
contour = ax.contourf(X, Y, Z, 
                     zdir='z',             # Project onto z-plane
                     offset=Z.min() - 0.5,  # Position just below the surface
                     cmap='terrain',
                     alpha=0.5)

# Highlight the "mountain peaks" with scatter points
# Find the 5 highest points
peak_indices = np.argpartition(Z.flatten(), -5)[-5:]
peak_rows, peak_cols = np.unravel_index(peak_indices, Z.shape)
peak_x = X[peak_rows, peak_cols]
peak_y = Y[peak_rows, peak_cols]
peak_z = Z[peak_rows, peak_cols]

# Plot the peaks
ax.scatter(peak_x, peak_y, peak_z + 0.2, color='red', s=50, marker='^')

# Add annotations for a few peaks
for i in range(min(3, len(peak_x))):
    ax.text(peak_x[i], peak_y[i], peak_z[i] + 0.4, f'Peak {i+1}', 
            color='black', ha='center', fontweight='bold')

# Customize the plot
ax.set_xlabel('X Distance (km)')
ax.set_ylabel('Y Distance (km)')
ax.set_zlabel('Elevation (km)')
ax.set_title('3D Terrain Visualization', fontsize=16)

# Adjust the view for better perspective
ax.view_init(elev=40, azim=30)

# Add a colorbar
fig.colorbar(terrain, ax=ax, shrink=0.5, aspect=10, label='Elevation (km)')

plt.tight_layout()
plt.show()
```


This terrain visualization example demonstrates how 3D plots can be used to create intuitive representations of geographic data, highlighting features like mountain peaks and using appropriate colormaps to enhance understanding.

#### Statistical Plots: Box Plots, Violin Plots, and Error Bars

Statistical visualizations help communicate data distributions, variance, and uncertainty. They're essential for data analysis and scientific reporting.

##### Box Plots (Box and Whisker Plots)

Box plots provide a standardized way of displaying data distribution based on a five-number summary: minimum, first quartile (Q1), median, third quartile (Q3), and maximum.



```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
np.random.seed(42)
data1 = np.random.normal(100, 15, 200)  # Normal distribution, mean=100, std=15
data2 = np.random.normal(90, 20, 200)   # Normal distribution, mean=90, std=20
data3 = np.random.exponential(20, 200) + 70  # Exponential distribution, shifted

# Combine data into a list
data = [data1, data2, data3]

# Create a figure
fig, ax = plt.subplots(figsize=(10, 6))

# Create box plot
box_plot = ax.boxplot(data, 
                     patch_artist=True,  # Fill the boxes with color
                     notch=True,         # Add notches for CI of median
                     vert=True,          # Vertical orientation
                     labels=['Group 1', 'Group 2', 'Group 3'])

# Customize colors
colors = ['lightblue', 'lightgreen', 'lightpink']
for patch, color in zip(box_plot['boxes'], colors):
    patch.set_facecolor(color)

# Add grid lines
ax.yaxis.grid(True, linestyle='--', alpha=0.7)

# Add title and labels
ax.set_title('Box Plot Comparison of Three Distributions', fontsize=14)
ax.set_xlabel('Data Group')
ax.set_ylabel('Value')

plt.tight_layout()
plt.show()
```


This example creates box plots for three different distributions. Box plots show:
- The median (line inside the box)
- The interquartile range or IQR (box)
- The whiskers (lines extending from the box, typically to min/max or 1.5*IQR)
- Outliers (individual points beyond the whiskers)

They're excellent for comparing distributions across groups, especially when you want to highlight central tendency, spread, and outliers.

##### Violin Plots

Violin plots combine box plots with kernel density estimates, showing the full distribution shape alongside summary statistics:



```python
import matplotlib.pyplot as plt
import numpy as np

# Use the same data as for the box plot
# Create a figure
fig, ax = plt.subplots(figsize=(10, 6))

# Create violin plot
violin_parts = ax.violinplot(data, 
                         showmeans=False, 
                         showmedians=True)

# Customize colors
for i, pc in enumerate(violin_parts['bodies']):
    pc.set_facecolor(colors[i])
    pc.set_edgecolor('black')
    pc.set_alpha(0.7)

# Add individual points for better visualization of data distribution
for i, d in enumerate(data):
    # Add jittered points
    x = np.random.normal(i+1, 0.04, size=len(d))
    ax.scatter(x, d, alpha=0.2, s=5, color='black')

# Add grid lines
ax.yaxis.grid(True, linestyle='--', alpha=0.7)

# Set axis labels
ax.set_title('Violin Plot Comparison of Three Distributions', fontsize=14)
ax.set_xlabel('Data Group')
ax.set_ylabel('Value')
ax.set_xticks([1, 2, 3])
ax.set_xticklabels(['Group 1', 'Group 2', 'Group 3'])

plt.tight_layout()
plt.show()
```


Violin plots provide more detailed distribution information than box plots, making them particularly useful when:
- The shape of the distribution matters (e.g., bimodal distributions)
- You need to see the density of values at different levels
- You want to compare complex distributions across groups

##### Error Bars

Error bars visualize the uncertainty or variability in measurements, which is crucial for scientific and statistical reporting:



```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
np.random.seed(42)
x = np.arange(1, 6)
y = np.array([2, 4, 6, 8, 10])
y_error = np.random.uniform(0.5, 1.5, size=len(y))  # Random error values

# Create a figure
fig, ax = plt.subplots(figsize=(10, 6))

# Create error bar plot
ax.errorbar(x, y, yerr=y_error,
          fmt='o',           # Marker style
          color='blue',      # Marker color
          ecolor='red',      # Error bar color
          elinewidth=2,      # Error bar line width
          capsize=6)         # Width of cap on error bars

# Add grid lines
ax.grid(True, linestyle='--', alpha=0.7)

# Add title and labels
ax.set_title('Error Bar Plot Example', fontsize=14)
ax.set_xlabel('X Value')
ax.set_ylabel('Y Value')
ax.set_xticks(x)

plt.tight_layout()
plt.show()
```


Error bars are essential for:
- Scientific data with measurement uncertainty
- Statistical results with confidence intervals
- Any visualization where conveying the precision of values is important

#### Geographic and Spatial Plots

Several Matplotlib plot types are designed specifically for spatial data and vector fields.

##### Hexbin Plots

Hexbin plots create a 2D histogram using hexagonal bins, which is excellent for visualizing the density of points in a scatter plot:



```python
import matplotlib.pyplot as plt
import numpy as np

# Generate some sample data (correlated with some noise)
np.random.seed(42)
n = 10000
x = np.random.normal(0, 1, n)
y = x + np.random.normal(0, 1, n)

# Create a figure
fig, ax = plt.subplots(figsize=(10, 8))

# Create hexbin plot
hb = ax.hexbin(x, y, 
              gridsize=30,           # Number of bins
              cmap='viridis',        # Colormap
              mincnt=1,              # Minimum count to color a bin
              bins='log',            # Use logarithmic scale for counts
              alpha=0.8)             # Transparency

# Add a colorbar
cb = fig.colorbar(hb, ax=ax)
cb.set_label('log10(N)')

# Add title and labels
ax.set_title('Hexbin Plot: Point Density Visualization', fontsize=14)
ax.set_xlabel('X Value')
ax.set_ylabel('Y Value')

# Add grid
ax.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```


Hexbin plots are superior to standard scatter plots when:
- You have very large datasets with potential overplotting
- The density or concentration of points is more important than individual points
- You want to visualize the relationship between two variables with emphasis on frequency

##### Stream Plots

Stream plots visualize vector fields by drawing streamlines that follow the vector direction:



```python
import matplotlib.pyplot as plt
import numpy as np

# Create a grid of points
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)

# Create vector field components (example: rotating vector field)
U = -Y  # x-component of the vector field
V = X   # y-component of the vector field

# Create a figure
fig, ax = plt.subplots(figsize=(10, 8))

# Create stream plot
strm = ax.streamplot(X, Y, U, V,
                    color=np.sqrt(U**2 + V**2),  # Color by vector magnitude
                    linewidth=1.5,               # Line width
                    cmap='autumn',               # Colormap
                    density=1.5,                 # Streamline density
                    arrowsize=1.5)               # Arrow size

# Add a colorbar
cbar = fig.colorbar(strm.lines)
cbar.set_label('Vector Magnitude')

# Add title and labels
ax.set_title('Stream Plot: Vector Field Visualization', fontsize=14)
ax.set_xlabel('X')
ax.set_ylabel('Y')

# Set equal aspect ratio
ax.set_aspect('equal')

plt.tight_layout()
plt.show()
```


Stream plots are valuable for:
- Visualizing fluid flow
- Electromagnetic field lines
- Any 2D vector field where direction and magnitude are important
- Showing paths that follow the vector field

##### Tri-Surface and Triangulation Plots

For irregular or unstructured grids, Matplotlib provides triangulation-based plots:



```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.tri import Triangulation

# Generate some scattered points
np.random.seed(42)
n = 100
x = np.random.uniform(-1, 1, n)
y = np.random.uniform(-1, 1, n)
z = np.sin(4*x) * np.cos(4*y)

# Create a triangulation
triang = Triangulation(x, y)

# Create a figure
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6), subplot_kw={'projection': '3d'})

# Create a 3D tri-surface plot
surf = ax1.plot_trisurf(triang, z,
                       cmap='viridis',
                       linewidth=0.2,
                       antialiased=True,
                       edgecolor='gray')
ax1.set_title('Triangulated Surface Plot', fontsize=14)
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Z')

# Add a wireframe view in the second subplot
wire = ax2.plot_trisurf(triang, z,
                       cmap='viridis',
                       linewidth=0.5,
                       antialiased=True,
                       alpha=0.8)
ax2.set_title('Triangulated Wireframe View', fontsize=14)
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_zlabel('Z')
ax2.view_init(elev=20, azim=30)  # Change the view angle

# Add a shared colorbar
fig.colorbar(surf, ax=[ax1, ax2], shrink=0.5, aspect=10, label='Z Value')

plt.tight_layout()
plt.show()
```


Triangulation plots are useful for:
- Irregular or scattered data points
- Terrain modeling from scattered elevation data
- Finite element analysis results
- Any surface representation where data points aren't on a regular grid

#### Specialized Statistical and Categorical Plots

Matplotlib includes several plot types designed for specific statistical or categorical data visualization needs.

##### Event Plots

Event plots show the occurrence of events over time, similar to a spike train in neural recording:



```python
import matplotlib.pyplot as plt
import numpy as np

# Generate some random event data
np.random.seed(42)
num_points = 100

# Generate 5 different "event trains"
events = [np.sort(np.random.uniform(0, 10, num_points)) for i in range(5)]

# Thin out some events randomly
for i in range(5):
    events[i] = events[i][np.random.uniform(0, 1, num_points) > 0.5]

# Create a figure
fig, ax = plt.subplots(figsize=(12, 6))

# Create an event plot
ax.eventplot(events,
            colors=['blue', 'red', 'green', 'orange', 'purple'],
            lineoffsets=1.0,   # Height of each event line
            linelengths=0.8,   # Length of each event marker
            linewidths=1.0)    # Width of each event line

# Add title and labels
ax.set_title('Event Plot: Temporal Events Visualization', fontsize=14)
ax.set_xlabel('Time (s)')
ax.set_ylabel('Event Group')
ax.set_yticks(range(5))
ax.set_yticklabels(['Group A', 'Group B', 'Group C', 'Group D', 'Group E'])

# Add grid
ax.grid(True, axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```


Event plots are useful for:
- Neuronal spike train data
- Transaction or activity logs
- Any time-series data where the occurrence of discrete events is important

##### Multi-panel Plots

While not a specific plot type, creating multi-panel plots (also called facets or small multiples) is a powerful technique for comparing several related visualizations:



```python
import matplotlib.pyplot as plt
import numpy as np

# Create a figure with a grid of subplots
fig, axs = plt.subplots(2, 3, figsize=(15, 8))
axs = axs.flatten()  # Flatten the 2D array of axes for easier indexing

plot_types = ['Line Plot', 'Scatter Plot', 'Bar Chart', 
              'Step Plot', 'Fill Between', 'Stem Plot']

# Generate some sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)
categories = ['A', 'B', 'C', 'D', 'E']
values = [3, 7, 2, 5, 8]

# Create different plot types in each subplot
# Line plot
axs[0].plot(x, y, 'b-')
axs[0].set_title(plot_types[0])
axs[0].grid(True, linestyle='--', alpha=0.7)

# Scatter plot
axs[1].scatter(x[::5], y[::5], c=y[::5], cmap='viridis', s=50)
axs[1].set_title(plot_types[1])
axs[1].grid(True, linestyle='--', alpha=0.7)

# Bar chart
axs[2].bar(categories, values, color='skyblue', edgecolor='navy')
axs[2].set_title(plot_types[2])
axs[2].grid(True, axis='y', linestyle='--', alpha=0.7)

# Step plot
axs[3].step(x, y, 'g-', where='mid')
axs[3].set_title(plot_types[3])
axs[3].grid(True, linestyle='--', alpha=0.7)

# Fill between
axs[4].fill_between(x, 0, y, alpha=0.5, color='purple')
axs[4].set_title(plot_types[4])
axs[4].grid(True, linestyle='--', alpha=0.7)

# Stem plot
axs[5].stem(categories, values, linefmt='r-', markerfmt='ro', basefmt='k-')
axs[5].set_title(plot_types[5])
axs[5].grid(True, linestyle='--', alpha=0.7)

# Add a main title
fig.suptitle('Gallery of Different Plot Types', fontsize=16)

# Adjust spacing between subplots
plt.tight_layout()
plt.subplots_adjust(top=0.9)  # Make room for the main title

plt.show()
```


Multi-panel plots are excellent for:
- Comparing different variables or aspects of the same dataset
- Showing the same data with different visualization techniques
- Presenting a comprehensive view of complex data
- Creating dashboard-like visualizations

#### Creating Custom Combined Plots

For more complex analytical needs, you can combine multiple plot types into a single visualization:



```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
np.random.seed(42)
x = np.linspace(0, 10, 100)
y = 2 * np.sin(x) + np.random.normal(0, 0.5, 100)  # Sine wave with noise

# Create a figure with a custom layout
fig = plt.figure(figsize=(12, 8))

# Main time series plot (2/3 of the figure height, full width)
ax_main = plt.subplot2grid((3, 3), (0, 0), colspan=3, rowspan=2)

# Histogram on the bottom left
ax_histx = plt.subplot2grid((3, 3), (2, 0), colspan=2)

# Box plot on the bottom right
ax_box = plt.subplot2grid((3, 3), (2, 2))

# Plot the main time series with error region
ax_main.plot(x, y, 'b-', label='Time Series Data')
ax_main.fill_between(x, y-0.5, y+0.5, alpha=0.2, color='blue')
ax_main.set_title('Noisy Sine Wave with Distribution Analysis', fontsize=14)
ax_main.set_xlabel('Time')
ax_main.set_ylabel('Value')
ax_main.grid(True, linestyle='--', alpha=0.7)
ax_main.legend()

# Plot histogram of values
ax_histx.hist(y, bins=20, color='skyblue', edgecolor='navy', alpha=0.7)
ax_histx.set_xlabel('Value')
ax_histx.set_ylabel('Frequency')
ax_histx.grid(True, axis='y', linestyle='--', alpha=0.7)

# Plot box plot
ax_box.boxplot(y, patch_artist=True)
ax_box.set_xticklabels(['Distribution'])
ax_box.grid(True, axis='y', linestyle='--', alpha=0.7)

# Adjust layout
plt.tight_layout()

plt.show()
```


This combined visualization provides multiple perspectives on the same dataset:
1. The main plot shows the time series data with an error region
2. The histogram shows the distribution of values
3. The box plot provides a statistical summary of the distribution

This approach creates a more comprehensive understanding than any single plot type could provide.

#### Interactive Elements in Matplotlib

While Matplotlib is primarily designed for static visualizations, it does offer some basic interactive capabilities:



```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.widgets import Slider, Button

# Create initial data
t = np.linspace(0, 10, 1000)
a = 5
f = 3
s = a * np.sin(2 * np.pi * f * t)

# Create the figure and the line that we will manipulate
fig, ax = plt.subplots(figsize=(10, 8))
line, = ax.plot(t, s, lw=2)
ax.set_xlabel('Time [s]')
ax.set_ylabel('Signal')
ax.grid(True)

# Add sliders for amplitude and frequency
axfreq = plt.axes([0.25, 0.1, 0.65, 0.03], facecolor='lightgoldenrodyellow')
axamp = plt.axes([0.25, 0.15, 0.65, 0.03], facecolor='lightgoldenrodyellow')

sfreq = Slider(axfreq, 'Frequency', 0.1, 10.0, valinit=f, valstep=0.1)
samp = Slider(axamp, 'Amplitude', 0.1, 10.0, valinit=a)

# The function to be called anytime a slider's value changes
def update(val):
    freq = sfreq.val
    amp = samp.val
    line.set_ydata(amp * np.sin(2 * np.pi * freq * t))
    fig.canvas.draw_idle()

# Register the update function with each slider
sfreq.on_changed(update)
samp.on_changed(update)

# Add a reset button
resetax = plt.axes([0.8, 0.025, 0.1, 0.04])
button = Button(resetax, 'Reset', color='lightgoldenrodyellow', hovercolor='0.975')

def reset(event):
    sfreq.reset()
    samp.reset()
    
button.on_clicked(reset)

# Adjust the main plot to make room for the sliders
plt.subplots_adjust(bottom=0.25)

plt.show()
```


This interactive example creates a sine wave visualization with sliders to adjust the amplitude and frequency in real-time. Note that for more advanced interactive visualizations, libraries like Plotly or Bokeh might be more suitable.

*Statistical Visualization:*



```python
# 1-3. Create box and violin plots of different distributions
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Generate three different distributions
n_samples = 200

# Normal distribution (mean=5, std=1)
normal_dist = np.random.normal(5, 1, n_samples)

# Skewed distribution (exponential, shifted)
skewed_dist = np.random.exponential(1.5, n_samples) + 2

# Bimodal distribution (combination of two normal distributions)
bimodal_dist1 = np.random.normal(3, 0.5, n_samples // 2)
bimodal_dist2 = np.random.normal(7, 0.5, n_samples // 2)
bimodal_dist = np.concatenate([bimodal_dist1, bimodal_dist2])

# Combine data into a list
data = [normal_dist, skewed_dist, bimodal_dist]
labels = ['Normal', 'Skewed', 'Bimodal']

# Create a figure with two subplots side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Create box plot in the first subplot
box_plot = ax1.boxplot(data, 
                      patch_artist=True,  # Fill the boxes with color
                      labels=labels,
                      notch=True)         # Add notches for CI of median

# Customize box colors
colors = ['lightblue', 'lightgreen', 'lightpink']
for patch, color in zip(box_plot['boxes'], colors):
    patch.set_facecolor(color)

# Customize whiskers and median lines
for whisker in box_plot['whiskers']:
    whisker.set_color('gray')
    whisker.set_linewidth(1.5)
    
for median in box_plot['medians']:
    median.set_color('red')
    median.set_linewidth(2)

# Add title and labels to box plot
ax1.set_title('Box Plot Comparison of Distributions', fontsize=14)
ax1.set_xlabel('Distribution Type')
ax1.set_ylabel('Value')
ax1.grid(True, linestyle='--', alpha=0.7, axis='y')

# Create violin plot in the second subplot
violin_parts = ax2.violinplot(data, showmeans=True, showmedians=True)

# Customize violin colors
for i, pc in enumerate(violin_parts['bodies']):
    pc.set_facecolor(colors[i])
    pc.set_edgecolor('black')
    pc.set_alpha(0.7)

# Customize other violin parts
for partname in ['cbars', 'cmins', 'cmaxes', 'cmeans', 'cmedians']:
    if partname in violin_parts:
        violin_parts[partname].set_edgecolor('black')
        violin_parts[partname].set_linewidth(1.5)

# Add title and labels to violin plot
ax2.set_title('Violin Plot Comparison of Distributions', fontsize=14)
ax2.set_xlabel('Distribution Type')
ax2.set_ylabel('Value')
ax2.set_xticks([1, 2, 3])
ax2.set_xticklabels(labels)
ax2.grid(True, linestyle='--', alpha=0.7, axis='y')

# Add a main title to the figure
fig.suptitle('Comparison of Different Distribution Types', fontsize=16)

plt.tight_layout()
plt.subplots_adjust(top=0.9)  # Adjust for the main title
plt.show()
```


*Spatial Visualization:*



```python
# 4-5. Create a hexbin plot with correlated data
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Generate correlated data
n_points = 5000
x = np.random.normal(0, 1, n_points)
y = x * 0.7 + np.random.normal(0, 0.5, n_points)  # Correlation with some noise

# Create a figure
plt.figure(figsize=(10, 8))

# Create hexbin plot
hexbin = plt.hexbin(x, y, 
                   gridsize=30,          # Number of hexagons in each direction
                   cmap='inferno',       # Color map
                   bins='log',           # Use logarithmic binning
                   mincnt=1,             # Minimum count to include a bin
                   edgecolors='black',   # Hexagon edge color
                   linewidths=0.1,       # Hexagon edge width
                   alpha=0.8)            # Transparency

# Add a colorbar with a label
cbar = plt.colorbar(hexbin, label='log10(N)')
cbar.set_label('log10(Count)', rotation=270, labelpad=20, fontsize=12)

# Add title and labels
plt.title('Hexbin Plot of Correlated Data', fontsize=14)
plt.xlabel('X Variable', fontsize=12)
plt.ylabel('Y Variable', fontsize=12)

# Add grid (optional, as it might clutter with hexagons)
plt.grid(False)

# Calculate correlation coefficient and add as text
correlation = np.corrcoef(x, y)[0, 1]
plt.text(0.05, 0.95, f'Correlation: {correlation:.2f}', 
         transform=plt.gca().transAxes,  # Use axis coordinates (0-1)
         fontsize=12,
         bbox=dict(facecolor='white', alpha=0.8))

plt.tight_layout()
plt.show()
```

## 📋 Chapter Summary

- **Hexbin Plot**: A 2D binning visualization using hexagonal cells to show point density
