## **Practical Exercise 3.3**: Solutions

Here`s a step-by-step solution using CLI commands, with explanations for each step:

---

**1. Activate your virtual environment.**

-  Open your command line interface (CLI).
-  Activate your environment:  

    ```cmd  
    C:\repos\python-for-engineers\my_first_venv_project\myenv\Scripts>activate
    ```
---

**2. Install the packages `numpy`, `pandas`, `matplotlib` and `scikit-learn`. Try to install all packages with one command.**


- *Input:*    
    ```cmd
    (myenv) C:\repos\python-for-engineers\my_first_venv_project\myenv\Scripts>pip install numpy pandas matplotlib scikit-learn
    ```
---

**6. Create a requirements.txt file:**  

- *Input:*    
    ```cmd
    (myenv) C:\repos\python-for-engineers\my_first_venv_project\myenv\Scripts>pip freeze > requirements.txt
    ```

-->  check with “dir” whether the txt file has been created

---

**7. Use the requirements.txt file from 6. to create a new environment with the same packages and the same versions:**  

- Create a new virtual environment:
    - *Input:*    
        ```cmd
        C:\repos\python-for-engineers\my_first_venv_project>python -m venv myenvnew
        ```

    - check with “dir” whether the txt file has been created
        ```cmd
        C:\repos\python-for-engineers\my_first_venv_project>dir
        ```

    - *Output:*    
        ```cmd
        20.09.2024  16:26    <DIR>          .
        20.09.2024  11:41    <DIR>          ..
        20.09.2024  15:23    <DIR>          myenv
        20.09.2024  16:26    <DIR>          myenvnew
                0 Datei(en),              0 Bytes
                4 Verzeichnis(se), 153.200.152.576 Bytes frei

        C:\repos\python-for-engineers\my_first_venv_project>
    ```

- Copy the requirements.txt into the directory of the new environment:  
    ```cmd
    C:\repos\python-for-engineers\my_first_venv_project\myenv\Scripts> move requirements.txt C:\repos\python-for-engineers\my_first_venv_project\myenvnew\Scripts
    ``` 

- Activate the new environment:  
    ```cmd
    (myenvnew) C:\repos\python-for-engineers\my_first_venv_project\myenvnew\Scripts>activate
    ``` 

- Install packages from requirements.txt:  
    ```cmd
    (myenvnew) C:\repos\python-for-engineers\my_first_venv_project\myenvnew\Scripts>pip install -r requirements.txt
    ```
