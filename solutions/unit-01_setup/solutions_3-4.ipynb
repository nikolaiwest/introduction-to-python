## **Practical Exercise 3.4**: Solutions

Here`s a step-by-step solution using Conda, with explanations for each step:

---

**1. Create a new directory for your project called "my_first_conda_project".**

- Open your command line interface.**
- Navigate to: `C:\repos\python-for-engineers>`
- `mkdir my_first_conda_project`

    *Input:*  
    ```cmd
    C:\repos\python-for-engineers> mkdir my_first_conda_projec
    C:\repos\python-for-engineers>cd my_first_conda_project   
    ```

---


**2. Set up a virtual environment in this directory.**

- `conda create --name conda_venv`
- *Input:*  
    ```cmd
    C:\repos\python-for-engineers\my_first_conda_project> conda create --name conda_venv   
    ```

---

  
**3. Activate the virtual environment.**  

- Windows: `conda activate conda_venv`
- macOS/Linux: `source myenv/bin/activate`

- *Input:*    
    ```cmd
    C:\repos\python-for-engineers\my_first_conda_project> conda activate conda_venv
    ```

- *Output:*    
    ```cmd
    (conda_venv) C:\repos\python-for-engineers\my_first_conda_project>
    ```

---


**4. Create a simple Python script that prints "Hello from my virtual environment!".**  

- Create a file with `echo.>`  
    ```cmd 
    (conda_venv) C:\repos\python-for-engineers\my_first_conda_project> echo.> script.py
    ```

- open file in Visual Studio Code and enter the content:

    ```python
    print("Hello from my virtual environment!")
    ```

- Save the scrpit with `Crtl + S`.

*Note: If you have done practical exercise 2.2 you can copy* `script.py` from C:\repos\python-for-engineers\my_first_venv_project\myenv\Scripts> with `cp <path\file.py> <new_path\new_file.py>`


---


**5. Run your script.**

- `python script.py`

---

**6. Install the packages `numpy`, `pandas`, `matplotlib` and `scikit-learn`. Try to install all packages with one command.**

- `conda install numpy pandas matplotlib scikit-learn`
- Wait a few seconds and confirm with `Y` if necessary.

--- 

**7. List all the packages, that are in your environment.**

- `pip list`

---

**8. Uninstall `matplotlib` and check if it's done with listing your packages again.**

- `conda uninstall matplotlib`

---


**9. `matplotlib` is used in unit 5, so you can install it for the futur Practical Exercises.**

- `conda install matplotlib`

---


**10. Create a Conda environment file.**

`conda env export > environment.yml`

---


**11. Use the Conda environment file to create a new environment with the same packages and the same versions.**
- Create a new environment and activate
- Create a new directory: C:\repos\python-for-engineers\my_second_conda_project
- Copy `environment.yml` in the new directory: C:\repos\python-for-engineers\my_second_conda_project>
- Use `conda env create -f environment.yml`
