## **Practical Exercise 4.3**: Solutions

Here`s a step-by-step solution installing *Pipx and CookieCutter Data Science (CCDS)*, with explanations for each step:

---

Solution - Install CCDS Template:

- Navigate into a folder where you want your project structur.  
    e.g. `C:\Repo\python-for-engineers\programs>cd project_structure`
- Create and **activate** a new virtual environment
    `see practical exercise 2.2`
- Check if python and pip is already installed:  
    `python --version`
    `pip --version`
- Install GIT (needed for CCDS)
    `winget install --id Git.Git -e --source winget`
- Install CCDS
    `pip install cookiecutter-data-science`
- run ccds
    - If you already used ccds, you will be asked to re-download it.
    ```cmd 
    You've downloaded C:\Users\User\.cookiecutters\cookiecutter-data-science
    before. Is it okay to delete and re-download it? [y/n] (y): y
    ```
    - Enter "y" to confirm.
- Depending on your answers, ccds will create the directory template.
- You will find the template in the folder where you installed it.
- To do this, open the repository folder that you created during the query.
- Example: Repository Name --> dsp
    - Template in: `C:\repos\python-for-engineers\my_first_venv_project\myccdsenv\Scripts\dsp`
