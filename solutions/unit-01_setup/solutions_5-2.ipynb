## **Practical Exercise 5.2**: Solutions

Here`s a step-by-step solution using Git for Version Control.

---

**1. Install Git.**

- `winget install --id Git.Git -e --source winget`

---

**2. Create a GitHub account.**

 Go to [github.com](https://github.com/). Enter your Email adress and klick *Sign up for GitHub*. Follow the further instructions at the website to create an account.

 ---

**3. Configure Git.**

- `git config --global user.name "Your Name"`
- `git config --global user.email "your.email@example.com"`

---

**4. Initialize the Git Repository.** `C:\Repo\python-for-engineers\programs\project_structure\python_for_engineers`
- Navigate to the directory with `cd C:\Repo\python-for-engineers\programs\project_structure\python_for_engineers`
- `git init`

---

**5. Copy the `filled_testfile.py` (Practical Exercise 1.3) in `C:\Repo\python-for-engineers\programs\project_structure\python_for_engineers\program_code`.**
- `cd ..\..` or `cd C:\Repo\python-for-engineers\programs`
- `move filled_testfile.py C:\Repo\python-for-engineers\programs\project_structure\python_for_engineers\program_code`

---

**6. Create a new file called `new_testfile.py` and fill the file with `print("First Line")`**

- `echo.> new_testfile.py`
- `echo print(First line") >> new_testfile.py`

---

**7. Stage the changes of both files.**

- `git add filled_testfile.py`
- `git add new_testfile.py`

---

**8. Take a look to the actual staged changes.**

- `git status`

---

**9. Commit both changes with the commit message: `Add 2 files: filled_testfile.py & new_testfile.py`**

- `git commit -m "Add 2 files: filled_testfile.py & new_testfile.py"`

---

**10. Change both of the files with adding `print("Changes in filled testfile")` and `print("Changes in new testfile")`**

- `echo print("Changes in filled testfile") > filled_testfile.py`
- `echo print("Changes in filled testfile") > new_testfile.py`

---

**11. Stage the changes of the `filled_testfile.py`**

- `git add filled_testfile.py`

---

**12. Take a look to the actual staged changes.**

- git status

---

**13. Commit the stanged change with the commit message: `Add print() in filled_testfile.py`**

- `git commit -m "Add print() in filled_testfile.py"`

---

**14. View the commit history.**

- `git log`
