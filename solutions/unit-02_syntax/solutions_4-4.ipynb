*Set creation and modification:*

```python
# 1. Create a set called `fruits` with at least 5 different fruit names.
fruits = {"apple", "banana", "cherry", "date", "elderberry"}
```

```python
# 2. Add a new fruit to the `fruits` set.
fruits.add("fig")
```

```python
# 3. Remove a fruit from the `fruits` set.
fruits.remove("banana")
```

```python
# 4. Try to add a fruit that already exists in the set (notice that it doesn't create a duplicate).
fruits.add("apple")
fruits  # The set remains unchanged
```

*Set operation:*

```python
# 5. Create two sets: `set1` with numbers 1-5 and `set2` with numbers 4-8.
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
```

```python
# 6. Find the union of `set1` and `set2`.
set1 | set2  # or set1.union(set2)
```

```python
# 7. Find the intersection of `set1` and `set2`.
set1 & set2  # or set1.intersection(set2)
```

```python
# 8. Find the difference between `set1` and `set2` (elements in set1 that are not in set2).
set1 - set2  # or set1.difference(set2)
```

*Set comprehension:*

```python
# 9. Use a set comprehension to create a set of the squares of numbers from 1 to 10.
{x**2 for x in range(1, 11)}
```

```python
# 10. Use a set comprehension to create a set of all the consonants in the word "python" (exclude vowels).
{char for char in "python" if char not in "aeiou"}
```
