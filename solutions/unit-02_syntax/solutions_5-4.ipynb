*Clean function design:*

````python
# 1. Refactor the function to follow the Single Responsibility Principle
def validate_user_data(username, email, age):
    """Validate user data."""
    if not username or not email or age < 0:
        print("Invalid data provided")
        return False
    return True

def format_user_data(username, email):
    """Format user data by standardizing case and removing whitespace."""
    username = username.lower().strip()
    email = email.lower().strip()
    return username, email

def save_to_database(username, email, age):
    """Save user data to the database."""
    print(f"Saving user {username} with email {email} and age {age} to database")
    return True

def send_welcome_email(email):
    """Send a welcome email to the user."""
    print(f"Sending welcome email to {email}")
    return True

def process_user_data(username, email, age):
    """Process user data by validating, formatting, saving, and sending a welcome email."""
    # First validate the data
    if not validate_user_data(username, email, age):
        return False
    
    # Format the data
    username, email = format_user_data(username, email)
    
    # Save to database
    save_to_database(username, email, age)
    
    # Send welcome email
    send_welcome_email(email)
    
    return True
````

````python
# 2. Improve the function with proper error handling
def calculate_average(numbers):
    """Calculate the average of a list of numbers."""
    try:
        # Check if the input is empty
        if not numbers:
            raise ValueError("Cannot calculate average of an empty list")
        
        total = sum(numbers)
        average = total / len(numbers)
        return average
    except TypeError as e:
        print(f"Error: Input must be an iterable of numbers. {e}")
        return None
    except ZeroDivisionError:
        print("Error: Cannot divide by zero (empty list)")
        return None
    except Exception as e:
        print(f"Unexpected error: {e}")
        return None
````

*Function Documentation:*


````python
# 3. Add a proper docstring to the function
def calculate_rectangle_area(length, width):
    """
    Calculate the area of a rectangle.
    
    This function multiplies the length and width to determine the area of a rectangle.
    
    Args:
        length (float): The length of the rectangle. Must be a positive number.
        width (float): The width of the rectangle. Must be a positive number.
        
    Returns:
        float: The area of the rectangle (length × width)
        
    Raises:
        TypeError: If length or width are not numbers
        ValueError: If length or width are negative or zero
        
    Examples:
        >>> calculate_rectangle_area(5, 10)
        50
        >>> calculate_rectangle_area(2.5, 3)
        7.5
    """
    if not isinstance(length, (int, float)) or not isinstance(width, (int, float)):
        raise TypeError("Length and width must be numbers")
    
    if length <= 0 or width <= 0:
        raise ValueError("Length and width must be positive numbers")
        
    return length * width
````

*Error Handling:*


````python
# 4. Create a safe_divide function
def safe_divide(numerator, denominator):
    """Safely divide two numbers, handling potential errors."""
    try:
        result = numerator / denominator
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")
        return None
    except TypeError:
        print("Error: Both inputs must be numbers")
        return None
    except Exception as e:
        print(f"Unexpected error: {e}")
        return None
````

````python
# 5. Create a get_element function
def get_element(data_list, index):
    """
    Safely retrieve an element from a list at the specified index.
    
    This function attempts to access an element in a list at the given index.
    It handles potential errors if the index is out of range or if the input
    is not a list.
    
    Args:
        data_list (list): The list to retrieve the element from
        index (int): The position of the element to retrieve (zero-based)
        
    Returns:
        Any: The element at the specified index, or None if an error occurs
        
    Raises:
        TypeError: If data_list is not a list or index is not an integer (handled internally)
        IndexError: If the index is out of range (handled internally)
        
    Examples:
        >>> get_element([10, 20, 30], 1)
        20
        >>> get_element([10, 20, 30], 5)
        Error: Index 5 is out of range for list of length 3
        None
    """
    try:
        if not isinstance(data_list, list):
            raise TypeError("First argument must be a list")
            
        if not isinstance(index, int):
            raise TypeError("Index must be an integer")
            
        return data_list[index]
    except IndexError:
        print(f"Error: Index {index} is out of range for list of length {len(data_list)}")
        return None
    except TypeError as e:
        print(f"Error: {e}")
        return None
    except Exception as e:
        print(f"Unexpected error: {e}")
        return None
````
