## 👨‍💻 Solutions 6.1: Creating Classes and Objects

**Basic class creation:**

1. Create a class called `Rectangle` with attributes for `width` and `height`. Include an `__init__` method to initialize these attributes.

2. Add a method called `area` to the `Rectangle` class that returns the area of the rectangle (width × height).

3. Add another method called `perimeter` that returns the perimeter of the rectangle (2 × (width + height)).

**Working with instances:**

4. Create two different instances of your `Rectangle` class with different dimensions.

5. Calculate and print the area and perimeter of each rectangle using your methods.

**Class and instance attributes:**

6. Add a class attribute called `shape_type` with the value "quadrilateral" to your `Rectangle` class.

7. Add an instance method called `describe` that prints a description of the rectangle, including its dimensions, area, perimeter, and shape type.

*Class and instance attributes:*

```python
# 1-3. Create Rectangle class with area and perimeter methods
class Rectangle:
    """A class representing a rectangle with width and height."""
    
    # Class attribute
    shape_type = "quadrilateral"
    
    def __init__(self, width, height):
        """Initialize rectangle with width and height."""
        self.width = width
        self.height = height
    
    def area(self):
        """Calculate and return the area of the rectangle."""
        return self.width * self.height
    
    def perimeter(self):
        """Calculate and return the perimeter of the rectangle."""
        return 2 * (self.width + self.height)
    
    def describe(self):
        """Return a description of the rectangle."""
        return (f"This {self.shape_type} is a rectangle with width {self.width}, "
                f"height {self.height}, area {self.area()}, and perimeter {self.perimeter()}.")
```

*Working with instances:*

```python
# 4-5. Create Rectangle instances and calculate area/perimeter
# Create two different rectangles
rect1 = Rectangle(5, 10)
rect2 = Rectangle(7, 3)

# Calculate and print area and perimeter for each rectangle
print(f"Rectangle 1 area: {rect1.area()}")  # 50
print(f"Rectangle 1 perimeter: {rect1.perimeter()}")  # 30

print(f"Rectangle 2 area: {rect2.area()}")  # 21
print(f"Rectangle 2 perimeter: {rect2.perimeter()}")  # 20
```

*Class and instance attributes:*

```python
# 6-7. Test class attribute and describe method
print(Rectangle.shape_type)  # Accessing class attribute directly
print(rect1.shape_type)      # Accessing via instance

# Test the describe method
print(rect1.describe())
print(rect2.describe())
```
