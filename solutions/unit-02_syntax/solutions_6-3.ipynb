## 👨‍💻 Solutions 6.3: Properties and Access Control

**Property getters and setters:**

1. Create a class called `Square` with a private attribute `_side_length`. Implement a property called `side_length` with a getter and a setter that validates that the side length is positive.

2. Add a read-only property called `area` that returns the area of the square.

3. Add a read-only property called `perimeter` that returns the perimeter of the square.

**Class and static methods:**

4. Add a class attribute to your `Square` class called `number_of_squares` that keeps track of how many square objects have been created.

5. Add a class method called `report_number_of_squares` that returns a string saying how many squares have been created.

6. Add a static method called `is_square` that takes a rectangle (with width and height) and returns True if it's a square (width equals height) and False otherwise.

*Property getters and setters:*

```python
# Tasks 1-3: Create Square class with properties

class Square:
    """A class representing a square."""
    
    def __init__(self, side_length):
        """Initialize square with a side length."""
        # Task 1: Create a class with private attribute _side_length
        self._side_length = 0  # Initialize first to avoid issues with the setter
        self.side_length = side_length  # This calls the setter method
    
    # Task 1: Implement property with getter and setter
    @property
    def side_length(self):
        """Getter for side_length property."""
        return self._side_length
    
    @side_length.setter
    def side_length(self, value):
        """Setter for side_length property with validation."""
        if value <= 0:
            raise ValueError("Side length must be positive")
        self._side_length = value
    
    # Task 2: Add a read-only property for area
    @property
    def area(self):
        """Calculate and return the area of the square."""
        return self._side_length ** 2
    
    # Task 3: Add a read-only property for perimeter
    @property
    def perimeter(self):
        """Calculate and return the perimeter of the square."""
        return 4 * self._side_length


# Test Tasks 1-3
square = Square(5)
print(f"Side length: {square.side_length}")  # 5
print(f"Area: {square.area}")                # 25
print(f"Perimeter: {square.perimeter}")      # 20

# Test validation
try:
    invalid_square = Square(-3)  # Should raise ValueError
except ValueError as e:
    print(f"Validation error: {e}")

# Test changing the side length
square.side_length = 7
print(f"New side length: {square.side_length}")  # 7
print(f"New area: {square.area}")                # 49
print(f"New perimeter: {square.perimeter}")      # 28
```

*Class and static methods:*

```python
# Tasks 4-6: Add class attribute, class method, and static method

class Square:
    """A class representing a square."""
    
    # Task 4: Add a class attribute to track number of squares
    number_of_squares = 0
    
    def __init__(self, side_length):
        """Initialize square with a side length."""
        self._side_length = 0
        self.side_length = side_length
        
        # Task 4: Increment the count of squares
        Square.number_of_squares += 1
    
    @property
    def side_length(self):
        """Getter for side_length property."""
        return self._side_length
    
    @side_length.setter
    def side_length(self, value):
        """Setter for side_length property with validation."""
        if value <= 0:
            raise ValueError("Side length must be positive")
        self._side_length = value
    
    @property
    def area(self):
        """Calculate and return the area of the square."""
        return self._side_length ** 2
    
    @property
    def perimeter(self):
        """Calculate and return the perimeter of the square."""
        return 4 * self._side_length
    
    # Task 5: Add a class method to report number of squares
    @classmethod
    def report_number_of_squares(cls):
        """Return a string reporting the number of squares created."""
        return f"Number of squares created: {cls.number_of_squares}"
    
    # Task 6: Add a static method to check if a rectangle is a square
    @staticmethod
    def is_square(rectangle):
        """Check if a rectangle is a square (static method)."""
        # Check if the rectangle has width and height attributes
        if not hasattr(rectangle, 'width') or not hasattr(rectangle, 'height'):
            raise AttributeError("Object must have width and height attributes")
        
        return rectangle.width == rectangle.height


# Test Task 4 - Class attribute
s1 = Square(5)
s2 = Square(10)
print(f"Number of squares: {Square.number_of_squares}")  # Should be 2

# Test Task 5 - Class method
print(Square.report_number_of_squares())  # "Number of squares created: 2"

# Test Task 6 - Static method
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

# Create two different rectangles
rect1 = Rectangle(4, 4)  # A square
rect2 = Rectangle(4, 5)  # Not a square

# Test the static method
print(f"Is rect1 a square? {Square.is_square(rect1)}")  # True
print(f"Is rect2 a square? {Square.is_square(rect2)}")  # False
```
