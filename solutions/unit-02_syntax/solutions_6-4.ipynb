## 👨‍💻 Solutions 6.4: Inheritance and Polymorphism

**Creating an inheritance hierarchy:**

1. Create a base class called `Vehicle` with attributes for `make`, `model`, and `year`. Include methods for `start_engine()` and `info()` that returns a string with the vehicle information.

2. Create two subclasses: `Car` and `Motorcycle`. Both should inherit from `Vehicle` and add specific attributes (e.g., `num_doors` for Car, `has_sidecar` for Motorcycle).

3. Override the `start_engine()` method in both subclasses to provide vehicle-specific messages.

**Using polymorphism:**

4. Create a function called `test_drive` that takes any vehicle object and calls its `start_engine()` method and `info()` method, then returns a string saying "Test driving: [vehicle info]".

5. Create several different vehicle objects and use the `test_drive` function with them to demonstrate polymorphism.

**Multiple inheritance:**

6. Create a `FlyingVehicle` class and an `ElectricVehicle` class, each with their own methods and attributes.

7. Create a `FlyingCar` class that inherits from both `Car` and `FlyingVehicle`.

8. Create an `ElectricMotorcycle` class that inherits from both `Motorcycle` and `ElectricVehicle`.

*Creating an inheritance hierarchy:*

```python
# Tasks 1-3: Create Vehicle class hierarchy

# Task 1: Create base Vehicle class
class Vehicle:
    """Base class for all vehicles."""
    
    def __init__(self, make, model, year):
        """Initialize vehicle with make, model, and year."""
        self.make = make
        self.model = model
        self.year = year
    
    def start_engine(self):
        """Start the vehicle's engine."""
        return f"The {self.year} {self.make} {self.model}'s engine is starting..."
    
    def info(self):
        """Return basic information about the vehicle."""
        return f"{self.year} {self.make} {self.model}"


# Task 2: Create Car subclass
class Car(Vehicle):
    """A car is a type of vehicle with doors."""
    
    def __init__(self, make, model, year, num_doors):
        """Initialize a car with make, model, year, and number of doors."""
        # Call the parent class's __init__ method
        super().__init__(make, model, year)
        self.num_doors = num_doors
    
    # Task 3: Override start_engine method
    def start_engine(self):
        """Override the start_engine method for cars."""
        return f"The {self.year} {self.make} {self.model} car turns on with a purr..."
    
    def info(self):
        """Extend the parent info method with car-specific details."""
        return f"{super().info()} - {self.num_doors} door car"


# Task 2: Create Motorcycle subclass
class Motorcycle(Vehicle):
    """A motorcycle is a type of vehicle that may have a sidecar."""
    
    def __init__(self, make, model, year, has_sidecar=False):
        """Initialize motorcycle with make, model, year, and sidecar flag."""
        super().__init__(make, model, year)
        self.has_sidecar = has_sidecar
    
    # Task 3: Override start_engine method
    def start_engine(self):
        """Override the start_engine method for motorcycles."""
        return f"The {self.year} {self.make} {self.model} motorcycle roars to life!"
    
    def info(self):
        """Extend the parent info method with motorcycle-specific details."""
        sidecar_info = "with sidecar" if self.has_sidecar else "without sidecar"
        return f"{super().info()} - Motorcycle {sidecar_info}"


# Test Tasks 1-3
generic_vehicle = Vehicle("Generic", "Transport", 2023)
sedan = Car("Toyota", "Camry", 2022, 4)
sport_bike = Motorcycle("Kawasaki", "Ninja", 2023)
touring_bike = Motorcycle("Harley-Davidson", "Road King", 2021, has_sidecar=True)

# Test method calls
print(generic_vehicle.info())
print(generic_vehicle.start_engine())
print()

print(sedan.info())
print(sedan.start_engine())
print()

print(sport_bike.info())
print(sport_bike.start_engine())
print()

print(touring_bike.info())
print(touring_bike.start_engine())
```

*Using polymorphism:*


```python
# Task 4: Create a test_drive function that works with any vehicle
def test_drive(vehicle):
    """Function that works with any vehicle object (polymorphism).
    
    Args:
        vehicle: Any object from the Vehicle hierarchy
        
    Returns:
        A string describing the test drive experience
    """
    # Call methods without knowing the specific type of vehicle
    engine_sound = vehicle.start_engine()
    vehicle_info = vehicle.info()
    return f"Test driving: {vehicle_info}\nEngine sound: {engine_sound}"
```

```python
# Task 5: Create different vehicles and demonstrate polymorphism
car1 = Car("Toyota", "Camry", 2022, 4)
car2 = Car("Ford", "Mustang", 2023, 2)
motorcycle1 = Motorcycle("Harley-Davidson", "Sportster", 2021)
motorcycle2 = Motorcycle("BMW", "R1250", 2022, has_sidecar=True)
generic = Vehicle("Generic", "Vehicle", 2020)

# Test polymorphic behavior
print("Testing polymorphism with different vehicles:\n")

# Create a list of different vehicle types
vehicles = [car1, car2, motorcycle1, motorcycle2, generic]

# The same function works with all vehicle types
for vehicle in vehicles:
    print(test_drive(vehicle))
    print("-" * 40)
```

*Multiple inheritance:*

```python
# Task 6: Create a FlyingVehicle class
class FlyingVehicle:
    """A mixin class for vehicles that can fly."""
    
    def __init__(self, max_altitude):
        """Initialize flying vehicle with max altitude."""
        self.max_altitude = max_altitude
    
    def take_off(self):
        """Method for taking off."""
        return "Taking off into the air!"
    
    def land(self):
        """Method for landing."""
        return "Landing safely on the ground."


class ElectricVehicle:
    """A mixin class for electric-powered vehicles."""
    
    def __init__(self, battery_capacity):
        """Initialize electric vehicle with battery capacity in kWh."""
        self.battery_capacity = battery_capacity
    
    def charge(self):
        """Method for charging the vehicle."""
        return "Charging the battery..."
    
    def range(self):
        """Calculate the vehicle's range based on battery capacity."""
        # Simple formula for demonstration purposes
        return self.battery_capacity * 4  # 4 miles per kWh
```

```python
# Task 7: Create a FlyingCar class with multiple inheritance
class FlyingCar(Car, FlyingVehicle):
    """A car that can also fly (multiple inheritance)."""
    
    def __init__(self, make, model, year, num_doors, max_altitude):
        """Initialize a flying car with attributes from both parent classes."""
        Car.__init__(self, make, model, year, num_doors)
        FlyingVehicle.__init__(self, max_altitude)
    
    def start_engine(self):
        """Override to provide flying car specific behavior."""
        return f"The {self.year} {self.make} {self.model} flying car starts its hybrid engine."
    
    def info(self):
        """Return comprehensive information about the flying car."""
        return f"{Car.info(self)} - Can fly up to {self.max_altitude} feet"
```

```python
# Task 8: Create an ElectricMotorcycle class with multiple inheritance
class ElectricMotorcycle(Motorcycle, ElectricVehicle):
    """A motorcycle that runs on electricity (multiple inheritance)."""
    
    def __init__(self, make, model, year, battery_capacity, has_sidecar=False):
        """Initialize with attributes from both parent classes."""
        Motorcycle.__init__(self, make, model, year, has_sidecar)
        ElectricVehicle.__init__(self, battery_capacity)
    
    def start_engine(self):
        """Override for electric motorcycle specific behavior."""
        return f"The {self.year} {self.make} {self.model} electric motorcycle silently powers up."
    
    def info(self):
        """Return comprehensive information about the electric motorcycle."""
        return f"{Motorcycle.info(self)} - Electric with {self.battery_capacity} kWh battery"
```

```python
# Test Tasks 6-8
print("Testing multiple inheritance:\n")

# Create instances using multiple inheritance
flying_car = FlyingCar("Terrafugia", "Transition", 2023, 2, 10000)
electric_motorcycle = ElectricMotorcycle("Zero", "SR/F", 2023, 14.4)

# Test FlyingCar
print("Flying Car Tests:")
print(flying_car.info())
print(flying_car.start_engine())
print(flying_car.take_off())
print(flying_car.land())
print()

# Test ElectricMotorcycle
print("Electric Motorcycle Tests:")
print(electric_motorcycle.info())
print(electric_motorcycle.start_engine())
print(electric_motorcycle.charge())
print(f"Range: {electric_motorcycle.range()} miles")

# Show method resolution order (MRO)
print("\nMethod Resolution Order for FlyingCar:")
print([cls.__name__ for cls in FlyingCar.__mro__])

print("\nMethod Resolution Order for ElectricMotorcycle:")
print([cls.__name__ for cls in ElectricMotorcycle.__mro__])
```
