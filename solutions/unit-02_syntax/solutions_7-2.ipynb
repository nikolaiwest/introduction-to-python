*Using standard library packages:*

```python
# 1. Use urllib.request to retrieve website content
import urllib.request

# Retrieve content from example.com
url = "http://example.com"
try:
    response = urllib.request.urlopen(url)
    html = response.read().decode('utf-8')
    # Print first 100 characters
    print(f"First 100 characters from {url}:")
    print(html[:100])
except Exception as e:
    print(f"Error retrieving content: {e}")
```

```python
# 2. Use json to serialize and deserialize data
import json

# Create a Python dictionary
person = {
    "name": "John Doe",
    "age": 30,
    "city": "New York",
    "languages": ["Python", "JavaScript", "SQL"],
    "is_student": False
}

# Convert to JSON string
json_string = json.dumps(person, indent=2)
print("JSON string:")
print(json_string)

# Parse back to Python object
parsed_person = json.loads(json_string)
print("\nParsed back to Python:")
print(f"Name: {parsed_person['name']}")
print(f"Age: {parsed_person['age']}")
print(f"Languages: {', '.join(parsed_person['languages'])}")
```

*Exploring installed packages:*

```python
# 3. Explore the collections package
import collections

# List all attributes in collections module
print("Contents of the collections package:")
collections_contents = [name for name in dir(collections) 
                       if not name.startswith('_')]
print(collections_contents)

# Next, demonstrating a few collections...
```

```python
# Counter (counter are dict subclass for counting hashable objects)
# They are an unordered collection where elements are stored as dictionary keys and their counts are stored as dictionary values

from collections import Counter

print("Using Counter:")

words = ['apple', 'orange', 'apple', 'banana', 'orange', 'apple']
word_counts = Counter(words)
print(word_counts)
```

```python
# namedtuple (named tuple is a tuple with named fields)
# They are useful for creating lightweight immutable objects
from collections import namedtuple


print("Using namedtuple:")

Point = namedtuple('Point', ['x', 'y'])
p = Point(10, 20)
print(p)
print(f"x: {p.x}, y: {p.y}")
```

```python
# deque (deque means double-ended queue)
# deque is preferred over list when we need to append or pop elements from both ends
from collections import deque

print("Using deque:")

d = deque(['a', 'b', 'c'])
d.append('d')
d.appendleft('z')
print(d)
```

```python
# 4. Work with pandas (if installed)
try:
    import pandas as pd
    
    # Create a simple DataFrame
    data = {
        'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 40],
        'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
    }
    
    df = pd.DataFrame(data)
    print("Pandas DataFrame:")
    print(df)
    
    # Basic DataFrame operations
    print("\nDataFrame info:")
    print(f"Shape: {df.shape}")
    print(f"Columns: {df.columns.tolist()}")
    print(f"Average age: {df['Age'].mean()}")
    
except ImportError:
    print("Pandas is not installed. To install it, run: pip install pandas")
```
