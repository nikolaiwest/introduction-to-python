*Exploring the module search path:*

```python
# 1. Print the module search path
import sys

print("Python's module search path:")
for i, path in enumerate(sys.path, 1):
    print(f"{i}. {path}")

# Identify the current directory
import os
current_dir = os.getcwd()
print(f"\nCurrent working directory: {current_dir}")

# Check if current directory is in the search path
if current_dir in sys.path:
    print("Current directory is in the search path")
else:
    print("Current directory is NOT in the search path")
```

```python
# 2. Write a function to check if a module exists
import importlib.util

def module_exists(module_name):
    """Check if a module can be imported without actually importing it.
    
    Args:
        module_name: Name of the module to check
        
    Returns:
        bool: True if the module exists, False otherwise
    """
    return importlib.util.find_spec(module_name) is not None

# Test with modules that should exist
print("\nChecking if modules exist:")
modules_to_check = ['sys', 'os', 'math', 'random', 'nonexistent_module',
                    'another_fake_module', 'numpy', 'django']

for module in modules_to_check:
    exists = module_exists(module)
    print(f"Module '{module}': {'Exists' if exists else 'Does not exist'}")
```

```python
# 4. Create a function that tries multiple imports
def get_math_module():
    """Try to import either numpy or math, in that order.
    
    Returns:
        module: The first available module, or None if neither is available
    """
    try:
        # First try numpy (more advanced)
        import numpy as np
        print("NumPy is available - using it for math operations")
        return np
    except ImportError:
        print("NumPy not found - falling back to math module")
        try:
            # Fallback to standard math module
            import math
            print("Math module is available")
            return math
        except ImportError:
            # This is very unlikely
            print("Neither NumPy nor math module is available!")
            return None

# Test the function
print("\nTesting multiple import function:")
math_module = get_math_module()

if math_module:
    print("Testing imported module:")
    # Try basic math operation with whichever module we got
    if hasattr(math_module, 'sqrt'):
        print(f"Square root of 16: {math_module.sqrt(16)}")
    else:
        print("Module doesn't have sqrt function")
```
