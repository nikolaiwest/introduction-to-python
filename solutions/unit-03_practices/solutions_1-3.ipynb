*Type Aliases:*

```python
# 3. Type aliases for complex data structures

from typing import TypeAlias, List, Dict, Union, Optional

# Define type aliases for student data
StudentId: TypeAlias = int
Grade: TypeAlias = float
StudentRecord: TypeAlias = Dict[str, Union[StudentId, str, Grade, List[str]]]
ClassRoster: TypeAlias = List[StudentRecord]
GradeBook: TypeAlias = Dict[str, ClassRoster]

# With Python 3.10+:
# StudentId = int
# Grade = float
# StudentRecord = dict[str, StudentId | str | Grade | list[str]]
# ClassRoster = list[StudentRecord]
# GradeBook = dict[str, ClassRoster]

# Function using the type aliases
def get_top_students(
    grade_book: GradeBook,
    course: str,
    min_grade: Grade = 90.0
) -> List[StudentRecord]:
    """Get students with grades at or above the minimum for a course."""
    if course not in grade_book:
        return []

    return [
        student for student in grade_book[course]
        if student.get("grade", 0.0) >= min_grade
    ]

# Example usage
science_students: ClassRoster = [
    {"id": 10001, "name": "Alex", "grade": 92.5, "interests": ["astronomy", "geology"]},
    {"id": 10002, "name": "Barbara", "grade": 88.0, "interests": ["chemistry", "biology"]},
    {"id": 10003, "name": "Carlos", "grade": 95.5, "interests": ["physics", "astronomy"]},
    {"id": 10004, "name": "Diana", "grade": 79.0, "interests": ["biology", "medicine"]}
]

math_students: ClassRoster = [
    {"id": 10001, "name": "Alex", "grade": 89.0, "interests": ["algebra", "statistics"]},
    {"id": 10005, "name": "Elena", "grade": 94.5, "interests": ["calculus", "geometry"]},
    {"id": 10006, "name": "Faisal", "grade": 91.0, "interests": ["statistics", "algebra"]}
]

school_grade_book: GradeBook = {
    "Science 101": science_students,
    "Math 101": math_students
}

top_science_students = get_top_students(school_grade_book, "Science 101")
print(f"Top science students: {[s['name'] for s in top_science_students]}")
```
