*Basic Arithmetic:*

````python
# 1. Create two 1D arrays
import numpy as np

array_a = np.array([10, 20, 30, 40])
array_b = np.array([1, 2, 3, 4])

print(f"Array A: {array_a}")
print(f"Array B: {array_b}")
````

````python
# 2. Perform arithmetic operations
addition = array_a + array_b
subtraction = array_a - array_b
multiplication = array_a * array_b
division = array_a / array_b

print(f"Addition: {addition}")        # [11 22 33 44]
print(f"Subtraction: {subtraction}")  # [9 18 27 36]
print(f"Multiplication: {multiplication}")  # [10 40 90 160]
print(f"Division: {division}")        # [10. 10. 10. 10.]
````


*Operations with Scalars:*



````python
# 3. Scalar operations with array_a
add_five = array_a + 5
multiply_two = array_a * 2
sqrt_values = np.sqrt(array_a)
squared = array_a ** 2

print(f"Original array: {array_a}")
print(f"Add 5: {add_five}")          # [15 25 35 45]
print(f"Multiply by 2: {multiply_two}")  # [20 40 60 80]
print(f"Square root: {sqrt_values}")  # [3.16 4.47 5.48 6.32]
print(f"Squared: {squared}")          # [100 400 900 1600]
````


*Matrix Operations:*



````python
# 4. Create two 2×2 matrices
matrix_a = np.array([[1, 2], [3, 4]])
matrix_b = np.array([[5, 6], [7, 8]])

print(f"Matrix A:\n{matrix_a}")
print(f"Matrix B:\n{matrix_b}")
````

````python
# 5. Matrix operations
element_wise_product = matrix_a * matrix_b
matrix_product = matrix_a @ matrix_b  # or np.dot(matrix_a, matrix_b)
sum_matrix_a = np.sum(matrix_a)
sum_matrix_b = np.sum(matrix_b)

print(f"Element-wise product:\n{element_wise_product}")
print(f"Matrix product:\n{matrix_product}")
print(f"Sum of Matrix A: {sum_matrix_a}")  # 10
print(f"Sum of Matrix B: {sum_matrix_b}")  # 26
````


*Combining Operations:*



````python
# 6. Combined operations in a single expression
values = np.array([1, 2, 3, 4, 5])
result = (values * 2 + 10) ** 2

print(f"Original values: {values}")
print(f"Result of (values * 2 + 10) ** 2: {result}")
````
