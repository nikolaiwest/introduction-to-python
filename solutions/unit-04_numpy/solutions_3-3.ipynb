*Simple Broadcasting:*



````python
# 1-2. Working with prices
import numpy as np

prices = np.array([10, 20, 30, 40, 50])

# Broadcasting operations
prices_plus_5 = prices + 5
discounted_prices = prices * 0.9  # 10% discount
prices_with_tax = prices * 1.08  # 8% tax

print(f"Original prices: {prices}")
print(f"Prices + 5: {prices_plus_5}")
print(f"After 10% discount: {discounted_prices}")
print(f"After 8% tax: {prices_with_tax}")
````


*Matrix Operations with Broadcasting:*



````python
# 3-4. Broadcasting with matrices
# Create a 3×4 matrix of random integers
data_matrix = np.random.randint(1, 100, size=(3, 4))
print(f"Original matrix:\n{data_matrix}")

# Subtract the mean of each column
column_means = np.mean(data_matrix, axis=0)  # Shape: (4,)
centered_matrix = data_matrix - column_means  # Broadcasting
print(f"Column means: {column_means}")
print(f"Centered matrix:\n{centered_matrix}")

# Normalize each row by dividing by row sum
row_sums = np.sum(data_matrix, axis=1).reshape(-1, 1)  # Shape: (3,1)
normalized_rows = data_matrix / row_sums  # Broadcasting
print(f"Row sums: {row_sums.flatten()}")
print(f"Row-normalized matrix:\n{normalized_rows}")

# Add a row vector to each row
row_vector = np.array([1, 2, 3, 4])  # Shape: (4,)
matrix_plus_vector = data_matrix + row_vector  # Broadcasting
print(f"Row vector: {row_vector}")
print(f"Matrix after adding vector to each row:\n{matrix_plus_vector}")
````


*Advanced Broadcasting:*



````python
# 5-6. Working with sales data
sales = np.array([
    [100, 120, 140, 90],    # Product 1
    [80, 95, 110, 70],      # Product 2
    [150, 160, 170, 180]    # Product 3
])
print(f"Sales data:\n{sales}")

# Calculate percentage contribution of each quarter to annual sales
annual_sales = np.sum(sales, axis=1).reshape(-1, 1)  # Shape: (3,1)
quarter_percentages = (sales / annual_sales) * 100  # Broadcasting
print(f"Annual sales by product: {annual_sales.flatten()}")
print(f"Quarter percentages:\n{quarter_percentages}")

# Apply different discount rates to each product
discount_rates = np.array([0.1, 0.15, 0.2]).reshape(-1, 1)  # Shape: (3,1)
discounted_sales = sales * (1 - discount_rates)  # Broadcasting
print(f"Discount rates by product: {discount_rates.flatten()}")
print(f"Discounted sales:\n{discounted_sales}")

# Calculate difference from average sales
product_averages = np.mean(sales, axis=1).reshape(-1, 1)  # Shape: (3,1)
sales_difference = sales - product_averages  # Broadcasting
print(f"Average sales by product: {product_averages.flatten()}")
print(f"Difference from average:\n{sales_difference}")
````


*Real-world Application - Temperature Conversion:*



````python
# 7-8. Temperature conversions
celsius_temps = np.array([20, 22, 19, 25, 27, 24, 21])
print(f"Daily temperatures in Celsius: {celsius_temps}")

# Convert to Fahrenheit: F = C × 9/5 + 32
fahrenheit_temps = (celsius_temps * 9/5) + 32
print(f"Daily temperatures in Fahrenheit: {fahrenheit_temps}")

# Calculate difference from weekly average
weekly_avg = np.mean(celsius_temps)
temp_diff_from_avg = celsius_temps - weekly_avg
print(f"Weekly average temperature: {weekly_avg:.1f}°C")
print(f"Difference from weekly average: {temp_diff_from_avg}")

# Compare with monthly normal temperatures
normal_temps = np.array([18, 19, 19, 20, 21, 21, 20])
diff_from_normal = celsius_temps - normal_temps
print(f"Monthly normal temperatures: {normal_temps}")
print(f"Difference from normal: {diff_from_normal}")
````
