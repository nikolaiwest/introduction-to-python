*Basic Reshaping:*

````python
# 1-3. Basic reshaping tasks
import numpy as np

# Create a 1D array with values from 0 to 23
arr = np.arange(24)
print(f"Original array: {arr}")
print(f"Shape: {arr.shape}\n")

# Reshape into a 4×6 matrix
matrix_4x6 = arr.reshape(4, 6)
print(f"4×6 matrix:\n{matrix_4x6}")
print(f"Shape: {matrix_4x6.shape}\n")

# Reshape into a 2×3×4 3D array
array_3d = arr.reshape(2, 3, 4)
print(f"2×3×4 3D array:\n{array_3d}")
print(f"Shape: {array_3d.shape}\n")

# Reshape into a 3×2×2×2 4D array
array_4d = arr.reshape(3, 2, 2, 2)
print(f"3×2×2×2 4D array:\n{array_4d}")
print(f"Shape: {array_4d.shape}\n")
````


*Using Automatic Dimension Calculation:*



````python
# 4-5. Using automatic dimension calculation
# Create an array of 30 evenly spaced values
values = np.linspace(0, 10, 30)
print(f"Original array: {values}")
print(f"Shape: {values.shape}\n")

# Reshape into a 5×? matrix
matrix_5x = values.reshape(5, -1)
print(f"5×? matrix:\n{matrix_5x}")
print(f"Shape: {matrix_5x.shape}\n")  # Should be (5, 6)

# Reshape into a 2×3×? 3D array
array_2x3x = values.reshape(2, 3, -1)
print(f"2×3×? 3D array:\n{array_2x3x}")
print(f"Shape: {array_2x3x.shape}\n")  # Should be (2, 3, 5)
````


*Flattening and Raveling:*



````python
# 6-8. Flattening and raveling
# Create a 3×4 matrix with random integers
matrix = np.random.randint(1, 101, size=(3, 4))
print(f"Original matrix:\n{matrix}")
print(f"Shape: {matrix.shape}\n")

# Flatten using flatten() - creates a copy
flat_copy = matrix.flatten()
print(f"Flattened (copy): {flat_copy}")

# Flatten using ravel() - creates a view when possible
flat_view = matrix.ravel()
print(f"Raveled (view): {flat_view}\n")

# Modify the first element of each
flat_copy[0] = 0
flat_view[0] = 0

print(f"After modifying flattened arrays:")
print(f"Original matrix:\n{matrix}")
print(f"Flattened (modified): {flat_copy}")
print(f"Raveled (modified): {flat_view}")
print("Note: The original matrix is only affected by the ravel() modification")
````


*Transposing:*



````python
# 9-11. Transposing arrays
# Create a 3×4 matrix of sequential numbers
matrix = np.arange(12).reshape(3, 4)
print(f"Original 3×4 matrix:\n{matrix}")
print(f"Shape: {matrix.shape}\n")

# Transpose the matrix
transposed = matrix.T
print(f"Transposed matrix:\n{transposed}")
print(f"Shape: {transposed.shape}\n")

# Create a 2×3×4 3D array
tensor = np.arange(24).reshape(2, 3, 4)
print(f"Original 3D array shape: {tensor.shape}")

# Default transpose (reverse axes)
transposed_default = tensor.transpose()
print(f"Default transpose shape: {transposed_default.shape}")

# Custom axis ordering (2, 0, 1)
transposed_custom = tensor.transpose(2, 0, 1)
print(f"Custom transpose shape: {transposed_custom.shape}")
````
