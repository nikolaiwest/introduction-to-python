*Basic Stacking:*

````python
# 1. Create the arrays for basic stacking
import numpy as np

# Create three 1D arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
c = np.array([7, 8, 9])

print("Original arrays:")
print(f"a: {a}")
print(f"b: {b}")
print(f"c: {c}\n")
````

````python
# 2. Use different stacking methods to create:

# Stack as rows
rows_array = np.vstack((a, b, c))
print(f"Stacked as rows:\n{rows_array}")
print(f"Shape: {rows_array.shape}\n")  # (3, 3)

# Stack as columns
# First reshape each 1D array into a column vector
cols_array = np.column_stack((a, b, c))
print(f"Stacked as columns:\n{cols_array}")
print(f"Shape: {cols_array.shape}\n")  # (3, 3)

# Stack as depth layers (3D array)
depth_array = np.stack((a, b, c), axis=2)
print(f"Stacked as depth layers (showing first layer):\n{depth_array[:,:,0]}")
print(f"Shape: {depth_array.shape}\n")  # (3, 3, 3)
````


*Working with 2D Arrays:*



````python
# 3. Create new arrays for stacking of 2D arrays

# Create two 2×3 matrices
matrix1 = np.array([[1, 2, 3], [4, 5, 6]])
matrix2 = np.array([[7, 8, 9], [10, 11, 12]])

print("Original matrices:")
print(f"Matrix 1:\n{matrix1}")
print(f"Matrix 2:\n{matrix2}\n")
````

````python
# 4. Apply the appropriate stacking methods 

# Stack vertically (along rows)
vertical_stack = np.vstack((matrix1, matrix2))
print(f"Vertical stack:\n{vertical_stack}")
print(f"Shape: {vertical_stack.shape}\n")  # (4, 3)

# Stack horizontally (along columns)
horizontal_stack = np.hstack((matrix1, matrix2))
print(f"Horizontal stack:\n{horizontal_stack}")
print(f"Shape: {horizontal_stack.shape}\n")  # (2, 6)

# Stack to create a 2×2×3 array (2 layers, each 2×3)
depth_stack = np.stack((matrix1, matrix2), axis=0)
print(f"Depth stack (showing first layer):\n{depth_stack[0]}")
print(f"Shape: {depth_stack.shape}")  # (2, 2, 3)
````


*General Stacking:*



````python
# 5. Create three 2×2 matrices
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
C = np.array([[9, 10], [11, 12]])

print("Original matrices:")
print(f"Matrix A:\n{A}")
print(f"Matrix B:\n{B}")
print(f"Matrix C:\n{C}\n")
````

````python
# 6. Use concatenations vertically and horizontally

# Join vertically with concatenate
vertical_concat = np.concatenate((A, B, C), axis=0)
print(f"Vertical concatenation:\n{vertical_concat}")
print(f"Shape: {vertical_concat.shape}\n")  # (6, 2)

# Join horizontally with concatenate
horizontal_concat = np.concatenate((A, B, C), axis=1)
print(f"Horizontal concatenation:\n{horizontal_concat}")
print(f"Shape: {horizontal_concat.shape}")  # (2, 6)
````


*Basic Splitting:*



````python
# 7. Create a 6×6 array of sequential numbers (0 to 35).

# Create a 6×6 array of sequential numbers
arr = np.arange(36).reshape(6, 6)
print(f"Original 6×6 array:\n{arr}\n")
````

````python
# 8. Splitting the arrays

# Split into 3 equal parts along axis 0 (rows)
row_splits = np.split(arr, 3, axis=0)
print("Split into 3 parts along rows:")
for i, split in enumerate(row_splits):
    print(f"Part {i+1}, shape {split.shape}:\n{split}\n")

# Split into 2 equal parts along axis 1 (columns)
col_splits = np.split(arr, 2, axis=1)
print("Split into 2 parts along columns:")
for i, split in enumerate(col_splits):
    print(f"Part {i+1}, shape {split.shape}:\n{split}\n")

# Split both ways (first rows, then columns)
print("Splitting both ways:")
row_then_col_splits = []
for row_split in row_splits:
    col_sub_splits = np.split(row_split, 2, axis=1)
    row_then_col_splits.extend(col_sub_splits)
    
for i, split in enumerate(row_then_col_splits):
    print(f"Part {i+1}, shape {split.shape}:\n{split}")
````


*Handling Uneven Splits:*



````python
# 9. Create a 5×5 array of random integers.

# Create a 5×5 array of random integers
random_arr = np.random.randint(1, 100, size=(5, 5))
print(f"Original 5×5 array:\n{random_arr}\n")
````

````python
# 10. Use `np.array_split()` to split this array

# Split into 3 parts along rows (5 is not divisible by 3)
row_splits = np.array_split(random_arr, 3, axis=0)
print("Uneven split into 3 parts along rows:")
for i, split in enumerate(row_splits):
    print(f"Part {i+1}, shape {split.shape}:\n{split}\n")

# Split into 4 parts along columns (5 is not divisible by 4)
col_splits = np.array_split(random_arr, 4, axis=1)
print("Uneven split into 4 parts along columns:")
for i, split in enumerate(col_splits):
    print(f"Part {i+1}, shape {split.shape}:\n{split}")
````
