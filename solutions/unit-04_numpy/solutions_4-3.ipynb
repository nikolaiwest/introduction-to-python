*Working with Built-in ufuncs:*

````python
# 1-2. Working with trigonometric ufuncs
import numpy as np

# Create an array of 12 angles from 0 to 2π
angles = np.linspace(0, 2*np.pi, 12)
print(f"Angles (radians): {angles}")
print(f"Angles (degrees): {angles * 180/np.pi}\n")

# Calculate trigonometric functions
sines = np.sin(angles)
cosines = np.cos(angles)
tangents = np.tan(angles)

print(f"Sine values: {sines}")
print(f"Cosine values: {cosines}")
print(f"Tangent values: {tangents}\n")

# Recover original angles using arcsine
# Note: arcsin only returns values in [-π/2, π/2], so we need to handle this
recovered_angles = np.arcsin(sines)
print(f"Recovered angles from arcsin: {recovered_angles}")
print(f"Note: arcsin only returns values in [-π/2, π/2]\n")

# Verify the identity sin²(θ) + cos²(θ) = 1
identity_check = np.sin(angles)**2 + np.cos(angles)**2
print(f"sin²(θ) + cos²(θ) = {identity_check}")
print(f"All values ≈ 1? {np.allclose(identity_check, 1)}")
````


*Comparison and Logical ufuncs:*



````python
# 3-4. Comparison and logical ufuncs
a = np.array([5, 7, 2, 9, 4, 8, 3])
b = np.array([6, 3, 9, 5, 7, 1, 4])

print(f"Array a: {a}")
print(f"Array b: {b}\n")

# Find elements where a > b
greater_mask = np.greater(a, b)
print(f"a > b: {greater_mask}")
print(f"Elements where a > b: {a[greater_mask]}")
print(f"Count: {np.sum(greater_mask)}\n")

# Find elements where both arrays have values > 5
both_gt5_mask = (a > 5) & (b > 5)
print(f"Both > 5: {both_gt5_mask}")
print(f"Elements in a where both > 5: {a[both_gt5_mask]}")
print(f"Elements in b where both > 5: {b[both_gt5_mask]}")
print(f"Count: {np.sum(both_gt5_mask)}\n")

# Find elements where either array has values < 4
either_lt4_mask = (a < 4) | (b < 4)
print(f"Either < 4: {either_lt4_mask}")
print(f"Elements in a where either < 4: {a[either_lt4_mask]}")
print(f"Elements in b where either < 4: {b[either_lt4_mask]}")
print(f"Count: {np.sum(either_lt4_mask)}")
````


*Using Output Parameters:*



````python
# 5. Timing with output parameters
import numpy as np
import time

# Create large arrays
size = 1_000_000
a = np.random.random(size)
b = np.random.random(size)

# Time standard approach
start_time = time.time()
for _ in range(100):  # Run multiple times for more reliable timing
    c1 = np.add(a, b)
standard_time = (time.time() - start_time) / 100
print(f"Standard approach average time: {standard_time:.9f} seconds")

# Time with out parameter
c2 = np.empty_like(a)
start_time = time.time()
for _ in range(100):
    np.add(a, b, out=c2)
out_param_time = (time.time() - start_time) / 100
print(f"Using out parameter average time: {out_param_time:.9f} seconds")

# Compare
print(f"Speed improvement: {(standard_time / out_param_time - 1) * 100:.2f}%")
print(f"Results are identical: {np.array_equal(c1, c2)}")
````


*Creating Custom ufuncs:*



````python
# 6-7. Custom temperature categorization ufunc
import numpy as np

# Define a function to categorize temperatures
def categorize_temp(celsius):
    if celsius < 0:
        return "Freezing"
    elif celsius < 15:
        return "Cold"
    elif celsius < 25:
        return "Mild"
    elif celsius < 35:
        return "Warm"
    else:
        return "Hot"

# Vectorize the function
vectorized_categorize = np.vectorize(categorize_temp)

# Create an array of temperatures
temperatures = np.array([-5, 0, 10, 20, 30, 40])
print(f"Temperatures (°C): {temperatures}")

# Apply the vectorized function
categories = vectorized_categorize(temperatures)
print(f"Temperature categories: {categories}")

# Alternative approach with boolean indexing
categories_alt = np.empty_like(temperatures, dtype='U8')  # Unicode strings up to 8 chars
categories_alt[temperatures < 0] = "Freezing"
categories_alt[(temperatures >= 0) & (temperatures < 15)] = "Cold"
categories_alt[(temperatures >= 15) & (temperatures < 25)] = "Mild"
categories_alt[(temperatures >= 25) & (temperatures < 35)] = "Warm"
categories_alt[temperatures >= 35] = "Hot"

print(f"Categories (alternative approach): {categories_alt}")
print(f"Both methods give same result: {np.array_equal(categories, categories_alt)}")
````
