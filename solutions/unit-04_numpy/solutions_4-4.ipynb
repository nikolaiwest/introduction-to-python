*Boolean Indexing:*

````python
# 1-2. Boolean indexing

import numpy as np

# Create a 5×5 array of random integers
data = np.random.randint(0, 100, size=(5, 5))
print(f"Original array:\n{data}\n")

# Find all elements greater than 50
gt_50_mask = data > 50
gt_50_values = data[gt_50_mask]
print(f"Elements > 50: {gt_50_values}")
print(f"Count: {len(gt_50_values)}\n")

# Find elements that are even and greater than 30
even_gt_30_mask = (data % 2 == 0) & (data > 30)
even_gt_30_values = data[even_gt_30_mask]
print(f"Elements that are even and > 30: {even_gt_30_values}")
print(f"Count: {len(even_gt_30_values)}\n")

# Find elements that are either less than 20 or greater than 80
lt_20_or_gt_80_mask = (data < 20) | (data > 80)
lt_20_or_gt_80_values = data[lt_20_or_gt_80_mask]
print(f"Elements < 20 or > 80: {lt_20_or_gt_80_values}")
print(f"Count: {len(lt_20_or_gt_80_values)}")
````


*Multi-dimensional Boolean Indexing:*



````python
# 3-4. Multi-dimensional boolean indexing

# Create a 4×6 array of random integers
data = np.random.randint(0, 10, size=(4, 6))
print(f"Array:\n{data}\n")

# Select rows where the sum of the row is greater than 30
row_sums = np.sum(data, axis=1)
print(f"Row sums: {row_sums}")
row_mask = row_sums > 30
selected_rows = data[row_mask]
print(f"Rows with sum > 30:\n{selected_rows}\n")

# Select columns where the maximum value in the column is greater than 8
col_maxes = np.max(data, axis=0)
print(f"Column maximums: {col_maxes}")
col_mask = col_maxes > 8
# Need to use a different approach since direct column boolean indexing would flatten
selected_cols = data[:, col_mask]
print(f"Columns with max > 8:\n{selected_cols}\n")

# Find the coordinates of all cells with the value 0
zero_positions = np.where(data == 0)
print(f"Coordinates of zeros: {list(zip(zero_positions[0], zero_positions[1]))}")
````


*Integer Array Indexing:*



````python
# 5-6. Integer array indexing

# Create a 1D array with values from 10 to 19
arr = np.arange(10, 20)
print(f"Original array: {arr}\n")

# Select elements at positions 0, 3, and 8
indices = np.array([0, 3, 8])
selected = arr[indices]
print(f"Elements at positions {indices}: {selected}\n")

# Reverse the array
reversed_arr = arr[::-1]  # Slice approach
reversed_arr_2 = arr[np.arange(len(arr)-1, -1, -1)]  # Integer indexing approach
print(f"Reversed array (slice): {reversed_arr}")
print(f"Reversed array (integer indexing): {reversed_arr_2}\n")

# Create a new array by selecting elements in a random order
random_indices = np.random.permutation(len(arr))
random_arr = arr[random_indices]
print(f"Random indices: {random_indices}")
print(f"Array in random order: {random_arr}")
````


*Combined Indexing Techniques:*



````python
# 7-8. Combined indexing techniques

# Create a 5×5 array of random integers
data = np.random.randint(0, 100, size=(5, 5))
print(f"Original array:\n{data}\n")

# Select the first three columns from rows where the row sum exceeds 200
row_sums = np.sum(data, axis=1)
print(f"Row sums: {row_sums}")
high_sum_rows = row_sums > 200
selected = data[high_sum_rows, :3]
print(f"First three columns from rows with sum > 200:\n{selected}\n")

# Replace all odd numbers greater than 50 with -1
odd_gt_50_mask = (data % 2 == 1) & (data > 50)
data_modified = data.copy()
data_modified[odd_gt_50_mask] = -1
print(f"After replacing odd numbers > 50 with -1:\n{data_modified}\n")

# Extract the diagonal elements where the value is greater than 30
diag_elements = np.diag(data)
print(f"Diagonal elements: {diag_elements}")
high_diag_mask = diag_elements > 30
high_diags = diag_elements[high_diag_mask]
print(f"Diagonal elements > 30: {high_diags}")
````
