*Basic CSV Importing:*

```python
# Import the required libraries (remember to use a venv)
import numpy as np
import pandas as pd
```

```python
# 1. Import employees.csv into a DataFrame
employees_df = pd.read_csv('data/employees.csv')
print("Basic CSV import:")
employees_df.head()
```

```python
# 2. Variations of importing

# Use index_col to set 'id' as the index
employees_indexed = pd.read_csv('data/employees.csv', index_col='id')
print("With 'id' as index:")
employees_indexed.head()
```

```python
# Use dtype to specify 'salary' as float
employees_typed = pd.read_csv('data/employees.csv', dtype={'salary': float})
print("With 'salary' as float:")
employees_typed.dtypes
```

```python
# Use usecols to import only specific columns
employees_subset = pd.read_csv('data/employees.csv', usecols=['name', 'salary'])
print("Only 'name' and 'salary' columns:")
employees_subset.head()
```


*Working with Different Delimiters:*



```python
# 3. Import sales.csv with semicolon delimiter
sales_df = pd.read_csv('data/sales.csv', delimiter=';')
print("Sales data with semicolon delimiter:")
sales_df.head()
```

```python
# 4. Converting dates during import
# Import with parse_dates
sales_dates = pd.read_csv('data/sales.csv', delimiter=';', parse_dates=['date'])

print("Date column type:", sales_dates['date'].dtype)

print("Sales data with parsed dates:")
sales_dates.head()
```


*Handling Missing Values During Import:*



```python
# 5. Handling missing values during import

# Import students.csv with na_values
students_df = pd.read_csv('data/students.csv', na_values=[''])

print("Data types:", students_df.dtypes)

print("Students data with empty strings as NaN:")
students_df
```


*Working with Excel Data:*



```python
# 6. Importing Excel data - first sheet only
try:
    products_df = pd.read_excel('data/inventory.xlsx')
    print("Excel import - first sheet:")
    print(products_df.head())
except FileNotFoundError:
    print("inventory.xlsx file not found")
```

```python
# 7. Importing all Excel sheets
try:
    all_sheets = pd.read_excel('data/inventory.xlsx', sheet_name=None)
    print("All Excel sheets:")
    for sheet_name, sheet_df in all_sheets.items():
        print(f"Sheet: {sheet_name}")
        print(sheet_df.head())
except FileNotFoundError:
    print("inventory.xlsx file not found")
```

*Import from JSON:*

```python
# 8. Importing JSON data
products_json = pd.read_json('data/products.json')
print("JSON import:")
products_json.head()
```

```python
# 9. JSON import with orient parameter
products_records = pd.read_json('data/products.json', orient='records')
print("JSON import with orient='records':")
print(products_records.head())
print("Comparison: Are the DataFrames identical?", products_json.equals(products_records))
```
