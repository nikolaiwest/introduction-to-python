*Working with Sample Data:*

````python
import pandas as pd
import numpy as np

# 1. Create a sample DataFrame

np.random.seed(42)  # For reproducible results

# Create a sample DataFrame
data = {
    'product_id': range(1, 21),
    'category': np.random.choice(['Electronics', 'Clothing', 'Books'], 20),
    'price': np.random.uniform(10, 500, 20).round(2),
    'stock': np.random.randint(0, 50, 20),
    'listed_date': pd.date_range(start='2023-01-01', periods=20)
}

df = pd.DataFrame(data)
df
````

````python
# 2. Exploring the DataFrame with head() and tail()
print("First 5 rows using head()")
print(df.head())
# head() shows the first n rows (default is 5)
# This gives us a quick look at the beginning of the dataset

print("Last 3 rows using tail(3)")
print(df.tail(3))
# tail() shows the last n rows (default is 5, but we specified 3)
# This lets us see what's at the end of our dataset
````

````python
# 3. Examining the DataFrame structure
print("DataFrame dimensions (shape)")
print(df.shape)
# shape returns a tuple: (rows, columns)
# In this case, we should see (20, 5)

print("Column names")
print(df.columns.tolist())
# columns attribute provides the column labels
# tolist() converts it to a standard Python list for easier reading

print("\Data types of each column")
print(df.dtypes)
# dtypes shows the data type of each column
# Note how pandas automatically selected appropriate types for each column
````

````python
# 4. Getting summary information
print("DataFrame info()")
df.info()
# info() provides a concise summary of the DataFrame
# Shows index type, column names, non-null counts, dtypes, and memory usage

print("Summary statistics with describe()")
print(df.describe())
# describe() generates summary statistics for numerical columns
# Includes count, mean, std, min, 25%, 50% (median), 75%, max
````

````python
# 5. Basic column exploration
print("Price column statistics")
price_mean = df['price'].mean()
price_median = df['price'].median()
price_std = df['price'].std()
print(f"Mean price: ${price_mean:.2f}")
print(f"Median price: ${price_median:.2f}")
print(f"Standard deviation: ${price_std:.2f}")
# These statistics help understand the central tendency and spread of prices

print("Stock min and max")
stock_min = df['stock'].min()
stock_max = df['stock'].max()
print(f"Minimum stock: {stock_min}")
print(f"Maximum stock: {stock_max}")
# This shows the range of our inventory levels
````

*Exploring Categorical Data:*

````python
# 6. Exploring categorical data
print("\n--- Category counts ---")
category_counts = df['category'].value_counts()
print(category_counts)
# value_counts() returns a Series with the frequency of each unique value
# This shows how many products are in each category

print("\n--- Category percentage distribution ---")
category_percentages = df['category'].value_counts(normalize=True) * 100
print(category_percentages.round(1))
# normalize=True returns proportions instead of counts
# Multiplying by 100 converts to percentages
# This shows the distribution of products across categories
````

*Working with Samples:*

````python
# 7. Working with samples
print("\n--- Random sample of 5 products ---")
print(df.sample(5))
# sample(n) returns n random rows
# This is useful for quickly inspecting a random subset of data

print("\n--- Random 25% sample ---")
print(df.sample(frac=0.25))
# sample(frac=x) returns x fraction of the rows
# frac=0.25 returns 25% of the data
# This is useful for creating random subsets for testing
````
