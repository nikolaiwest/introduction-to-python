*Setup:*

````python
# 1. Setup a sample DataFrame
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Create employee data
data = {
    'employee_id': range(1001, 1016),
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Emma', 'Frank', 'Grace', 'Henry', 
             'Isabella', 'Jack', 'Kate', 'Liam', 'Mia', 'Noah', 'Olivia'],
    'department': np.random.choice(['HR', 'IT', 'Sales', 'Marketing'], 15),
    'salary': np.random.randint(50000, 100000, 15),
    'years_employed': np.random.randint(0, 10, 15)
}

# Create DataFrame
employees = pd.DataFrame(data)

# Set employee_id as the index
employees.set_index('employee_id', inplace=True)

print("DataFrame with employee_id as index:")
employees.head()
````

*Basic selection:*

````python
# 2. Use basic selection with brackets 

# Select just the 'name' column
names = employees['name']
print("1. Name column as Series:")
print(names.head())
print(f"Type: {type(names)}")

# Select the 'name' and 'department' columns together
name_dept = employees[['name', 'department']]
print("2. Name and department columns as DataFrame:")
print(name_dept.head())
print(f"Type: {type(name_dept)}")

# Select the first 5 rows using slicing
first_five = employees[:5]
print("3. First 5 rows:")
print(first_five)
````

*Label-Based Indexing:*


````python
# 3. Use the loc accessor for label-based indexing:

# Select the employee with ID 1005
employee_1005 = employees.loc[1005]
print("Employee with ID 1005:")
print(employee_1005)

# Select employees with IDs 1003 and 1010
selected_employees = employees.loc[[1003, 1010]]
print("Employees with IDs 1003 and 1010:")
print(selected_employees)

# Select the 'name' and 'salary' columns for employees with IDs from 1005 to 1008
name_salary_range = employees.loc[1005:1008, ['name', 'salary']]
print("Name and salary for employees with IDs 1005-1008:")
print(name_salary_range)

# Select all columns for employees with IDs 1001, 1005, and 1010
specific_employees = employees.loc[[1001, 1005, 1010]]
print("All columns for employees with IDs 1001, 1005, and 1010:")
print(specific_employees)
````

*Position-based indexing:*

````python
# 4. Getting summary information

# Select the 3rd employee (remember Python is 0-indexed)
third_employee = employees.iloc[2]
print("3rd employee (index position 2):")
print(third_employee)

# Select rows 2 through 5 (exclusive of 6)
rows_2_to_5 = employees.iloc[2:6]
print("Rows 2 through 5 (exclusive of 6):")
print(rows_2_to_5)

# Select the first 3 rows and the first 2 columns
first_3_rows_2_cols = employees.iloc[:3, :2]
print("First 3 rows and first 2 columns:")
print(first_3_rows_2_cols)

# Select the 'salary' column (by position) for all rows
# First, find the position of the 'salary' column
salary_position = employees.columns.get_loc('salary')
salary_column = employees.iloc[:, salary_position]
print(f"Salary column (position {salary_position}) for all rows:")
print(salary_column.head())
````

*Boolean indexing:*

````python
# 5. Basic column exploration

# Select employees in the 'IT' department
it_employees = employees[employees['department'] == 'IT']
print("Employees in the IT department:")
print(it_employees)

# Select employees with a salary greater than 75000
high_salary = employees[employees['salary'] > 75000]
print("Employees with salary > 75000:")
print(high_salary)

# Select employees who have been employed for more than 5 years
senior_employees = employees[employees['years_employed'] > 5]
print("Employees with more than 5 years employment:")
print(senior_employees)
````

*Exploring Categorical Data:*

````python
# 6. Exploring categorical data

# Select employees in either IT or HR departments
it_or_hr = employees[(employees['department'] == 'IT') | (employees['department'] == 'HR')]
print("Employees in either IT or HR departments:")
print(it_or_hr)

# Alternative approach using isin()
it_or_hr_alt = employees[employees['department'].isin(['IT', 'HR'])]
print("Employees in IT or HR using isin():")
print(it_or_hr_alt)

# Select employees in Marketing with a salary less than 70000
marketing_low_salary = employees[(employees['department'] == 'Marketing') & (employees['salary'] < 70000)]
print("Marketing employees with salary < 70000:")
print(marketing_low_salary)
````
