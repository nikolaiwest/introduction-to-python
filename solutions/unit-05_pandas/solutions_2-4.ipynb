*Data Import:*

```python
import pandas as pd
import numpy as np

# 1. Import the messy customer data
# Import the CSV file
messy_df = pd.read_csv('data/customers_messy.csv')

# Display the first few rows
print("First few rows of the messy data:")
print(messy_df.head())

# Check dimensions
print(f"DataFrame dimensions: {messy_df.shape}")
# This shows the number of rows and columns in the DataFrame
```

*Handle missing values:*

```python
# 2. Identify missing values

# Find missing values
missing_mask = messy_df.isna()
print("\nMissing value mask (first 5 rows):")
print(missing_mask.head())
# isna() creates a DataFrame with True for missing values and False for non-missing values

# Count missing values per column
missing_counts = messy_df.isna().sum()
print("\nMissing values per column:")
print(missing_counts)
# sum() counts the True values (missing data) in each column

# Calculate percentage of missing values
missing_percentages = messy_df.isna().mean() * 100
print("\nPercentage of missing values per column:")
print(missing_percentages.round(2))
# mean() calculates the proportion of True values, which we multiply by 100 for percentages

# Check for empty strings in customer_id
empty_ids = (messy_df['customer_id'] == '').sum()
print(f"\nEmpty customer_id values: {empty_ids}")
# This checks specifically for empty strings, which aren't technically NULL values
```

```python
# 3. Handle missing values

# Create a copy for cleaning
clean_df = messy_df.copy()

# Remove rows with missing or empty customer_id
# First, create a mask for valid customer IDs
valid_id_mask = (clean_df['customer_id'].notna()) & (clean_df['customer_id'] != '')
clean_df = clean_df[valid_id_mask]
print(f"\nRows after removing missing/empty customer_ids: {clean_df.shape[0]}")
# We use both notna() and checking for empty strings to handle all invalid IDs

# Fill missing name values with "Unknown"
clean_df['name'] = clean_df['name'].fillna("Unknown")
# fillna() replaces NULL values with the specified value

# Convert age to numeric and handle missing values
clean_df['age'] = pd.to_numeric(clean_df['age'], errors='coerce')
# errors='coerce' converts invalid values to NaN

# Calculate the median age and fill missing values
median_age = clean_df['age'].median()
clean_df['age'] = clean_df['age'].fillna(median_age)
print(f"\nMedian age used for filling: {median_age}")
# Using median is often better than mean when data might have outliers

# Create a standard replacement for missing emails
clean_df['email'] = clean_df['email'].fillna("missing@example.com")
```

*Detecting and Removing Duplicates:*

```python
# 4. Work with duplicate records

# Check for duplicates
duplicate_mask = clean_df.duplicated()
duplicate_count = duplicate_mask.sum()
print(f"\nNumber of duplicate rows: {duplicate_count}")
# duplicated() returns True for rows that are duplicates of earlier rows

# View the duplicate rows
if duplicate_count > 0:
    print("\nDuplicate rows (and their first occurrences):")
    print(clean_df[clean_df.duplicated(keep=False)])
    # keep=False marks all duplicates, including first occurrences

# Remove duplicates
clean_df = clean_df.drop_duplicates()
print(f"\nRows after removing duplicates: {clean_df.shape[0]}")
# drop_duplicates() removes duplicate rows, keeping the first occurrence by default
```

*Data Type Conversion:*

```python
# 5. Fix data types

# Age was already converted to numeric in step 3

# Remove $ from purchase_amount and convert to float
clean_df['purchase_amount'] = clean_df['purchase_amount'].str.replace('$', '', regex=False)
clean_df['purchase_amount'] = pd.to_numeric(clean_df['purchase_amount'])
# str.replace() removes the dollar sign, and pd.to_numeric() converts to a floating-point number

# Verify data types
print("\nData types after conversion:")
print(clean_df.dtypes)
# This confirms that our type conversions worked correctly
```

*Handling Outliers:*

```python
# 6. Deal with outliers
# Identify age outliers
age_outliers = (clean_df['age'] < 0) | (clean_df['age'] > 100)
outlier_count = age_outliers.sum()
print(f"\nNumber of age outliers: {outlier_count}")
# This identifies impossible or implausible age values

# Replace outliers with the median
if outlier_count > 0:
    clean_df.loc[age_outliers, 'age'] = median_age
    print(f"Replaced {outlier_count} outliers with median age: {median_age}")

# Create age group categories
bins = [0, 18, 65, float('inf')]
labels = ['Child', 'Adult', 'Senior']
clean_df['age_group'] = pd.cut(clean_df['age'], bins=bins, labels=labels)

print("\nAge groups distribution:")
print(clean_df['age_group'].value_counts())
# pd.cut() divides the age values into the specified bins and assigns the corresponding labels
```
