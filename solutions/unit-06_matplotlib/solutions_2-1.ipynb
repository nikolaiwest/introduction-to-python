*Basic line plotting:*

````python
# 1-2. Quadratic function for basic line plotting

# 1. Create a line plot of the function y = x² for x values from -5 to 5
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)
y = x**2

plt.figure(figsize=(8, 6))
plt.plot(x, y)

# 2. Add appropriate title and axis labels
plt.title('Quadratic Function: y = x²')
plt.xlabel('x values')
plt.ylabel('y values')

plt.show()
````


*Multiple line plots:*



````python
# 3-5. Creating multiple lines for trigonometrical functions

# 3. Plot both y = sin(x) and y = cos(x) for x values from 0 to 2π
x = np.linspace(0, 2*np.pi, 100)
sin_y = np.sin(x)
cos_y = np.cos(x)

plt.figure(figsize=(10, 6))

# 4. Use a blue solid line for sin(x) and a red dashed line for cos(x)
plt.plot(x, sin_y, color='blue', linestyle='-', label='sin(x)')
plt.plot(x, cos_y, color='red', linestyle='--', label='cos(x)')

# 5. Add a legend
plt.legend()

plt.title('Sine and Cosine Functions')
plt.xlabel('x values')
plt.ylabel('y values')

plt.show()
````


*Customization:*



````python
# 6-9. Customization with temperature data

# 6. Create a new figure and plot monthly temperature data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temperatures = [5, 7, 10, 15, 20, 25, 28, 27, 23, 17, 10, 6]

plt.figure(figsize=(10, 6))

# 7. Add circular markers connected by a solid green line
plt.plot(months, temperatures, marker='o', color='green', linestyle='-')

# 8. Add a grid with dotted lines
plt.grid(True, linestyle=':')

# 9. Set the y-axis limits to better visualize the temperature range
plt.ylim(0, 30)
# Optionally, update the chart labeling 
plt.title('Monthly Average Temperatures')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')

plt.show()
````

*Customization for multiple line plots:*

````python
# 10-12. Multiple line plots with customization

# 10. Create a new figure for comparing growth rates
years = [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020]
company_a = [10, 15, 25, 40, 55, 80, 110, 140, 170, 200, 220]
company_b = [8, 13, 20, 35, 60, 90, 105, 150, 190, 210, 250]

plt.figure(figsize=(12, 6))

# 11. Plot both datasets with different styles
plt.plot(years, company_a, marker='^', color='blue', linestyle='-', 
         linewidth=2, markersize=8, label='Company A')
plt.plot(years, company_b, marker='s', color='red', linestyle='--', 
         linewidth=2, markersize=8, label='Company B')

# 12. Add customization elements
plt.grid(True, alpha=0.3)
plt.legend(loc='upper left')
plt.title('Growth Comparison of Two Companies (2010-2020)')
plt.xlabel('Year')
plt.ylabel('Revenue (Million $)')
plt.xticks(years, rotation=45)

plt.show()
````
