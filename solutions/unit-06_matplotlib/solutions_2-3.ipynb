```python
import numpy as np
import matplotlib.pyplot as plt
```

*Basic scatter plots:*

```python
# 1-2. Basic scatter plots

# 1. Generate two arrays of 50 random numbers between 0 and 100
np.random.seed(42)  # For reproducibility
x1 = np.random.rand(50) * 100
y1 = np.random.rand(50) * 100

# Create basic scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(x1, y1)

# 2. Add title, axis labels, and grid
plt.title('Basic Scatter Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```


*Customizing markers:*



```python
# 3-4. Customizing markers

# 3. Create new scatter plot with triangular markers of size 80
plt.figure(figsize=(10, 6))
plt.scatter(x1, y1, marker='^', s=80, 
           # 4. Make markers semi-transparent with blue fill and red edges
           alpha=0.5, color='blue', edgecolor='red')
plt.title('Customized Scatter Plot with Triangular Markers')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```


*Representing additional dimensions:*



```python
# 5-7. Representing additional dimensions

# 5. Generate a third array for color dimension
z1 = np.random.rand(50) * 100

# Generate a fourth array for size dimension
# 7. Use this for marker sizes - scale up for better visibility
sizes = np.random.rand(50) * 500 + 100  # Values between 100 and 600

plt.figure(figsize=(10, 8))
# Create scatter plot with color and size dimensions
scatter = plt.scatter(x1, y1, c=z1, s=sizes, 
                     cmap='viridis', alpha=0.7)

# 6. Add a colorbar
plt.colorbar(scatter, label='Z Values')

plt.title('Scatter Plot with Color and Size Dimensions')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```


*Analyzing relationships:*



```python
# 8-10. Analyzing relationships

# 8. Create arrays where y = x² + random noise
x2 = np.linspace(0, 10, 50)
y2 = x2**2 + np.random.normal(0, 5, 50)

# 9. Create scatter plot and add quadratic trend line
plt.figure(figsize=(10, 6))
plt.scatter(x2, y2, alpha=0.7)

# Find coefficients for a 2nd-degree polynomial (quadratic)
coeffs = np.polyfit(x2, y2, 2)
# Create polynomial function
p = np.poly1d(coeffs)

# Create smooth x values for the trend line
x_line = np.linspace(0, 10, 100)
plt.plot(x_line, p(x_line), 'r-', linewidth=2)

# 10. Add text annotation explaining the relationship
# Extract coefficients (a, b, c) from the quadratic equation y = ax² + bx + c
a, b, c = coeffs
equation = f'y = {a:.2f}x² + {b:.2f}x + {c:.2f}'

# Place the text in an appropriate location
plt.text(0.5, 80, equation, fontsize=12, 
         bbox=dict(facecolor='white', alpha=0.5))

plt.title('Quadratic Relationship with Trend Line')
plt.xlabel('X Variable')
plt.ylabel('Y Variable')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
